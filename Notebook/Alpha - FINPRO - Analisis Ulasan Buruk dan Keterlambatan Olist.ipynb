## Ringkasan Eksekutif

**Permasalahan**

Sekitar **1 dari 7 pesanan Olist (14,61%)** berakhir dengan ulasan buruk (bintang 1–2). Project ini mencari faktor operasional di balik ulasan buruk tersebut. Pertanyaan yang ingin dijawab: *seller dan rute pengiriman mana yang perlu diprioritaskan agar proporsi ulasan buruk dapat ditekan?* Analisis menggunakan ±99 ribu pesanan Olist pada periode Januari 2017 – Agustus 2018.

**Temuan Utama**

1. **Keterlambatan adalah faktor operasional yang paling berkaitan dengan ulasan buruk.** Sebanyak **62,4%** pesanan telat mendapat ulasan buruk, ±6,8 kali lipat dibanding pesanan tepat waktu (9,2%). Meskipun pesanan telat hanya ±6,5% dari seluruh pesanan, kelompok ini menyumbang **±28% dari seluruh ulasan buruk**.
2. **Keterlambatan terkonsentrasi pada rute tertentu.** **11 rute prioritas** (±23% pesanan) menyumbang **±40% pesanan telat**. Rute **SP → RJ** memberi dampak terbesar (14,4% pesanan telat), padahal jaraknya hanya ±383 km. Pengiriman berjarak lebih dari 1.500 km dan pengiriman ke region **Nordeste** juga paling sering telat (11,7% dan 12,8%).
3. **Pesanan telat paling lama tertahan di tahap transit (±26 vs ±7 hari). Selain itu, 27,8% pesanan telat juga telat diserahkan oleh seller.** Pesanan yang telat diserahkan seller ke kurir ±3,9 kali lebih sering telat sampai ke pelanggan. Masalah ini terkonsentrasi pada **121 seller prioritas** yang menyumbang ±40% pesanan telat serah.
4. **Keterlambatan melonjak serentak di semua region** pada November 2017 dan Februari–Maret 2018, dengan tingkat keterlambatan bulanan mencapai 19,0%.
5. **Di luar keterlambatan**, pesanan yang tidak sampai dan pesanan kompleks (multi-seller atau berisi banyak barang) juga sering mendapat ulasan buruk. Ongkos kirim dan nilai pesanan hanya berkaitan secara tidak langsung. Metode pembayaran **tidak terbukti berkaitan** dengan ulasan buruk (p = 0,057 ≥ 0,05, gagal menolak Ho). Isi ulasan menunjukkan bahwa ulasan buruk pada pesanan yang **tepat waktu** umumnya mengeluhkan **barang yang kurang/salah dan kualitas produk**, bukan pengiriman.

**Rekomendasi Utama**

1. Evaluasi 11 rute prioritas dan perbaiki estimasi tanggal sampai.
2. Siapkan kapasitas tambahan menjelang bulan-bulan sibuk.
3. Tangani pesanan yang tidak sampai, pesanan multi-paket, serta kualitas dan kesesuaian produk pada seller dengan keluhan produk berulang.

**Potensi Dampak (Skenario)**

Perbaikan pada rute dan seller prioritas berpotensi mengurangi **±22% pesanan telat** (Late Delivery Rate turun dari 6,79% menjadi ±5,3%) dan mencegah **±750 ulasan buruk** selama periode analisis (±450 per tahun). Dalam metrik utama, **Low Review Rate turun dari 14,61% menjadi ±13,84%** (−0,8 poin persen, atau ±5% ulasan buruk lebih sedikit). Sebagai batas atas, jika **seluruh** keterlambatan hilang pun, Low Review Rate hanya turun ke **±11,16%** (±24% ulasan buruk berkurang), sehingga sumber ulasan buruk di luar keterlambatan tetap perlu ditangani.

# 1. Business Understanding

## 1.1 Latar Belakang

Olist adalah *marketplace* di Brasil yang menghubungkan pelanggan dengan banyak seller dari berbagai state. Satu pesanan melewati beberapa tahap: pelanggan *checkout*, pembayaran disetujui, seller menyerahkan barang ke kurir, lalu kurir mengantar barang ke pelanggan. Saat *checkout*, pelanggan sudah dijanjikan tanggal estimasi barang sampai.

Jika salah satu tahap terlambat, pelanggan menunggu lebih lama dari yang dijanjikan. Setelah barang sampai, atau setelah tanggal estimasi lewat, Olist mengirim survei kepuasan dan pelanggan memberi skor 1–5. Skor 1–2 menandakan pelanggan kecewa.

Masalahnya, skor ulasan hanya menunjukkan **bahwa** pelanggan kecewa, tetapi tidak menjelaskan **mengapa**. Karena itu, analisis ini menelusuri faktor operasional di balik ulasan buruk, mulai dari performa pengiriman, rute dan jarak pengiriman, hingga kinerja seller.

> **Kerangka berpikir:** ulasan buruk diperlakukan sebagai **hasil akhir** yang ingin dijelaskan. Keterlambatan pengiriman, kondisi geografis, dan kinerja seller diperlakukan sebagai faktor operasional yang **diuji dan ditelusuri**, bukan diasumsikan sebagai penyebab sejak awal.

## 1.2 Stakeholder

| Stakeholder | Peran | Cara memakai hasil analisis |
|---|---|---|
| Tim Operasional & Logistik Olist | Stakeholder utama | Memantau keterlambatan dan menentukan rute pengiriman yang perlu diprioritaskan |
| Tim Seller Management Olist | Stakeholder pendukung | Menentukan seller yang perlu dipantau atau dibina, serta mengevaluasi kepatuhan seller terhadap batas waktu serah barang |
| Tim Customer Experience Olist | Stakeholder pendukung | Mengomunikasikan keterlambatan kepada pelanggan, menangani keluhan, dan mengatur pengiriman survei kepuasan |
| Pelanggan | Penerima dampak | Mendapat pengiriman yang lebih sesuai janji |

## 1.3 Problem Statement

**Pertanyaan utama:**

> **Seller dan rute pengiriman mana yang perlu diprioritaskan perbaikannya agar proporsi ulasan buruk di Olist dapat ditekan?**

Pertanyaan utama dijabarkan menjadi tiga pilar:

| Pilar | Pertanyaan | Yang dianalisis | Output |
|---|---|---|---|
| **1 — WHAT** | Faktor operasional apa saja yang paling berkaitan dengan ulasan buruk di Olist? | Status pesanan (sampai vs tidak sampai), ketepatan waktu (telat vs tepat waktu), ongkir, nilai pesanan, jumlah barang, jumlah seller per pesanan, metode pembayaran utama, serta **isi komentar** ulasan buruk (keluhan utama dan pihak yang disalahkan) | Faktor yang paling berkaitan dengan ulasan buruk, serta faktor yang terbukti **tidak** berkaitan |
| **2 — WHERE & WHEN** | Di rute, jarak, dan bulan apa keterlambatan pengiriman paling sering terjadi? | Rute state seller → state pelanggan, pengiriman sama-state vs beda-state, jarak seller–pelanggan, bulan pembelian | Daftar rute, jarak, dan bulan rawan keterlambatan |
| **3 — WHY & WHO** | Seberapa sering keterlambatan sudah terjadi di sisi seller, dan seller mana yang paling sering telat? | Durasi per tahap (proses seller vs transit kurir), tanggal serah ke kurir dibandingkan batas serah (`shipping_limit_date`), per seller | Daftar seller prioritas pembinaan |

Alurnya bertahap: Pilar 1 membuktikan dulu masalah mana yang paling berkaitan dengan ulasan buruk, lalu Pilar 2 dan Pilar 3 menelusuri sumber keterlambatan dari sisi rute dan sisi seller.

**Pemetaan 5W-1H:**

| | Pertanyaan yang dijawab |
|---|---|
| **WHAT** | Faktor apa yang paling berkaitan dengan ulasan buruk? |
| **WHERE** | Rute dan wilayah mana yang memiliki risiko keterlambatan tinggi? |
| **WHEN** | Pada periode kapan keterlambatan meningkat? |
| **WHY** | Apakah hambatan terbesar ada di sisi seller (serah barang) atau di tahap transit? |
| **WHO** | Seller mana yang perlu diprioritaskan? |
| **HOW** | Tindakan apa yang dapat difokuskan pada seller dan rute berisiko? |

## 1.4 Goals

Tujuan project adalah menghasilkan **daftar prioritas rute dan seller yang dapat ditindaklanjuti** untuk menekan ulasan buruk. Tujuan ini dievaluasi melalui lima output:

1. Mengukur Low Review Rate saat ini dan membandingkannya antar kelompok pesanan (misalnya telat vs tepat waktu, sampai vs tidak sampai).
2. Menentukan rute, jarak, dan bulan dengan Late Delivery Rate tertinggi.
3. Mengukur seberapa besar keterlambatan pengiriman yang diawali keterlambatan serah barang dari seller.
4. Menyusun daftar prioritas rute dan seller berdasarkan kombinasi tingkat keterlambatan dan jumlah pesanan.
5. Memberikan **estimasi dampak berbasis skenario**, misalnya: jika Late Delivery Rate pada rute prioritas turun ke rata-rata nasional, berapa perkiraan penurunan Low Review Rate.

**Catatan evaluasi:** keberhasilan project tidak didefinisikan sebagai penurunan X% yang belum dapat dibuktikan oleh data historis, melainkan sebagai kemampuan menghasilkan prioritas tindakan yang **terukur, transparan, dan dapat dijelaskan**. Estimasi pada poin 5 bersifat skenario, karena dampak sebenarnya baru dapat diukur setelah rekomendasi diterapkan.

## 1.5 Business Metrics

Analisis menggunakan tiga metrik utama yang saling berhubungan. Ketiganya dihitung pada **tabel level pesanan** (1 baris = 1 pesanan), lihat bagian 1.6.

**1. Low Review Rate (metrik hasil)**

> Low Review Rate = jumlah pesanan dengan skor ulasan 1–2 ÷ jumlah pesanan yang memiliki ulasan valid × 100%

- Fungsi: ukuran utama ketidakpuasan pelanggan; metrik yang ingin diturunkan.
- Risiko bisnis: ulasan buruk menurunkan kepercayaan calon pembeli terhadap seller dan platform, serta membuat pelanggan enggan berbelanja kembali.

**2. Late Delivery Rate (metrik operasional utama)**

> Late Delivery Rate = jumlah pesanan yang sampai setelah tanggal estimasi ÷ jumlah pesanan berstatus `delivered` yang memiliki tanggal sampai × 100%

- Fungsi: mengukur seberapa sering janji tanggal sampai kepada pelanggan dilanggar. Dipakai di Pilar 1 dan Pilar 2.
- Risiko bisnis: janji yang dilanggar langsung dirasakan pelanggan dan menambah beban keluhan.

**3. Seller Late Handover Rate (metrik seller)**

> Seller Late Handover Rate = jumlah pesanan yang diserahkan ke kurir setelah `shipping_limit_date` ÷ jumlah pesanan 1 seller yang memiliki tanggal serah ke kurir × 100%

- Fungsi: mengukur seberapa sering keterlambatan sudah terjadi sebelum barang diterima kurir. Dipakai di Pilar 3.
- Risiko bisnis: keterlambatan di sisi seller sebenarnya dapat dicegah oleh Olist melalui aturan dan pembinaan seller.

**Hierarki metrik:**

```
CUSTOMER DISSATISFACTION      ← masalah bisnis
          │
LOW REVIEW RATE               ← metrik hasil yang ingin diturunkan
          │ berkaitan dengan (Pilar 1)
LATE DELIVERY RATE            ← masalah operasional utama
          │ bersumber dari
    ┌─────┴──────────────┐
SELLER (Pilar 3)      LOGISTIK & RUTE (Pilar 2)
Seller Late           Durasi transit, rute,
Handover Rate         jarak, bulan
```

**Metrik pendukung:**

| Metrik | Kolom yang dipakai | Fungsi |
|---|---|---|
| Rata-rata / median skor ulasan | `review_score` | Membandingkan kepuasan antar kelompok |
| Lama pengiriman total (hari) | `order_purchase_timestamp` → `order_delivered_customer_date` | Durasi dari checkout sampai barang diterima |
| Waktu proses seller (hari) | `order_approved_at` → `order_delivered_carrier_date` | Durasi di sisi seller |
| Waktu transit kurir (hari) | `order_delivered_carrier_date` → `order_delivered_customer_date` | Durasi di sisi kurir |
| Selisih terhadap estimasi (hari) | `order_delivered_customer_date` − `order_estimated_delivery_date` | Positif = terlambat, negatif = lebih cepat dari janji |
| Jarak seller–pelanggan (km) | koordinat seller & pelanggan | Gambaran kasar jauh-dekatnya pengiriman |
| Multi-seller order rate | jumlah seller per pesanan | Mengamati kompleksitas pemenuhan pesanan |
| Jumlah pesanan | `order_id` (unik) | Syarat minimum sebelum seller atau rute diberi peringkat |

## 1.6 Unit Analisis

Data hasil penggabungan 9 tabel memiliki tingkat detail **1 baris = 1 item dalam 1 pesanan** (pasangan `order_id` + `order_item_id`). Karena itu, analisis menggunakan dua tingkat tabel:

| Tabel | 1 baris = | Dibuat pada | Dipakai untuk |
|---|---|---|---|
| `df_clean` | 1 item dalam 1 pesanan | Tahap 3 (Data Cleaning) | Pengecekan kualitas data per item (missing value, konsistensi, outlier) dan sumber lokasi seller yang sudah dibersihkan |
| `df_order` | 1 pesanan (`order_id`) | Tahap 4.1 (Feature Engineering) | Menghitung ketiga metrik utama; unit analisis utama |

Pada tabel level pesanan, kolom yang nilainya sama untuk setiap item (status, tanggal, state pelanggan, skor ulasan, total pembayaran) diambil satu kali per pesanan. Informasi item diringkas menjadi jumlah barang, total harga, total ongkir, dan jumlah seller.

Untuk **Pilar 3**, analisis seller hanya memakai **pesanan dengan 1 seller** (±98,7% dari pesanan yang memiliki item), karena tanggal serah ke kurir hanya dicatat satu kali per pesanan. Pada pesanan multi-seller, tidak dapat diketahui seller mana yang terlambat menyerahkan barang.

## 1.7 Ruang Lingkup dan Sumber Data

- **Sumber data:** [Brazilian E-Commerce Public Dataset by Olist](https://www.kaggle.com/datasets/olistbr/brazilian-ecommerce) (Kaggle), data transaksi nyata Olist yang telah dianonimkan. Terdiri dari **9 tabel** (orders, customers, order_items, order_payments, order_reviews, products, sellers, geolocation, dan product_category_name_translation) dengan **±99 ribu pesanan** pada periode **September 2016 – Oktober 2018**.
- **Periode analisis:** pesanan dengan tanggal pembelian **Januari 2017 – Agustus 2018** (20 bulan analisis; data akhir Agustus 2018 bersifat parsial, lihat 3.9). Bulan di luar periode ini berisi sangat sedikit pesanan (beberapa bulan di bawah 20 pesanan) dan November 2016 tidak memiliki data sama sekali, sehingga tidak mewakili kondisi normal (lihat langkah 3.9).
- **Analisis wilayah:** memakai pasangan state seller → state pelanggan. Kolom koordinat (`customer_lat`, `customer_lng`, `seller_lat`, `seller_lng`) dipakai untuk menghitung jarak seller–pelanggan dan untuk visualisasi peta di dashboard.
- **Label isi ulasan:** komentar ulasan buruk (bahasa Portugis) diberi label masalah utama, posisi pesanan, dan pihak yang disalahkan oleh **Claude (Anthropic)** di luar notebook. Metode dan keterbatasannya dijelaskan di 5.1.8.
- **Fokus:** ulasan → pengiriman → rute → seller. Project tidak membahas marketing, profitabilitas, stok barang, maupun kinerja perusahaan kurir tertentu, karena datanya tidak tersedia.

## 1.8 Asumsi

- Skor ulasan 1–2 dianggap ulasan buruk. Skor 3 dianggap netral.
- Pesanan dianggap **telat** jika tanggal `order_delivered_customer_date` melewati tanggal `order_estimated_delivery_date`. Perbandingan dilakukan **per tanggal** (jam diabaikan), karena tanggal estimasi tidak memiliki keterangan jam.
- Seller dianggap **telat serah** jika `order_delivered_carrier_date` melewati `shipping_limit_date`.
- Untuk pesanan yang memiliki lebih dari 1 ulasan, dipakai ulasan yang dijawab paling akhir (`review_answer_timestamp`).
- **Kode pos** dianggap sebagai acuan lokasi yang paling dapat dipercaya. State, kota, dan koordinat yang tidak sesuai dengan kode pos dikoreksi mengikuti kode pos (langkah 3.7).
- Label isi komentar dari Claude dianggap cukup mewakili keluhan utama pelanggan, dengan akurasi yang sudah dicek melalui QA manual pada 150 ulasan (5.1.8).
- Jarak seller–pelanggan (dihitung dari titik tengah kode pos) dipakai sebagai gambaran sederhana jauh-dekatnya pengiriman.
- Seller dan rute hanya diberi peringkat jika memiliki jumlah pesanan minimum tertentu (nilainya ditentukan saat analisis), agar seller atau rute dengan sedikit pesanan tidak mendominasi peringkat hanya karena persentasenya ekstrem.

## 1.9 Batasan Project

| Batasan | Konsekuensi | Cara menyampaikan hasil |
|---|---|---|
| Ulasan diberikan per pesanan, bukan per barang atau per seller | Pada pesanan berisi beberapa barang/seller, tidak dapat diketahui barang atau seller mana yang membuat pelanggan kecewa | Analisis seller dibatasi pada pesanan 1 seller |
| Tanggal serah ke kurir dicatat per pesanan | Keterlambatan serah tidak dapat dikaitkan ke seller tertentu pada pesanan multi-seller | Pilar 3 hanya memakai pesanan 1 seller |
| Survei pada pesanan telat sering dikirim sebelum barang sampai (survei dikirim setelah tanggal estimasi lewat) | Skor pada pesanan telat lebih mencerminkan keterlambatan daripada kualitas produk | Dinyatakan saat menginterpretasi hubungan telat vs ulasan |
| Tidak ada data identitas kurir | Tidak dapat menilai perusahaan kurir tertentu | "Terdapat hambatan pada tahap transit setelah barang diserahkan ke kurir" |
| Tidak ada data penyebab internal seller (stok, kapasitas, proses packing) | Penyebab seller terlambat tidak diketahui | "Seller terlambat menyerahkan barang", bukan penyebab spesifiknya |
| Tidak ada harga pembanding pasar | Tidak dapat menyatakan produk terlalu mahal/murah | Hanya melihat apakah harga atau ongkir berkaitan dengan skor ulasan |
| Data historis (bukan eksperimen) | Hubungan yang ditemukan adalah keterkaitan, bukan sebab-akibat | "Pesanan telat lebih sering mendapat ulasan buruk", bukan "keterlambatan pasti menyebabkan ulasan buruk" |
| Komentar ulasan bersifat opsional dan berbahasa Portugis | Isi keluhan hanya diketahui dari ±75% ulasan buruk yang berkomentar, dan labelnya dibuat oleh model bahasa (Claude) | Disampaikan sebagai **persepsi pelanggan**; akurasi label sudah dicek manual pada 150 ulasan (Pain Points & Review Stage ±94%, Blame Target ±79%) |
| Kode pos hanya berisi 5 digit awal | Satu kode pos dapat mencakup beberapa kota yang bertetangga | Rute dianalisis per state; nama kota dibaca sebagai wilayah kota |
| Periode data 2017–2018 | Hasil belum tentu mewakili kondisi Olist saat ini | Selalu menyebut periode data saat menyampaikan hasil |

# 2. Data Merging

Dataset Olist terdiri dari 9 tabel dengan `olist_orders_dataset.csv` sebagai tabel pusat. Tabel-tabel ini digabung satu per satu ke tabel orders. Pada setiap langkah dilakukan merge, lalu pengecekan hasil (jumlah baris & kolom).

| Langkah | Tabel | 1 baris = | Digabung lewat | Cara penggabungan | Kolom penting | Dipakai untuk |
|---|---|---|---|---|---|---|
| — | `orders` | 1 pesanan | — | Tabel pusat | Status, tanggal pembelian → approval → serah ke kurir → diterima, estimasi | Status pesanan & Late Delivery Rate (Pilar 1), bulan (Pilar 2), durasi tahap (Pilar 3) |
| 1 | `customers` | 1 `customer_id` | `customer_id` | Left join langsung (1 pesanan = 1 `customer_id`) | Kode pos, kota, state pelanggan | Rute & region tujuan (Pilar 2) |
| 2 | `order_items` | 1 barang dalam pesanan | `order_id` | Left join; baris bertambah karena 1 pesanan bisa berisi beberapa barang | Seller, harga, ongkir, `shipping_limit_date` | Jumlah barang & seller, nilai pesanan, ongkir (Pilar 1); Seller Late Handover Rate (Pilar 3) |
| 3 | `products` | 1 produk | `product_id` | Left join langsung | Kategori (Portugis), berat & dimensi | Penjelasan outlier ongkir lewat berat (3.8); kategori produk (5.1.8) |
| 4 | `category_translation` | 1 kategori | `product_category_name` | Left join langsung | Nama kategori bahasa Inggris | Label kategori produk (5.1.8) |
| 5 | `sellers` | 1 seller | `seller_id` | Left join langsung | Kode pos, kota, state seller | Rute asal (Pilar 2), seller prioritas (Pilar 3) |
| 6 | `order_reviews` | 1 ulasan | `order_id` | Left join; sebagian kecil pesanan punya > 1 ulasan sehingga barisnya tergandakan (ditangani di 3.3) | Skor & komentar ulasan | **Low Review Rate** (hasil akhir yang diukur di semua pilar), isi keluhan (5.1.8) |
| 7 | `order_payments` | 1 transaksi pembayaran | `order_id` | **Diringkas dulu menjadi 1 baris per pesanan**, baru di-left join, agar baris tidak tergandakan | Metode bayar utama, total bayar, cicilan | Metode pembayaran (Pilar 1) |
| 8 | `geolocation` | 1 titik koordinat | Kode pos | **Diringkas dulu menjadi 1 titik per kode pos (median)**, lalu di-left join **2 kali**: ke kode pos pelanggan dan ke kode pos seller | Latitude & longitude | Jarak seller → pelanggan (Pilar 2) |

Semua penggabungan memakai **left join**, sehingga tidak ada pesanan yang hilang walaupun datanya di tabel lain tidak lengkap (misalnya pesanan tanpa ulasan). Kecocokan key antar tabel dicek di akhir bagian ini.

## Langkah 7 — + order_payments (diagregasi per pesanan)

**Order Payments** (`olist_order_payments_dataset.csv`) — 1 baris = 1 transaksi pembayaran.

| Feature | Description |
|---|---|
| order_id | ID pesanan (penghubung ke tabel orders) |
| payment_sequential | Urutan pembayaran jika pelanggan memakai lebih dari satu pembayaran |
| payment_type | Metode bayar: credit_card, boleto, voucher, debit_card, not_defined |
| payment_installments | Jumlah cicilan |
| payment_value | Nilai pembayaran (BRL) |

````python
df_payments = pd.read_csv(DATA_PATH + 'olist_order_payments_dataset.csv')

print('order_payments:', df_payments.shape)
df_payments.head()
````

````python
payments_per_order = df_payments['order_id'].value_counts()

print('Pesanan yang memiliki data pembayaran:', df_payments['order_id'].nunique())
print('Pesanan dengan >1 baris pembayaran   :', (payments_per_order > 1).sum())
print('Maksimal baris pembayaran per pesanan:', payments_per_order.max())
````

````python
# contoh pesanan dengan baris pembayaran terbanyak
df_payments[df_payments['order_id'] == payments_per_order.index[0]].sort_values('payment_sequential').head(10)
````

Satu pesanan bisa dibayar dengan beberapa transaksi, misalnya kartu kredit + voucher, atau banyak voucher sekaligus (maksimal 29 baris untuk satu pesanan).

### Kenapa tidak langsung di-merge?

`df_merged` saat ini sudah berlevel item (1 pesanan bisa punya banyak baris). Jika tabel pembayaran (yang juga bisa punya banyak baris per pesanan) langsung digabung, setiap item akan dipasangkan dengan setiap baris pembayaran pada pesanan yang sama, sehingga baris berlipat (item × pembayaran):

````python
rows_if_direct = df_merged.merge(df_payments, on='order_id', how='left').shape[0]

print('Baris df_merged saat ini          :', df_merged.shape[0])
print('Baris jika payments langsung digabung:', rows_if_direct)
print('Baris tambahan yang tidak semestinya :', rows_if_direct - df_merged.shape[0])
````

Selain menggandakan baris, merge langsung juga membuat nilai seperti `price` dan `payment_value` terhitung berkali-kali jika dijumlahkan.

**Solusi:** tabel pembayaran **diagregasi terlebih dahulu menjadi 1 baris per pesanan**, baru digabung. Kolom hasil agregasi:

| Kolom baru | Isi |
|---|---|
| payment_count | Jumlah baris/transaksi pembayaran pada pesanan |
| payment_types | Kombinasi metode bayar yang dipakai (misal `credit_card + voucher`) |
| payment_type_main | Metode bayar dengan nilai pembayaran terbesar pada pesanan |
| payment_installments_max | Jumlah cicilan terbanyak pada pesanan |
| payment_value_total | Total nilai pembayaran pesanan (BRL) |

````python
# metode bayar utama = metode dengan total nilai pembayaran terbesar dalam satu pesanan
payment_main = (
    df_payments.groupby(['order_id', 'payment_type'])['payment_value'].sum()
    .reset_index()
    .sort_values(['order_id', 'payment_value'], ascending=[True, False])
    .drop_duplicates('order_id')
    .set_index('order_id')['payment_type']
)

df_payments_agg = df_payments.groupby('order_id').agg(
    payment_count=('payment_sequential', 'size'),
    payment_types=('payment_type', lambda x: ' + '.join(sorted(x.unique()))),
    payment_installments_max=('payment_installments', 'max'),
    payment_value_total=('payment_value', 'sum'),
)
df_payments_agg['payment_type_main'] = payment_main
df_payments_agg = df_payments_agg.reset_index()

print('Hasil agregasi:', df_payments_agg.shape)
print('order_id unik setelah agregasi:', df_payments_agg['order_id'].is_unique)
df_payments_agg.head()
````

````python
df_payments_agg['payment_types'].value_counts()
````

### Merge

- `how='left'`: pesanan tanpa data pembayaran tetap dipertahankan.

````python
rows_before = df_merged.shape[0]

df_merged = df_merged.merge(df_payments_agg, on='order_id', how='left')

print('Sebelum merge:', rows_before, 'baris')
print('Hasil merge  :', df_merged.shape)
print('Pesanan tanpa data pembayaran:', df_merged.loc[df_merged['payment_count'].isna(), 'order_id'].nunique())
````

**Hasil:** jumlah baris **tetap 114.092** (tidak ada penggandaan) dan jumlah kolom menjadi **41** (36 + 5 kolom pembayaran hasil agregasi).

Ada **1 pesanan yang tidak memiliki data pembayaran** sama sekali. Hal ini akan dicek pada tahap Data Cleaning.

**Catatan:** nilai `payment_value_total` adalah total **per pesanan**, sehingga pada pesanan dengan beberapa item nilainya akan terulang di setiap baris item. Untuk menjumlahkan pembayaran, gunakan data level pesanan (`drop_duplicates('order_id')`), bukan langsung menjumlahkan seluruh baris.

## Langkah 8 — + geolocation (diagregasi per kode pos)

**Geolocation** (`olist_geolocation_dataset.csv`) — 1 baris = 1 titik koordinat untuk sebuah kode pos.

| Feature | Description |
|---|---|
| geolocation_zip_code_prefix | 5 digit awal kode pos (penghubung ke customers dan sellers) |
| geolocation_lat | Latitude (garis lintang) |
| geolocation_lng | Longitude (garis bujur) |
| geolocation_city | Nama kota |
| geolocation_state | State |

Tabel ini tidak memiliki key langsung ke orders. Tabel ini terhubung melalui **kode pos**, baik kode pos pelanggan (`customer_zip_code_prefix`) maupun kode pos seller (`seller_zip_code_prefix`). Koordinat ini berguna untuk menghitung jarak pengiriman seller → pelanggan dan untuk visualisasi peta.

````python
df_geolocation = pd.read_csv(DATA_PATH + 'olist_geolocation_dataset.csv')

print('geolocation:', df_geolocation.shape)
df_geolocation.head()
````

````python
points_per_zip = df_geolocation['geolocation_zip_code_prefix'].value_counts()

print('Kode pos unik di geolocation :', points_per_zip.shape[0])
print('Rata-rata titik per kode pos :', round(points_per_zip.mean(), 1))
print('Maksimal titik per kode pos  :', points_per_zip.max())
print('Baris duplikat penuh         :', df_geolocation.duplicated().sum())
````

Satu kode pos memiliki banyak titik koordinat (rata-rata ±53, maksimal 1.146), karena satu kode pos prefix mencakup satu area, bukan satu alamat. Di sisi `df_merged`, satu kode pos juga dipakai oleh banyak pelanggan/seller. Jika digabung langsung, setiap baris akan tergandakan sebanyak jumlah titik pada kode posnya.

**Solusi:** koordinat **diagregasi menjadi 1 titik per kode pos** menggunakan **median** latitude dan longitude. Median dipilih (bukan mean) karena lebih tahan terhadap titik koordinat yang salah/ekstrem.

````python
df_geo_agg = (
    df_geolocation.groupby('geolocation_zip_code_prefix')[['geolocation_lat', 'geolocation_lng']]
    .median()
    .reset_index()
)

print('Hasil agregasi:', df_geo_agg.shape)
print('Kode pos unik setelah agregasi:', df_geo_agg['geolocation_zip_code_prefix'].is_unique)
df_geo_agg.head()
````

**Cek: apakah baris duplikat memengaruhi median?** Tabel geolocation berisi banyak baris duplikat penuh (lihat output di atas). Duplikat tetap ikut dihitung dalam median, sehingga titik yang tercatat berkali-kali memiliki bobot lebih besar. Pengaruhnya dicek dengan membandingkan median sebelum dan sesudah duplikat dibuang:

````python
geo_agg_dedup = (
    df_geolocation.drop_duplicates()
    .groupby('geolocation_zip_code_prefix')[['geolocation_lat', 'geolocation_lng']]
    .median()
)
compare = df_geo_agg.set_index('geolocation_zip_code_prefix').join(geo_agg_dedup, rsuffix='_dedup')

# perkiraan jarak (km) antara dua median
shift_km = 111 * np.sqrt(
    (compare['geolocation_lat'] - compare['geolocation_lat_dedup']) ** 2
    + ((compare['geolocation_lng'] - compare['geolocation_lng_dedup']) * np.cos(np.radians(compare['geolocation_lat']))) ** 2
)

print(f'Median pergeseran        : {shift_km.median():.2f} km')
print('Kode pos bergeser > 1 km :', (shift_km > 1).sum(), 'dari', shift_km.shape[0])
print('Kode pos bergeser > 5 km :', (shift_km > 5).sum())
````

**Keputusan:** duplikat **tetap dipakai** dalam perhitungan median.

**Alasan:**
- Baris duplikat kemungkinan berasal dari beberapa alamat berbeda yang berada di titik koordinat yang sama. Median yang memperhitungkan duplikat justru mengikuti lokasi tempat alamat paling banyak berada.
- Pengaruhnya kecil: sebagian besar kode pos hampir tidak bergeser, dan hanya sedikit kode pos yang bergeser lebih dari 5 km. Angka ini jauh lebih kecil daripada lebar kelompok jarak yang dipakai di analisis (minimal 100 km, lihat 4.7).

### Merge

Tabel koordinat digabung **dua kali**:
1. Dengan `customer_zip_code_prefix` → kolom `customer_lat`, `customer_lng`
2. Dengan `seller_zip_code_prefix` → kolom `seller_lat`, `seller_lng`

Kolom koordinat diberi nama ulang sebelum merge agar lokasi pelanggan dan seller tidak tertukar.

````python
rows_before = df_merged.shape[0]

geo_customer = df_geo_agg.rename(columns={
    'geolocation_zip_code_prefix': 'customer_zip_code_prefix',
    'geolocation_lat': 'customer_lat',
    'geolocation_lng': 'customer_lng',
})
df_merged = df_merged.merge(geo_customer, on='customer_zip_code_prefix', how='left')

geo_seller = df_geo_agg.rename(columns={
    'geolocation_zip_code_prefix': 'seller_zip_code_prefix',
    'geolocation_lat': 'seller_lat',
    'geolocation_lng': 'seller_lng',
})
df_merged = df_merged.merge(geo_seller, on='seller_zip_code_prefix', how='left')

print('Sebelum merge:', rows_before, 'baris')
print('Hasil merge  :', df_merged.shape)
````

````python
has_item = df_merged['seller_id'].notna()
print('Baris tanpa koordinat pelanggan:', df_merged['customer_lat'].isna().sum())
print('Baris (yang punya item) tanpa koordinat seller:', df_merged.loc[has_item, 'seller_lat'].isna().sum())
````

**Hasil:** jumlah baris **tetap 114.092** dan jumlah kolom menjadi **45** (41 + 4 kolom koordinat).

Sebagian kecil baris tidak mendapat koordinat karena kode posnya tidak tercatat di tabel geolocation. Baris ini tetap dipertahankan. Koordinat hanya dibutuhkan untuk analisis jarak dan peta, dan koordinat yang kosong ditangani pada tahap Data Cleaning (3.5d dan 3.7b).

### Catatan: koordinat di luar Brasil

````python
# batas kotak wilayah Brasil (perkiraan)
def outside_brazil(lat, lng):
    return (lat > 5.3) | (lat < -33.8) | (lng > -34.7) | (lng < -74.0)

print('Titik mentah di luar Brasil       :', outside_brazil(df_geolocation['geolocation_lat'], df_geolocation['geolocation_lng']).sum())
print('Kode pos yang median-nya di luar Brasil:', outside_brazil(df_geo_agg['geolocation_lat'], df_geo_agg['geolocation_lng']).sum())
````

Ada titik koordinat mentah yang berada di luar wilayah Brasil (salah input). Sebagian besar sudah tertutupi oleh median, tetapi masih ada beberapa kode pos yang titik median-nya di luar Brasil. Ini akan ditangani pada tahap **Data Cleaning**.

### Cek Integritas Key Antar Tabel

Semua merge memakai `how='left'`, sehingga jumlah baris yang tetap **belum membuktikan** bahwa setiap key berhasil dicocokkan. Baris yang key-nya tidak ditemukan tetap ada, hanya saja kolom dari tabel kanannya kosong. Karena itu dicek dua hal:
1. Setiap key di tabel anak harus ditemukan di tabel induknya (tidak ada data "yatim").
2. Key di tabel induk harus unik, agar merge tidak menggandakan baris.

````python
key_checks = [
    ('orders -> customers', df_orders['customer_id'], df_customers['customer_id']),
    ('order_items -> orders', df_items['order_id'], df_orders['order_id']),
    ('order_items -> products', df_items['product_id'], df_products['product_id']),
    ('order_items -> sellers', df_items['seller_id'], df_sellers['seller_id']),
    ('order_reviews -> orders', df_reviews['order_id'], df_orders['order_id']),
    ('order_payments -> orders', df_payments['order_id'], df_orders['order_id']),
    ('products -> category_translation', df_products['product_category_name'].dropna(), df_category['product_category_name']),
]
integrity = pd.DataFrame([
    {'relasi (anak -> induk)': name, 'jumlah_key': child.shape[0], 'tidak_ditemukan_di_induk': (~child.isin(parent)).sum()}
    for name, child, parent in key_checks
]).set_index('relasi (anak -> induk)')

print('Key induk unik:')
for name, key in [('orders.order_id', df_orders['order_id']), ('customers.customer_id', df_customers['customer_id']),
                  ('products.product_id', df_products['product_id']), ('sellers.seller_id', df_sellers['seller_id']),
                  ('category_translation.product_category_name', df_category['product_category_name'])]:
    print(f'  {name:45s}: {key.is_unique}')
integrity
````

**Hasil:** seluruh key induk unik, dan seluruh key di tabel anak ditemukan di tabel induknya. Satu-satunya pengecualian adalah **13 produk** dengan **2 nama kategori** yang tidak ada di tabel terjemahan (langkah 4), yang sudah dicatat untuk ditangani di Data Cleaning.

Artinya, kolom yang kosong setelah merge **bukan** akibat key yang gagal dicocokkan, melainkan karena datanya memang tidak ada: pesanan tanpa item, tanpa ulasan, atau tanpa pembayaran.

## 3.2 Tipe Data

**Temuan:**
- 8 kolom tanggal masih bertipe teks (`str`), sehingga belum bisa dihitung selisihnya (misal lama pengiriman atau keterlambatan).
- Beberapa kolom yang seharusnya bilangan bulat (ID urutan, kode pos, skor ulasan, jumlah/cicilan pembayaran) berubah menjadi `float64`. Ini terjadi karena pandas mengubah kolom integer menjadi float ketika ada nilai kosong (NaN) setelah proses merge.

````python
date_cols = [
    'order_purchase_timestamp', 'order_approved_at', 'order_delivered_carrier_date',
    'order_delivered_customer_date', 'order_estimated_delivery_date',
    'shipping_limit_date', 'review_creation_date', 'review_answer_timestamp',
]

for col in date_cols:
    df_clean[col] = pd.to_datetime(df_clean[col])

df_clean[date_cols].dtypes
````

Untuk kolom bilangan bulat, digunakan tipe **`Int64`** (huruf I kapital). Tipe ini adalah integer milik pandas yang **bisa menyimpan nilai kosong** (`<NA>`), berbeda dengan `int64` biasa yang akan error jika ada NaN. Dengan begitu nilai kosong tetap terjaga, tetapi angka ditampilkan sebagai bilangan bulat (misal `1` bukan `1.0`).

````python
int_cols = [
    'order_item_id', 'seller_zip_code_prefix', 'review_score',
    'payment_count', 'payment_installments_max',
    'product_name_lenght', 'product_description_lenght', 'product_photos_qty',
]

# pastikan nilainya memang bilangan bulat sebelum diubah (tidak ada angka desimal yang hilang)
for col in int_cols:
    non_integer = (df_clean[col].dropna() % 1 != 0).sum()
    print(f'{col:28s} nilai desimal: {non_integer}')
````

````python
for col in int_cols:
    df_clean[col] = df_clean[col].astype('Int64')

df_clean[int_cols].dtypes
````

**Keputusan:**
- 8 kolom tanggal diubah ke `datetime`.
- 8 kolom bilangan bulat diubah ke `Int64` (setelah dipastikan tidak ada nilai desimal).

**Alasan:** tipe `datetime` dibutuhkan untuk menghitung metrik utama (Late Delivery Rate dan Seller Late Handover Rate) serta durasi pengiriman. Tipe `Int64` membuat data lebih rapi dan sesuai maknanya tanpa menghilangkan informasi nilai kosong.

**Catatan dari rentang tanggal:** `shipping_limit_date` memiliki nilai maksimal di tahun **2020**, jauh setelah periode transaksi berakhir (2018). Nilai ini tidak wajar dan akan dicek pada langkah konsistensi logika (3.6).

## 3.3 Duplikat

````python
print('Baris duplikat penuh:', df_clean.duplicated().sum())
print('Baris dengan kombinasi (order_id, order_item_id) berulang:',
      df_clean.duplicated(subset=['order_id', 'order_item_id']).sum())
````

**Temuan:** tidak ada baris yang duplikat penuh, tetapi ada kombinasi `order_id` + `order_item_id` yang muncul lebih dari sekali. Seharusnya kombinasi ini unik, karena satu item dalam satu pesanan hanya dicatat sekali.

Duplikasi ini berasal dari pesanan yang memiliki **lebih dari satu ulasan** (terlihat pada merge langkah 6): setiap item pada pesanan tersebut ikut tergandakan sesuai jumlah ulasannya. Berikut gambaran ulasan ganda tersebut:

````python
multi_review = df_reviews[df_reviews['order_id'].duplicated(keep=False)].copy()
multi_review['is_low_review'] = multi_review['review_score'] <= 2

review_check = multi_review.groupby('order_id').agg(
    jumlah_ulasan=('review_id', 'size'),
    variasi_skor=('review_score', 'nunique'),
    variasi_status_low_review=('is_low_review', 'nunique'),
)

print('Pesanan dengan >1 ulasan                         :', review_check.shape[0])
print('Pesanan yang skor ulasannya berbeda-beda          :', (review_check['variasi_skor'] > 1).sum())
print('Pesanan yang status low review (1-2)-nya berbeda  :', (review_check['variasi_status_low_review'] > 1).sum())
````

````python
# contoh pesanan dengan ulasan ganda yang skornya berbeda
example_order = review_check[review_check['variasi_skor'] > 1].index[0]
df_reviews[df_reviews['order_id'] == example_order].sort_values('review_answer_timestamp')
````

**Keputusan:** untuk setiap pesanan, dipertahankan **1 ulasan terbaru** berdasarkan `review_answer_timestamp`.

**Alasan:**
- **Low Review Rate** dihitung per pesanan (*reviewed orders*), sehingga satu pesanan harus diwakili oleh satu skor. Tanpa langkah ini, pesanan dengan ulasan ganda akan terhitung 2–3 kali.
- Skor pada ulasan ganda tidak selalu sama, bahkan ada pesanan yang status low review-nya berbeda antar ulasan. Karena itu dibutuhkan satu aturan yang konsisten.
- Ulasan **terbaru** dipilih karena paling mencerminkan penilaian akhir pelanggan, misalnya setelah barang benar-benar diterima atau setelah masalahnya diselesaikan.

````python
rows_before = df_clean.shape[0]

df_clean = (
    df_clean
    .sort_values('review_answer_timestamp')
    .drop_duplicates(subset=['order_id', 'order_item_id'], keep='last')
    .sort_values(['order_purchase_timestamp', 'order_id', 'order_item_id'])
    .reset_index(drop=True)
)

print('Baris sebelum :', rows_before)
print('Baris dihapus :', rows_before - df_clean.shape[0])
print('Baris sekarang:', df_clean.shape[0])
print('Sisa duplikat (order_id, order_item_id):', df_clean.duplicated(subset=['order_id', 'order_item_id']).sum())
print('Jumlah pesanan (tetap):', df_clean['order_id'].nunique())
````

**Hasil:** 667 baris duplikat dihapus sehingga jumlah baris kembali menjadi **113.425**, sama dengan jumlah baris sebelum tabel ulasan digabung (langkah 5). Setiap item kini muncul tepat satu kali, dan **jumlah pesanan tetap 99.441** (tidak ada pesanan yang hilang).

**Apakah `review_id` bisa dipakai sebagai kunci ulasan?** Pada profil tabel (3.1) ditemukan `review_id` yang muncul lebih dari sekali. Dicek apakah `review_id` yang sama dipakai oleh pesanan dari pelanggan yang sama atau dari pelanggan yang berbeda:

````python
orders_per_review = df_reviews.groupby('review_id')['order_id'].nunique()
shared_review_ids = orders_per_review[orders_per_review > 1].index

# pelanggan (customer_unique_id) pemilik pesanan-pesanan yang memakai review_id yang sama
shared = (
    df_reviews[df_reviews['review_id'].isin(shared_review_ids)]
    .merge(df_orders[['order_id', 'customer_id']], on='order_id', how='left')
    .merge(df_customers[['customer_id', 'customer_unique_id']], on='customer_id', how='left')
)
customers_per_review = shared.groupby('review_id')['customer_unique_id'].nunique()

print('review_id yang dipakai >1 pesanan        :', len(shared_review_ids))
print('... semuanya milik 1 pelanggan yang sama :', (customers_per_review == 1).sum())
````

**Temuan:** 789 `review_id` dipakai oleh lebih dari 1 pesanan, dan **ke-789-nya milik 1 pelanggan yang sama**. Polanya seperti satu survei yang dijawab untuk beberapa pesanan sekaligus dari pelanggan yang sama, sehingga bukan kesalahan input.

**Keputusan:** `review_id` **tidak dipakai** sebagai kunci deduplikasi. Kuncinya tetap `order_id` + `order_item_id`, seperti langkah di atas.

**Alasan:** jika deduplikasi memakai `review_id`, ulasan milik pesanan lain dari pelanggan yang sama akan ikut terhapus. Temuan ini juga menjadi dasar dugaan pada anomali tanggal ulasan (3.6d).

## 3.4 Missing Value: Tanggal Proses Pesanan

Tiga kolom tanggal proses pesanan memiliki nilai kosong: `order_approved_at`, `order_delivered_carrier_date`, dan `order_delivered_customer_date`. Tanggal-tanggal ini terisi sesuai tahapan yang sudah dilalui pesanan, sehingga pengecekan dilakukan berdasarkan `order_status`.

Karena tanggal ini melekat pada pesanan (bukan item), pengecekan dilakukan pada **level pesanan** (1 baris per `order_id`) agar pesanan dengan banyak item tidak terhitung berkali-kali.

````python
date_process_cols = ['order_approved_at', 'order_delivered_carrier_date', 'order_delivered_customer_date']

df_order_level = df_clean.drop_duplicates('order_id')

missing_by_status = df_order_level.groupby('order_status').agg(
    total_pesanan=('order_id', 'size'),
    **{f'kosong_{col}': (col, lambda x: x.isna().sum()) for col in date_process_cols},
)
missing_by_status.sort_values('total_pesanan', ascending=False)
````

**Temuan:** pola nilai kosong **sesuai dengan tahapan status pesanan**:
- `shipped` (dikirim tapi belum diterima): semua punya tanggal serah ke kurir, tetapi **belum** punya tanggal diterima pelanggan.
- `invoiced`, `processing`, `unavailable`: belum pernah diserahkan ke kurir, sehingga kedua tanggal pengiriman kosong.
- `created`: pembayaran belum disetujui, sehingga ketiga tanggal kosong.
- `canceled`: sebagian besar dibatalkan sebelum dikirim, sehingga tanggal pengiriman kosong.

Hanya ada sedikit pengecualian pada pesanan `delivered`: 14 pesanan tanpa tanggal approval, 2 tanpa tanggal serah ke kurir, dan 8 tanpa tanggal diterima pelanggan. Pengecualian ini dicek pada langkah konsistensi logika (3.6).

**Keputusan:** nilai kosong pada tanggal proses pesanan **tidak diimputasi (tidak diisi)**.

**Alasan:**
- Nilai kosong tersebut **bermakna**: tahapan itu memang belum atau tidak pernah terjadi. Mengisi tanggalnya sama dengan mengarang kejadian pengiriman yang tidak ada, dan akan membuat metrik keterlambatan menjadi bias.
- Setiap metrik cukup menggunakan pesanan yang datanya relevan, sesuai definisi di Business Understanding:
  - **Late Delivery Rate** hanya menggunakan pesanan berstatus `delivered` yang memiliki tanggal diterima pelanggan.
  - **Seller Late Handover Rate** hanya menggunakan **pesanan dengan 1 seller** yang memiliki tanggal serah ke kurir (pada pesanan multi-seller, tanggal serah ke kurir tidak bisa dikaitkan ke seller tertentu).

````python
eligible_delivery = (
    (df_order_level['order_status'] == 'delivered')
    & df_order_level['order_delivered_customer_date'].notna()
)

# Pilar 3 hanya memakai pesanan dengan 1 seller (lihat bagian 1.6)
sellers_per_order = df_clean.dropna(subset=['seller_id']).groupby('order_id')['seller_id'].nunique()
single_seller_orders = sellers_per_order[sellers_per_order == 1].index

df_single_seller = df_order_level[df_order_level['order_id'].isin(single_seller_orders)]
eligible_handover = df_single_seller['order_delivered_carrier_date'].notna()

print('Pesanan yang bisa dipakai untuk Late Delivery Rate      :', eligible_delivery.sum(),
      f'dari {df_order_level.shape[0]} pesanan')
print('Pesanan 1 seller untuk Seller Late Handover Rate        :', eligible_handover.sum(),
      f'dari {df_single_seller.shape[0]} pesanan 1 seller')
````

**Hasil:** tidak ada data yang diubah pada langkah ini. Sebanyak **96.470 pesanan** dapat dipakai untuk menghitung Late Delivery Rate. Untuk Seller Late Handover Rate, dapat dipakai **96.380 pesanan 1 seller** (dari 97.388 pesanan 1 seller). Angka ini masih sebelum pengecekan anomali dan pembatasan periode di langkah berikutnya.

## 3.6 Konsistensi & Anomali Logika

Pada langkah ini dicek data yang **secara logika tidak mungkin terjadi**. Data yang bermasalah tidak langsung dihapus, tetapi diberi **flag** (kolom penanda `True/False`). Dengan begitu, setiap metrik dapat mengecualikan data yang relevan saja, tanpa membuang informasi lain dari pesanan tersebut.

Ada empat pengecekan:
- a. Status pesanan vs tanggal
- b. Urutan waktu proses pesanan
- c. Batas waktu serah ke kurir (`shipping_limit_date`)
- d. Tanggal ulasan

### a. Status Pesanan vs Tanggal

Pesanan `delivered` seharusnya memiliki tanggal diterima pelanggan, dan pesanan `canceled` seharusnya tidak memiliki tanggal diterima.

````python
df_clean['is_status_inconsistent'] = (
    ((df_clean['order_status'] == 'delivered') & df_clean['order_delivered_customer_date'].isna())
    | ((df_clean['order_status'] == 'canceled') & df_clean['order_delivered_customer_date'].notna())
)

(df_clean.drop_duplicates('order_id')
 .groupby('order_status')['is_status_inconsistent'].sum()
 .loc[lambda x: x > 0]
 .to_frame('jumlah_pesanan'))
````

**Temuan:** 8 pesanan berstatus `delivered` tanpa tanggal diterima, dan 6 pesanan berstatus `canceled` tetapi memiliki tanggal diterima.

**Keputusan:** diberi flag `is_status_inconsistent`.

**Alasan:** jumlahnya sangat kecil (14 pesanan). Pesanan `delivered` tanpa tanggal sudah otomatis tidak dipakai di Late Delivery Rate (langkah 3.4), sedangkan pesanan `canceled` tidak termasuk pesanan terkirim. Flag ini menjaga agar kasus yang saling bertentangan ini tidak ikut terbaca sebagai pesanan normal pada analisis status pesanan (Pilar 1). Pesanan ini dikecualikan saat tabel analisis ulasan (`df_review`) dibuat di awal bagian 5.1.

### b. Urutan Waktu Proses Pesanan

Alur normal pesanan: **pembelian → approval pembayaran → diserahkan ke kurir → diterima pelanggan**. Tanggal yang urutannya terbalik tidak mungkin terjadi.

````python
time_checks = {
    'approved < purchase': df_clean['order_approved_at'] < df_clean['order_purchase_timestamp'],
    'carrier < purchase': df_clean['order_delivered_carrier_date'] < df_clean['order_purchase_timestamp'],
    'carrier < approved': df_clean['order_delivered_carrier_date'] < df_clean['order_approved_at'],
    'delivered < carrier': df_clean['order_delivered_customer_date'] < df_clean['order_delivered_carrier_date'],
    'delivered < approved': df_clean['order_delivered_customer_date'] < df_clean['order_approved_at'],
    'delivered < purchase': df_clean['order_delivered_customer_date'] < df_clean['order_purchase_timestamp'],
}

pd.DataFrame({
    'jumlah_pesanan': {name: df_clean.loc[mask, 'order_id'].nunique() for name, mask in time_checks.items()}
})
````

````python
# seberapa jauh selisih waktunya pada kasus terbanyak (serah ke kurir sebelum approval)
df_order_level = df_clean.drop_duplicates('order_id')
mask = df_order_level['order_delivered_carrier_date'] < df_order_level['order_approved_at']
gap_hours = (df_order_level.loc[mask, 'order_approved_at'] - df_order_level.loc[mask, 'order_delivered_carrier_date']).dt.total_seconds() / 3600

print('Median selisih (jam)         :', round(gap_hours.median(), 1))
print('Selisih < 24 jam (%)         :', round((gap_hours < 24).mean() * 100, 1))
````

**Temuan:**
- Tanggal awal dan akhir aman: tidak ada approval sebelum pembelian dan tidak ada pesanan yang diterima sebelum dibeli.
- Kasus terbanyak adalah **serah ke kurir tercatat sebelum approval pembayaran**. Selisihnya umumnya hanya hitungan jam (median ±17 jam, mayoritas di bawah 24 jam).
- Ada juga sebagian kecil pesanan yang tanggal serah ke kurir tercatat sebelum pembelian, atau tanggal diterima sebelum serah ke kurir.

Pola selisih yang kecil ini lebih mengarah pada **masalah pencatatan waktu di sistem** (misalnya perbedaan zona waktu atau keterlambatan input timestamp), bukan kejadian nyata.

**Keputusan:** diberi flag `is_time_anomaly`.

**Alasan:**
- Kesalahan hanya terjadi pada tanggal di tengah proses. Tanggal pembelian, estimasi, dan diterima pelanggan tetap valid, sehingga pesanan ini **tetap dipakai untuk Late Delivery Rate** (diterima vs estimasi).
- Pesanan ber-flag **dikecualikan dari Seller Late Handover Rate dan perhitungan durasi per tahap** (proses seller vs pengantaran kurir), karena tanggal serah ke kurirnya tidak bisa dipercaya dan akan menghasilkan durasi negatif.

````python
df_clean['is_time_anomaly'] = np.logical_or.reduce(list(time_checks.values()))
print('Pesanan dengan anomali urutan waktu:', df_clean.loc[df_clean['is_time_anomaly'], 'order_id'].nunique())
````

### c. Batas Waktu Serah ke Kurir (`shipping_limit_date`)

Pada langkah 3.2 ditemukan `shipping_limit_date` yang mencapai tahun 2020, padahal transaksi berakhir di 2018. Kolom ini menjadi dasar **Seller Late Handover Rate** (Pilar 3), sehingga perlu dicek lebih teliti. Pengecekan dilakukan pada level pasangan pesanan–seller, karena batas waktu berlaku per seller dalam satu pesanan.

````python
df_seller_order = df_clean.dropna(subset=['seller_id']).drop_duplicates(['order_id', 'seller_id']).copy()
df_seller_order['limit_days'] = (
    df_seller_order['shipping_limit_date'] - df_seller_order['order_purchase_timestamp']
).dt.total_seconds() / 86400

df_seller_order['limit_days'].describe(percentiles=[0.5, 0.9, 0.99, 0.999]).round(1).to_frame('jarak_hari_pembelian_ke_batas_serah')
````

````python
for threshold in [30, 60, 90, 365]:
    print(f'Batas serah > {threshold:>3} hari setelah pembelian:', (df_seller_order['limit_days'] > threshold).sum())

df_seller_order.loc[df_seller_order['limit_days'] > 365,
                    ['order_id', 'order_purchase_timestamp', 'shipping_limit_date', 'order_delivered_carrier_date', 'order_status']]
````

**Temuan:**
- Normalnya seller diberi waktu sekitar **6 hari** setelah pembelian untuk menyerahkan barang ke kurir (99% di bawah 19 hari).
- Nilai 30–60 hari masih membentuk sebaran yang menyambung (bukan lompatan), sehingga masih masuk akal, misalnya untuk produk yang butuh waktu persiapan lebih lama.
- Ada **3 pasangan pesanan–seller** dengan batas serah **lebih dari 2,5 tahun** setelah pembelian (tahun 2020), padahal barangnya bahkan sudah diserahkan ke kurir di 2017. Ini jelas salah input.

**Keputusan:** 3 pasangan tersebut diberi flag `is_limit_anomaly`.

**Alasan:** dengan batas serah di tahun 2020, seller akan selalu terlihat "tepat waktu", sehingga data ini akan membiaskan Seller Late Handover Rate. Data ini dikecualikan dari perhitungan metrik tersebut, tetapi pesanannya tetap dipakai untuk analisis lain.

````python
df_clean['is_limit_anomaly'] = (
    (df_clean['shipping_limit_date'] - df_clean['order_purchase_timestamp']).dt.days > 365
)
print('Baris item dengan anomali batas serah:', df_clean['is_limit_anomaly'].sum())
````

### d. Tanggal Ulasan

Survei kepuasan dikirim setelah pembelian, sehingga tanggal ulasan dibuat (`review_creation_date`) tidak mungkin lebih awal dari tanggal pembelian.

````python
df_order_level = df_clean.drop_duplicates('order_id')
review_before_purchase = df_order_level['review_creation_date'] < df_order_level['order_purchase_timestamp'].dt.normalize()

print('Pesanan dengan ulasan dibuat sebelum pembelian:', review_before_purchase.sum())
df_order_level.loc[review_before_purchase,
                   ['order_id', 'order_purchase_timestamp', 'review_creation_date', 'order_status', 'review_score']].head()
````

**Temuan:** 61 pesanan memiliki ulasan yang dibuat **sebelum** pesanannya dibeli (rata-rata ±20 hari lebih awal). Ini tidak mungkin terjadi. Kemungkinan ulasan tersebut sebenarnya milik pesanan lain dari pelanggan yang sama, karena sebelumnya ditemukan bahwa satu `review_id` bisa tercatat di beberapa pesanan.

**Keputusan:** diberi flag `is_review_anomaly`.

**Alasan:** skor ulasan pada pesanan ini tidak bisa dipastikan benar-benar menilai pesanan tersebut, sehingga dikecualikan dari perhitungan **Low Review Rate**. Pesanannya tetap dipakai untuk analisis pengiriman.

````python
df_clean['is_review_anomaly'] = (
    df_clean['review_creation_date'] < df_clean['order_purchase_timestamp'].dt.normalize()
)
print('Pesanan dengan anomali tanggal ulasan:', df_clean.loc[df_clean['is_review_anomaly'], 'order_id'].nunique())
````

### Ringkasan Flag

| Flag | Level | Jumlah | Dikecualikan dari |
|---|---|---|---|
| `is_status_inconsistent` | pesanan | 14 pesanan | Analisis status pesanan |
| `is_time_anomaly` | pesanan | ±1.400 pesanan | Seller Late Handover Rate & durasi per tahap |
| `is_limit_anomaly` | pesanan–seller | 3 pasangan | Seller Late Handover Rate |
| `is_review_anomaly` | pesanan | 61 pesanan | Low Review Rate |

Tidak ada baris yang dihapus pada langkah ini.

````python
flag_cols = ['is_status_inconsistent', 'is_time_anomaly', 'is_limit_anomaly', 'is_review_anomaly']
pd.DataFrame({
    'jumlah_pesanan': {col: df_clean.loc[df_clean[col], 'order_id'].nunique() for col in flag_cols},
    'jumlah_baris': {col: df_clean[col].sum() for col in flag_cols},
})
````

## 3.7 Konsistensi Data Lokasi

Data lokasi dipakai pada **Pilar 2** (rute asal–tujuan, same-state vs cross-state, dan jarak seller–pelanggan) dan untuk membandingkan apakah rute sebaiknya dihitung per kota atau per state (5.2.4). Bagian ini disusun sebagai berikut:
- a. Menentukan acuan utama lokasi
- b. Koordinat
- c. State pelanggan dan seller
- d. Nama kota
- e. Ringkasan

### a. Menentukan Acuan Utama Lokasi

Lokasi pelanggan dan seller tercatat di beberapa kolom yang bisa saling bertentangan:
- **kode pos**, **nama kota**, dan **state** pada tabel customers dan sellers,
- tabel **geolocation**, yang menghubungkan kode pos dengan koordinat, nama kota, dan state.

Sebelum membersihkan, perlu ditentukan dulu **sumber mana yang paling dapat dipercaya** sebagai acuan. Pengecekan dilakukan pada **tabel mentah** (sebelum merge). Sebagai pembanding yang independen dari dataset, dipakai **pembagian rentang kode pos (CEP) per state** yang berlaku resmi di Brasil. Contohnya, kode pos 80000–87999 selalu berada di state PR.

Catatan: kode pos di dataset tersimpan sebagai angka, sehingga angka 0 di depannya hilang (`1003` = `01003`, di SP).

````python
import re
import unicodedata

# rentang 5 digit awal kode pos (CEP) untuk setiap state di Brasil
cep_ranges = [
    (1000, 19999, 'SP'), (20000, 28999, 'RJ'), (29000, 29999, 'ES'), (30000, 39999, 'MG'),
    (40000, 48999, 'BA'), (49000, 49999, 'SE'), (50000, 56999, 'PE'), (57000, 57999, 'AL'),
    (58000, 58999, 'PB'), (59000, 59999, 'RN'), (60000, 63999, 'CE'), (64000, 64999, 'PI'),
    (65000, 65999, 'MA'), (66000, 68899, 'PA'), (68900, 68999, 'AP'), (69000, 69299, 'AM'),
    (69300, 69399, 'RR'), (69400, 69899, 'AM'), (69900, 69999, 'AC'), (70000, 72799, 'DF'),
    (72800, 72999, 'GO'), (73000, 73699, 'DF'), (73700, 76799, 'GO'), (76800, 76999, 'RO'),
    (77000, 77999, 'TO'), (78000, 78899, 'MT'), (79000, 79999, 'MS'), (80000, 87999, 'PR'),
    (88000, 89999, 'SC'), (90000, 99999, 'RS'),
]

def state_from_zip(zip_code):
    # state menurut rentang kode pos; NaN jika di luar semua rentang
    for start, end, state in cep_ranges:
        if start <= zip_code <= end:
            return state
    return np.nan

# 1) kode pos dan state pada tabel mentah dibandingkan dengan rentang kode pos
source_check = []
for name, data, zip_col, state_col in [
    ('customers', df_customers, 'customer_zip_code_prefix', 'customer_state'),
    ('sellers', df_sellers, 'seller_zip_code_prefix', 'seller_state'),
    ('geolocation', df_geolocation, 'geolocation_zip_code_prefix', 'geolocation_state'),
]:
    zip_state = data[zip_col].map({z: state_from_zip(z) for z in data[zip_col].dropna().unique()})
    source_check.append({
        'tabel_mentah': name,
        'jumlah_baris': data.shape[0],
        'kode_pos_kosong': data[zip_col].isna().sum(),
        'kode_pos_di_luar_rentang': (data[zip_col].notna() & zip_state.isna()).sum(),
        'state_tidak_sesuai_kode_pos': (zip_state.notna() & (data[state_col] != zip_state)).sum(),
    })

source_check = pd.DataFrame(source_check).set_index('tabel_mentah')
source_check['persen_state_tidak_sesuai'] = (source_check['state_tidak_sesuai_kode_pos'] / source_check['jumlah_baris'] * 100).round(3)
source_check
````

````python
# 2) kualitas penulisan nama kota pada tabel mentah
city_check = []
for name, city in [('customers', df_customers['customer_city']), ('sellers', df_sellers['seller_city']),
                   ('geolocation', df_geolocation['geolocation_city'])]:
    names = pd.Series(city.dropna().unique())
    has_accent = names.map(lambda x: unicodedata.normalize('NFKD', x).encode('ascii', 'ignore').decode() != x)
    city_check.append({
        'tabel_mentah': name,
        'nama_kota_unik': len(names),
        'beraksen': has_accent.sum(),
        'berisi_simbol_angka_email': names.str.contains(r'[/\\@\d,%]').sum(),
    })

pd.DataFrame(city_check).set_index('tabel_mentah')
````

**Temuan:**

| Sumber | Hasil pengecekan |
|---|---|
| **Kode pos** | Berupa angka, tidak ada yang kosong, dan **100% berada di rentang kode pos resmi** pada ketiga tabel. Tidak mungkin salah ketik seperti teks. |
| **State pelanggan** | **100% sesuai** dengan state menurut kode pos (0 dari 99.441). |
| **State seller** | **35 seller (±1,1%)** tidak sesuai dengan state menurut kode pos. Hampir semuanya tertulis SP. |
| **Kota pelanggan** | Rapi: tanpa aksen, tanpa simbol (1 nama berisi angka, `quilometro 14 do mutum`, yang memang nama distrik). Kemungkinan besar terisi otomatis dari kode pos. |
| **Kota seller** | Diketik bebas oleh seller: ada simbol, alamat email, dan angka (dibahas di bagian d). |
| **Geolocation** | Penulisan kotanya paling beragam (±8.000 variasi, ±2.000 beraksen, bahkan ada kode HTML seperti `d%26apos%3b`). Ada 9 baris yang state-nya tidak sesuai dan 42 titik di luar Brasil (langkah 8). Namun setiap kode pos memiliki banyak titik, sehingga kesalahan per baris tertutupi jika data **diringkas per kode pos**. |

Konsistensi 100% pada data pelanggan menunjukkan bahwa hubungan kode pos → state dapat diandalkan. Ketidaksesuaian pada seller berasal dari state yang **diketik** seller, bukan dari kode posnya.

**Keputusan: kode pos dijadikan acuan utama lokasi.**

| Urutan | Sumber | Peran |
|---|---|---|
| 1 | **Kode pos** | Acuan utama untuk semua informasi lokasi |
| 2 | **Rentang kode pos resmi** | Menentukan state dari kode pos |
| 3 | **Geolocation** | Kamus yang dibaca **lewat kode pos**: median koordinat dari titik-titik yang valid (bagian b) dan nama kota yang paling sering tercatat per kode pos (bagian d) |
| 4 | Kota dan state yang **ditulis** | Pembanding, dan dipakai hanya jika kode pos tidak dapat dipetakan |

Dengan acuan ini, **koordinat, state, dan kota semuanya berasal dari sumber yang sama**, sehingga rute, jarak, dan nama lokasi selalu konsisten satu sama lain. Acuan ini diterapkan pada koordinat (b), state (c), dan nama kota (d).

### b. Koordinat

Koordinat pelanggan dan seller berasal dari tabel geolocation, yaitu **median seluruh titik per kode pos** (langkah 8). Sesuai acuan di bagian a, tabel geolocation diperlakukan sebagai kamus yang **dicek dulu terhadap kode pos** sebelum dipakai. Ada dua langkah:
1. **Membuang titik yang tidak valid sebelum median dihitung ulang**, yaitu titik yang state-nya tidak sesuai rentang kode pos, atau koordinatnya berada di luar Brasil. Dengan cara ini, kesalahan diperbaiki di sumbernya, bukan setelah merge.
2. **Memperkirakan koordinat yang kosong** dari kode pos tetangga, jika akurasinya terbukti cukup baik.

**Langkah 1: cek setiap titik geolocation terhadap acuan**

````python
geo_points = df_geolocation[['geolocation_zip_code_prefix', 'geolocation_lat', 'geolocation_lng', 'geolocation_state']].copy()
geo_points['state_kode_pos'] = geo_points['geolocation_zip_code_prefix'].map(
    {z: state_from_zip(z) for z in geo_points['geolocation_zip_code_prefix'].unique()}
)
wrong_state = geo_points['geolocation_state'] != geo_points['state_kode_pos']
out_of_brazil = outside_brazil(geo_points['geolocation_lat'], geo_points['geolocation_lng'])

print('Titik geolocation                  :', geo_points.shape[0])
print('State tidak sesuai rentang kode pos:', wrong_state.sum())
print('Koordinat di luar Brasil           :', out_of_brazil.sum())
print('Titik yang dibuang                 :', (wrong_state | out_of_brazil).sum())

zip_coord_old = geo_points.groupby('geolocation_zip_code_prefix')[['geolocation_lat', 'geolocation_lng']].median()
zip_coord = geo_points[~wrong_state & ~out_of_brazil].groupby('geolocation_zip_code_prefix')[['geolocation_lat', 'geolocation_lng']].median()

print('\nKode pos dengan median di luar Brasil:',
      outside_brazil(zip_coord_old['geolocation_lat'], zip_coord_old['geolocation_lng']).sum(), '->',
      outside_brazil(zip_coord['geolocation_lat'], zip_coord['geolocation_lng']).sum())
print('Kode pos yang tidak punya titik valid sama sekali:', zip_coord_old.shape[0] - zip_coord.shape[0])
````

**Langkah 2: memperkirakan koordinat yang kosong**

Koordinat kosong berasal dari kode pos yang tidak tercatat di tabel geolocation (3.5d), ditambah kode pos yang kehilangan seluruh titiknya pada langkah 1.

Kode pos Brasil tersusun **berjenjang**: semakin banyak digit awal yang sama, semakin berdekatan lokasinya. Karena itu, koordinat kode pos yang kosong dapat diperkirakan dari **median koordinat kode pos lain dengan 4 digit awal yang sama**. Kelompok 4 digit tidak pernah melewati batas state, karena batas rentang kode pos antar-state selalu berada di kelipatan 100.

Sebelum dipakai, akurasinya **diuji**: 2.000 kode pos yang koordinatnya diketahui "disembunyikan", lalu diperkirakan dengan cara yang sama dan dibandingkan dengan koordinat aslinya.

````python
def approx_km(lat1, lng1, lat2, lng2):
    # perkiraan jarak (km) untuk selisih koordinat yang kecil
    return 111 * np.sqrt((lat1 - lat2) ** 2 + ((lng1 - lng2) * np.cos(np.radians(lat1))) ** 2)

test_zip = zip_coord.sample(2000, random_state=42)
known_zip = zip_coord.drop(test_zip.index)
prefix4_known = known_zip.groupby(known_zip.index // 10).median()

estimate = prefix4_known.reindex(test_zip.index // 10)
error_km = approx_km(test_zip['geolocation_lat'].values, test_zip['geolocation_lng'].values,
                     estimate['geolocation_lat'].values, estimate['geolocation_lng'].values)
error_km = pd.Series(error_km).dropna()

print(f'Kode pos uji yang dapat diperkirakan: {error_km.shape[0]} dari {test_zip.shape[0]}')
print(f'Median error                        : {error_km.median():.1f} km')
print(f'90% error di bawah                  : {error_km.quantile(0.9):.1f} km')
print(f'Error di atas 100 km                : {(error_km > 100).mean() * 100:.1f}%')
````

**Hasil uji:** dari 2.000 kode pos uji, 1.704 dapat diperkirakan (sisanya tidak memiliki tetangga dengan 4 digit awal yang sama). **Median error hanya ±2 km**, 90% perkiraan meleset kurang dari ±30 km, dan hanya ±3% yang meleset lebih dari 100 km.

**Keputusan:** perkiraan ini **dipakai** untuk mengisi koordinat yang kosong. Kode pos yang tidak memiliki tetangga 4 digit tetap dibiarkan kosong.

**Alasan:** akurasinya cukup untuk kebutuhan analisis. Jarak dipakai dalam kelompok yang lebarnya minimal 100 km (4.7), sehingga error ±2 km hampir tidak pernah memindahkan pesanan ke kelompok lain. Jumlah pesanan yang diperkirakan juga sangat sedikit, sehingga risiko dari ±3% perkiraan yang meleset jauh dapat diabaikan.

````python
prefix4_coord = zip_coord.groupby(zip_coord.index // 10).median()
coord_summary = []

for role in ['customer', 'seller']:
    zip_code = pd.to_numeric(df_clean[f'{role}_zip_code_prefix'], errors='coerce')
    old_lat, old_lng = df_clean[f'{role}_lat'].copy(), df_clean[f'{role}_lng'].copy()

    # koordinat dari kode pos sendiri; jika tidak ada, dari kode pos dengan 4 digit awal yang sama
    own = pd.DataFrame(zip_coord.reindex(zip_code.values).values, index=df_clean.index, columns=['lat', 'lng'])
    nearby = pd.DataFrame(prefix4_coord.reindex((zip_code // 10).values).values, index=df_clean.index, columns=['lat', 'lng'])
    is_estimated = own['lat'].isna() & nearby['lat'].notna()
    coord = own.fillna(nearby)

    df_clean[f'{role}_lat'] = coord['lat']
    df_clean[f'{role}_lng'] = coord['lng']

    entity = pd.DataFrame({
        'id': df_clean[f'{role}_id'],
        'dari_kode_pos_sendiri': own['lat'].notna(),
        'diperkirakan': is_estimated,
        'tetap_kosong': coord['lat'].isna() & zip_code.notna(),
        'bergeser_lebih_1_km': approx_km(old_lat, old_lng, coord['lat'], coord['lng']) > 1,
    }).dropna(subset=['id']).drop_duplicates('id')
    coord_summary.append({
        'pihak': role,
        'jumlah': entity.shape[0],
        'dari_kode_pos_sendiri': entity['dari_kode_pos_sendiri'].sum(),
        'diperkirakan_4_digit': entity['diperkirakan'].sum(),
        'tetap_kosong': entity['tetap_kosong'].sum(),
        'koordinat_bergeser_>1_km': entity['bergeser_lebih_1_km'].sum(),
        'pesanan_mendapat_koordinat': df_clean.loc[is_estimated, 'order_id'].nunique(),
    })

print('Sisa koordinat pelanggan di luar Brasil:', outside_brazil(df_clean['customer_lat'], df_clean['customer_lng']).sum())
print('Sisa koordinat seller di luar Brasil   :', outside_brazil(df_clean['seller_lat'], df_clean['seller_lng']).sum())
pd.DataFrame(coord_summary).set_index('pihak')
````

**Hasil:**
- **51 titik geolocation yang tidak valid dibuang** sebelum median dihitung ulang. Hasilnya, **tidak ada lagi kode pos dengan koordinat di luar Brasil** (sebelumnya 8).
- 4 pesanan yang koordinat pelanggannya sebelumnya berada di sekitar Portugal kini memiliki koordinat yang sesuai dengan state-nya (PR, PA, BA). Pesanan ini tidak lagi dikosongkan, tetapi diperbaiki dari titik yang valid atau dari kode pos tetangganya.
- 25 pelanggan mengalami pergeseran koordinat lebih dari 1 km, karena titik yang salah tidak lagi ikut dihitung.
- Koordinat **229 pelanggan** (229 pesanan) dan **4 seller** (176 pesanan) diperkirakan dari kode pos tetangga, sehingga jarak pengirimannya kini dapat dihitung.
- **50 pelanggan dan 3 seller** tetap tanpa koordinat karena tidak memiliki kode pos tetangga. Koordinat mereka dibiarkan kosong (tidak dikarang), tetapi pesanannya tetap dipakai pada analisis berbasis state.

### c. State Pelanggan dan Seller

Brasil memiliki **27 unit federasi** (26 negara bagian + Distrito Federal), masing-masing dengan kode 2 huruf. Pertama dicek apakah kode state yang tercatat valid:

````python
brazil_states = {
    'AC', 'AL', 'AP', 'AM', 'BA', 'CE', 'DF', 'ES', 'GO', 'MA', 'MT', 'MS', 'MG', 'PA',
    'PB', 'PR', 'PE', 'PI', 'RJ', 'RN', 'RS', 'RO', 'RR', 'SC', 'SP', 'SE', 'TO',
}

customer_states = set(df_clean['customer_state'].dropna())
seller_states = set(df_clean['seller_state'].dropna())

print('Jumlah state pelanggan:', len(customer_states), '| kode tidak valid:', customer_states - brazil_states)
print('Jumlah state seller   :', len(seller_states), '| kode tidak valid:', seller_states - brazil_states)
````

Seluruh kode state pelanggan dan seller valid, berupa 2 huruf kapital, dan tidak ada nilai kosong selain pada pesanan tanpa item. Namun kode yang **valid belum tentu benar**. Karena itu state dibandingkan dengan **state menurut kode pos** (acuan bagian a):

````python
customer_loc = df_clean[['customer_id', 'customer_zip_code_prefix', 'customer_state']].drop_duplicates('customer_id')
seller_loc = df_clean[['seller_id', 'seller_zip_code_prefix', 'seller_city', 'seller_state']].dropna(subset=['seller_id']).drop_duplicates('seller_id')

customer_state_zip = customer_loc['customer_zip_code_prefix'].map(state_from_zip)
seller_loc['state_kode_pos'] = seller_loc['seller_zip_code_prefix'].map(state_from_zip)
state_mismatch = seller_loc[seller_loc['seller_state'] != seller_loc['state_kode_pos']].copy()

# pembanding kedua: state yang paling sering tercatat untuk kode pos tersebut di tabel geolocation
zip_to_state_geo = (
    df_geolocation.groupby(['geolocation_zip_code_prefix', 'geolocation_state']).size().reset_index(name='n')
    .sort_values('n').drop_duplicates('geolocation_zip_code_prefix', keep='last')
    .set_index('geolocation_zip_code_prefix')['geolocation_state']
)
state_mismatch['state_geolocation'] = state_mismatch['seller_zip_code_prefix'].map(zip_to_state_geo)

print('Pelanggan dengan state tidak sesuai kode pos:', (customer_loc['customer_state'] != customer_state_zip).sum(), 'dari', customer_loc.shape[0])
print('Seller dengan state tidak sesuai kode pos   :', state_mismatch.shape[0], 'dari', seller_loc.shape[0])
print('Rentang kode pos dan geolocation sepakat    :',
      (state_mismatch['state_kode_pos'] == state_mismatch['state_geolocation']).sum(), 'dari', state_mismatch.shape[0], 'seller')
state_mismatch[['seller_city', 'seller_state', 'state_kode_pos', 'state_geolocation']].sort_values('state_kode_pos')
````

**Temuan:**
- **Pelanggan:** state seluruhnya sesuai dengan kode pos.
- **Seller:** 35 seller memiliki state yang tidak sesuai dengan kode posnya, dan hampir semuanya tertulis **SP**. Dua acuan yang independen, yaitu **rentang kode pos resmi** dan **tabel geolocation**, sepakat pada seluruh 35 seller.
- Nama kota yang ditulis seller umumnya ikut membuktikan bahwa state tertulis yang salah. Contohnya seller yang menulis kota `curitiba` (ibu kota PR) atau `porto alegre` (ibu kota RS), tetapi state-nya SP.
- Satu-satunya pengecualian adalah seller yang menulis `sao paulo` / SP, padahal kode posnya berada di MG. Seller ini tetap mengikuti kode pos, karena koordinat (dan jarak pengiriman) seller ini juga berasal dari kode pos yang sama. Jika state-nya dibiarkan SP, rute dan jaraknya akan saling bertentangan.

**Keputusan:** `seller_state` pada 35 seller tersebut diganti dengan **state menurut kode pos**. `customer_state` tidak diubah karena sudah sesuai.

**Alasan:** analisis rute (Pilar 2) memakai `seller_state` sebagai asal pengiriman. Seller yang tercatat di SP padahal berada di Curitiba (PR) membuat pesanannya masuk ke rute `SP → PR`, padahal seharusnya `PR → PR`. Barisnya tidak di-drop karena kolom lain pada pesanan tersebut tetap valid.

````python
state_fix = state_mismatch.set_index('seller_id')['state_kode_pos']
to_fix = df_clean['seller_id'].isin(state_fix.index)
df_clean.loc[to_fix, 'seller_state'] = df_clean.loc[to_fix, 'seller_id'].map(state_fix)

seller_after = df_clean.dropna(subset=['seller_id']).drop_duplicates('seller_id')
print('Seller yang state-nya dikoreksi:', state_fix.shape[0])
print('Pesanan yang terdampak         :', df_clean.loc[to_fix, 'order_id'].nunique())
print('Sisa seller dengan state tidak sesuai kode pos:',
      (seller_after['seller_state'] != seller_after['seller_zip_code_prefix'].map(state_from_zip)).sum())
````

### d. Nama Kota

````python
seller_cities = pd.Series(df_clean['seller_city'].dropna().unique())
odd_seller_city = seller_cities[seller_cities.str.contains(r'[/\\@\d]')]

print('Jumlah nama kota seller unik:', len(seller_cities))
print('Contoh nama kota seller tidak konsisten:')
print(odd_seller_city.head(10).to_string(index=False))
````

**Temuan:**
- Nama kota **seller** diketik bebas, sehingga tidak konsisten. Ada yang ditulis bersama state (`sbc/sp`, `lages - sc`), berisi alamat email atau angka kode pos, salah ketik (`sao pauo`, `garulhos`), berupa singkatan (`sbc`, `sp`), atau berupa nama state (`bahia`, `minas gerais`).
- Ada juga seller yang menulis nama kota yang benar, tetapi berbeda dengan lokasi kode posnya. Contohnya menulis `sao paulo`, padahal kode posnya di kota sekitar São Paulo.
- Nama kota **pelanggan** jauh lebih rapi (bagian a).

**Keputusan:** sesuai acuan di bagian a, **nama kota diambil dari kode pos**, untuk seller maupun pelanggan:

1. **Standardisasi penulisan.** Diterapkan pada nama kota di tabel geolocation dan pada nama kota yang ditulis, agar keduanya sebanding:
   - kode HTML `%26apos%3b` dikembalikan menjadi apostrof,
   - huruf kecil dan tanpa aksen (`são paulo` → `sao paulo`),
   - hanya mengambil nama kota sebelum tanda `/`, `\`, `,`, atau `(` (`maua/sao paulo` → `maua`),
   - tanda hubung dan apostrof diganti spasi, sehingga `mogi-guacu` dan `mogi guacu` menjadi sama,
   - akhiran kode state dihapus **jika sama dengan state baris tersebut** (`lages - sc` → `lages`). Syarat ini mencegah kota asli ikut terpotong, misalnya `sento se` (kota Sento Sé di state BA),
   - spasi ganda dirapikan.
2. **Kota menurut kode pos.** Untuk setiap kode pos, diambil nama kota yang **paling sering tercatat** di tabel geolocation. Nama inilah yang dipakai sebagai kota seller dan pelanggan.
3. **Cadangan.** Jika kode pos tidak ada di tabel geolocation, dipakai nama kota yang ditulis (setelah distandardisasi).

**Alasan:**
- Kode pos tidak mungkin salah ketik, sedangkan nama kota yang diketik bisa salah dengan banyak cara. Mengambil kota dari kode pos menghindari menebak typo satu per satu.
- Koordinat, state, dan kota berasal dari kode pos yang sama, sehingga ketiganya selalu konsisten.
- Seller dan pelanggan memakai **kamus nama kota yang sama**, sehingga nama kota keduanya dapat dicocokkan pada analisis rute kota → kota.
- Barisnya tidak di-drop karena kolom lain pada pesanan tersebut tetap valid.

````python
def clean_city_text(city, state):
    # standardisasi penulisan nama kota
    if not isinstance(city, str):
        return city
    city = city.replace('%26apos%3b', "'")                    # kode HTML -> apostrof
    city = unicodedata.normalize('NFKD', city).encode('ascii', 'ignore').decode().lower()
    city = re.split(r'[/\\,(]', city)[0]                    # 'maua/sao paulo' -> 'maua'
    city = re.sub(r"[-'´]", ' ', city)                       # 'mogi-guacu' -> 'mogi guacu'
    city = re.sub(r'\s+', ' ', city).strip()                 # 'sao  paulo' -> 'sao paulo'
    if isinstance(state, str):
        city = re.sub(rf' {state.lower()}$', '', city)       # 'lages - sc' -> 'lages' (hanya state miliknya sendiri)
    return city.strip()

# kamus kode pos -> kota: nama kota (sudah distandardisasi) yang paling sering tercatat per kode pos
geo_pairs = df_geolocation[['geolocation_city', 'geolocation_state']].drop_duplicates()
geo_pairs['city'] = [clean_city_text(c, s) for c, s in zip(geo_pairs['geolocation_city'], geo_pairs['geolocation_state'])]
geo_city = df_geolocation.merge(geo_pairs, on=['geolocation_city', 'geolocation_state'])

zip_to_city = (
    geo_city.groupby(['geolocation_zip_code_prefix', 'city']).size().reset_index(name='n')
    .sort_values('n').drop_duplicates('geolocation_zip_code_prefix', keep='last')
    .set_index('geolocation_zip_code_prefix')['city']
)

print('Variasi nama kota geolocation:', df_geolocation['geolocation_city'].nunique(), '->', geo_city['city'].nunique(), 'setelah standardisasi')
print('Kode pos di kamus            :', zip_to_city.shape[0])
````

````python
original_city = df_clean[['customer_city', 'seller_city']].copy()
typed_city = {}

for role in ['customer', 'seller']:
    typed_city[role] = pd.Series(
        [clean_city_text(c, s) for c, s in zip(df_clean[f'{role}_city'], df_clean[f'{role}_state'])], index=df_clean.index
    )
    # kota menurut kode pos; cadangan: kota yang ditulis (sudah distandardisasi)
    df_clean[f'{role}_city'] = df_clean[f'{role}_zip_code_prefix'].map(zip_to_city).fillna(typed_city[role])

city_summary = []
for role in ['customer', 'seller']:
    entity = pd.DataFrame({
        'id': df_clean[f'{role}_id'],
        'kota_tertulis': original_city[f'{role}_city'],
        'kota_tertulis_standar': typed_city[role],
        'kota_final': df_clean[f'{role}_city'],
        'kode_pos_ada_di_kamus': df_clean[f'{role}_zip_code_prefix'].isin(zip_to_city.index),
    }).dropna(subset=['id']).drop_duplicates('id')
    changed = entity['kota_tertulis_standar'] != entity['kota_final']
    city_summary.append({
        'pihak': role,
        'jumlah': entity.shape[0],
        'nama_kota_unik_awal': entity['kota_tertulis'].nunique(),
        'nama_kota_unik_akhir': entity['kota_final'].nunique(),
        'kota_dari_kode_pos': entity['kode_pos_ada_di_kamus'].sum(),
        'kota_dari_tulisan': (~entity['kode_pos_ada_di_kamus']).sum(),
        'berbeda_dari_tulisan': changed.sum(),
    })
    if role == 'seller':
        seller_city_changes = entity.loc[changed, ['kota_tertulis', 'kota_final']]

city_summary = pd.DataFrame(city_summary).set_index('pihak')
city_summary['persen_berbeda'] = (city_summary['berbeda_dari_tulisan'] / city_summary['jumlah'] * 100).round(2)
city_summary
````

````python
# contoh kota seller yang berbeda antara tulisan seller dan kode posnya
seller_city_changes.drop_duplicates().sort_values('kota_final').head(25)
````

**Hasil:**
- **Seller:** 69 seller (±2,2%) memiliki kota yang berbeda antara tulisan dan kode posnya. Sebagian besar adalah tulisan yang salah: salah ketik (`garulhos` → `guarulhos`), singkatan (`sbc` → `sao bernardo do campo`), alamat email (→ `maringa`), dan nama state (`bahia` → `paulo afonso`). Sisanya adalah nama kota yang sah tetapi bukan lokasi kode posnya, misalnya `sao paulo` untuk seller di kota sekitar São Paulo. Kode `sp` juga tidak selalu berarti kota São Paulo: beberapa seller yang menulis `sp` ternyata berada di `braganca paulista` dan `aracatuba` menurut kode posnya.
- **Pelanggan:** hanya ±0,2% yang berbeda, umumnya variasi ejaan atau nama distrik di kota yang sama (`paraty`/`parati`, `embu`/`embu das artes`). Ini sejalan dengan dugaan bahwa kota pelanggan terisi otomatis dari kode pos.
- Variasi nama kota seller berkurang tajam, sedangkan kota pelanggan hampir tidak berubah.
- Pelanggan dan seller yang kode posnya tidak ada di tabel geolocation (±280 pelanggan dan 7 seller) tetap memakai nama kota yang ditulis. Nama kota 7 seller tersebut sudah valid (misalnya `curitiba`, `porto alegre`).

**Keterbatasan:** kode pos di dataset hanya berisi **5 digit awal**, sehingga satu kode pos bisa mencakup beberapa kota yang bertetangga. Pada kasus ini, kota yang dipakai adalah kota yang paling dominan pada kode pos tersebut. Contohnya seller yang menulis `sao jose` menjadi `florianopolis`, atau `contagem` menjadi `belo horizonte`. Perbedaan ini masih berada dalam **wilayah metropolitan dan state yang sama**, sehingga tidak memengaruhi analisis rute state maupun jarak. Untuk analisis rute kota, nama kota sebaiknya dibaca sebagai **wilayah kota** (kota beserta sekitarnya).

## 3.8 Outlier

Outlier dicek pada kolom numerik utama yang akan dipakai di analisis:
- `price` dan `freight_value` (level item): faktor nilai transaksi dan ongkos kirim pada Pilar 1
- **lama pengiriman** dan **selisih terhadap estimasi** (level pesanan): dasar analisis keterlambatan pada Pilar 1 dan 2

Lama pengiriman dihitung sementara untuk keperluan pengecekan. Kolom resminya akan dibuat pada tahap Feature Engineering.

Metode yang digunakan adalah **IQR**: nilai di atas Q3 + 1,5 × IQR dianggap outlier.

````python
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
df_item_level = df_clean[df_clean['product_id'].notna()]

df_delivered = df_clean.drop_duplicates('order_id')
df_delivered = df_delivered[
    (df_delivered['order_status'] == 'delivered') & df_delivered['order_delivered_customer_date'].notna()
]
delivery_days_check = (df_delivered['order_delivered_customer_date'] - df_delivered['order_purchase_timestamp']).dt.days
delay_days_check = (df_delivered['order_delivered_customer_date'].dt.normalize() - df_delivered['order_estimated_delivery_date']).dt.days

outlier_cols = {
    'price (BRL)': df_item_level['price'],
    'freight_value (BRL)': df_item_level['freight_value'],
    'lama pengiriman (hari)': delivery_days_check,
    'selisih vs estimasi (hari)': delay_days_check,
}

def iqr_summary(s):
    q1, q3 = s.quantile([0.25, 0.75])
    upper = q3 + 1.5 * (q3 - q1)
    return {
        'min': s.min(), 'Q1': q1, 'median': s.median(), 'Q3': q3,
        'batas_atas_IQR': upper, 'max': s.max(),
        'jumlah_outlier': (s > upper).sum(),
        'persen_outlier': round((s > upper).mean() * 100, 2),
    }

pd.DataFrame({name: iqr_summary(s) for name, s in outlier_cols.items()}).T.round(2)
````

````python
fig, axes = plt.subplots(2, 2, figsize=(14, 7))
for ax, (name, s) in zip(axes.flatten(), outlier_cols.items()):
    sns.boxplot(x=s, ax=ax, color='steelblue')
    ax.set_title(f'Sebaran {name}')
    ax.set_xlabel(name)
fig.suptitle('Keempat kolom miring ke kanan dengan banyak outlier, sehingga analisis memakai median dan pengelompokan', fontsize=12)
plt.tight_layout()
plt.show()
````

**Temuan:**
- Keempat kolom memiliki outlier di sisi kanan (distribusi **miring ke kanan**): sebagian besar nilai berkumpul di angka kecil, dengan ekor panjang ke nilai besar.
- **price**: median sekitar 75 BRL, tetapi ada item hingga 6.735 BRL.
- **freight_value**: median sekitar 16 BRL, tetapi ada ongkir hingga ratusan BRL.
- **Lama pengiriman**: median sekitar 10 hari, tetapi ada pesanan yang baru diterima setelah lebih dari 200 hari.
- **Selisih vs estimasi**: nilai negatif berarti tiba lebih cepat dari estimasi. Mayoritas pesanan tiba lebih cepat, tetapi ada pesanan yang terlambat hingga lebih dari 180 hari.

Untuk memastikan nilai ekstrem ini wajar (bukan salah input), berikut item dengan harga dan ongkir tertinggi:

````python
df_item_level.nlargest(5, 'price')[['product_category_name_english', 'price', 'freight_value', 'product_weight_g']]
````

````python
df_item_level.nlargest(5, 'freight_value')[['product_category_name_english', 'price', 'freight_value', 'product_weight_g', 'customer_state', 'seller_state']]
````

Selain melihat item termahal, dicek juga seberapa besar porsi outlier harga terhadap nilai transaksi, apakah ongkir tinggi sejalan dengan berat barang, dan bagaimana ulasan pada pesanan yang sangat lama sampai:

````python
price_q1, price_q3 = df_item_level['price'].quantile([0.25, 0.75])
price_outlier = df_item_level['price'] > price_q3 + 1.5 * (price_q3 - price_q1)
print(f"Kontribusi item outlier harga terhadap total nilai item: {df_item_level.loc[price_outlier, 'price'].sum() / df_item_level['price'].sum() * 100:.1f}%")
print('Korelasi ongkir dengan berat barang                    :', round(df_item_level['freight_value'].corr(df_item_level['product_weight_g']), 2))

# pesanan yang sangat lama sampai: apakah ini noise atau justru inti masalah?
long_delivery = delivery_days_check > 60
long_review = df_delivered.loc[long_delivery & df_delivered['review_score'].notna(), 'review_score']
print('Pesanan dengan lama kirim > 60 hari                    :', long_delivery.sum())
print(f'... yang mendapat ulasan bintang 1-2                   : {(long_review <= 2).mean() * 100:.1f}%')
````

**Temuan tambahan:**
- Item outlier harga hanya ±7% dari item, tetapi menyumbang **±35,6% dari total nilai item**. Menghapusnya berarti membuang lebih dari sepertiga nilai transaksi.
- Ongkir **berkorelasi positif dengan berat barang** (r = 0,61), sehingga ongkir yang tinggi wajar untuk barang berat.
- Ada **288 pesanan yang sampai lebih dari 60 hari**, dan **±65% di antaranya mendapat ulasan bintang 1–2**. Pesanan ekstrem ini justru hal yang ingin dijelaskan (keterlambatan parah → ulasan buruk), bukan kesalahan data.

Harga tertinggi berasal dari kategori yang wajar bernilai mahal (`housewares`, `computers`, `small_appliances`) dengan berat hingga 30 kg, dan ongkir tertinggi berasal dari produk berat (±15–30 kg) yang dikirim antar-state (misalnya PR → PI, SP → MT).

**Keputusan:** outlier **tidak dihapus**.

**Alasan:**
- **price & freight_value**: nilai tinggi menggambarkan produk mahal, berat, atau pengiriman jarak jauh yang memang ada. Sesuai batasan project, harga tidak dinilai "terlalu mahal"; yang dianalisis adalah hubungannya dengan ulasan.
- **Lama pengiriman & keterlambatan**: pesanan yang sangat terlambat justru merupakan **inti masalah bisnis** yang dianalisis. Menghapusnya akan menghilangkan bukti keterlambatan terparah dan membuat performa pengiriman terlihat lebih baik dari kenyataannya.

**Konsekuensi untuk analisis:** karena distribusinya miring dan banyak outlier, analisis berikutnya memakai:
- **median**, bukan rata-rata,
- **pengelompokan** (misalnya kelompok hari keterlambatan atau rentang harga),
- **uji statistik non-parametrik** (Mann-Whitney di 5.3.1), yang tidak mensyaratkan data berdistribusi normal.

````python
print('Awal periode:')
print(orders_per_month.head(5).to_string())
print('\nAkhir periode:')
print(orders_per_month.tail(4).to_string())
````

**Temuan:**
- Data di awal periode sangat sedikit: September 2016 hanya 4 pesanan, Oktober 2016 sekitar 300 pesanan, **November 2016 tidak ada pesanan sama sekali**, dan Desember 2016 hanya 1 pesanan.
- Data di akhir periode juga sangat sedikit: September dan Oktober 2018 hanya belasan pesanan.
- Bulan-bulan lainnya (Januari 2017 – Agustus 2018) berisi ribuan pesanan per bulan.

**Keputusan:** analisis dibatasi pada pesanan dengan tanggal pembelian **1 Januari 2017 – 31 Agustus 2018** (20 bulan analisis; data akhir Agustus 2018 bersifat parsial, lihat di bawah). Pesanan di luar periode ini dihapus dari `df_clean`.

**Alasan:**
- Bulan dengan jumlah pesanan sangat kecil menghasilkan persentase yang tidak stabil. Contohnya, jika satu bulan hanya berisi 4 pesanan dan 1 di antaranya terlambat, Late Delivery Rate bulan itu langsung 25% dan bisa terbaca sebagai lonjakan palsu.
- Bulan terakhir belum lengkap karena pengumpulan data berhenti sebelum sebagian besar pesanannya sempat dikirim dan diulas.
- Pesanan yang dikeluarkan hanya sekitar 0,35% dari total, sehingga tidak mengubah gambaran keseluruhan.

````python
rows_before = df_clean.shape[0]
orders_before = df_clean['order_id'].nunique()

df_clean = df_clean[
    (df_clean['order_purchase_timestamp'] >= '2017-01-01')
    & (df_clean['order_purchase_timestamp'] < '2018-09-01')
].reset_index(drop=True)

print('Pesanan dikeluarkan:', orders_before - df_clean['order_id'].nunique(),
      f"({(orders_before - df_clean['order_id'].nunique()) / orders_before * 100:.2f}%)")
print('Baris dikeluarkan  :', rows_before - df_clean.shape[0])
print('Periode sekarang   :', df_clean['order_purchase_timestamp'].min().date(), 's.d.', df_clean['order_purchase_timestamp'].max().date())
````

**Catatan untuk analisis bulan terakhir:** pesanan yang dibeli menjelang akhir periode (terutama Agustus 2018) mungkin belum sempat diterima ketika data dikumpulkan. Pesanan terlambat yang belum diterima tidak akan terhitung di Late Delivery Rate (karena hanya pesanan `delivered` yang dihitung), sehingga Late Delivery Rate bulan-bulan terakhir berpotensi sedikit lebih rendah dari kondisi sebenarnya.

````python
not_delivered_recent = (
    df_clean.drop_duplicates('order_id')
    .assign(bulan=lambda d: d['order_purchase_timestamp'].dt.to_period('M'))
    .query("bulan >= '2018-06'")
    .groupby('bulan')['order_status']
    .apply(lambda s: round((s != 'delivered').mean() * 100, 2))
    .rename('persen_belum_delivered')
)
not_delivered_recent.to_frame()
````

Selain itu, dicek apakah pesanan bulan terakhir (Agustus 2018) tercatat lengkap sampai akhir bulan, dengan membandingkan rata-rata jumlah pesanan per hari:

````python
august = df_clean.drop_duplicates('order_id')
august = august[august['order_purchase_timestamp'] >= '2018-08-01']
daily_august = august.groupby(august['order_purchase_timestamp'].dt.day)['order_id'].count()

print('Rata-rata pesanan per hari, 1–20 Agustus 2018 :', round(daily_august.loc[:20].mean(), 1))
print('Rata-rata pesanan per hari, 21–31 Agustus 2018:', round(daily_august.loc[21:].mean(), 1))
````

**Temuan:** rata-rata pesanan per hari turun dari **±278** (tanggal 1–20) menjadi **±86** (tanggal 21–31). Data akhir Agustus 2018 kemungkinan terpotong saat dataset diambil.

**Keputusan:** Agustus 2018 tetap dipakai.

**Alasan:** 20 hari pertamanya lengkap, dan menghapus satu bulan penuh akan membuang ±6.500 pesanan yang valid. Konsekuensinya dicatat sebagai batasan: angka Agustus 2018 pada analisis bulanan (5.2.5) perlu dibaca hati-hati.

## 3.10 Ringkasan Data Cleaning

````python
df_order_level = df_clean.drop_duplicates('order_id')

eligible_delivery = (df_order_level['order_status'] == 'delivered') & df_order_level['order_delivered_customer_date'].notna()
eligible_review = df_order_level['review_score'].notna() & ~df_order_level['is_review_anomaly']

# Pilar 3: hanya pesanan 1 seller, tanpa anomali urutan waktu dan anomali batas serah
sellers_per_order = df_clean.dropna(subset=['seller_id']).groupby('order_id')['seller_id'].nunique()
single_seller_orders = sellers_per_order[sellers_per_order == 1].index
limit_anomaly_orders = df_clean.loc[df_clean['is_limit_anomaly'], 'order_id'].unique()

df_single_seller = df_order_level[df_order_level['order_id'].isin(single_seller_orders)]
eligible_handover = (
    df_single_seller['order_delivered_carrier_date'].notna()
    & ~df_single_seller['is_time_anomaly']
    & ~df_single_seller['order_id'].isin(limit_anomaly_orders)
)

print('Jumlah baris (item)  :', df_clean.shape[0])
print('Jumlah kolom         :', df_clean.shape[1])
print('Jumlah pesanan       :', df_clean['order_id'].nunique())
print('Periode              :', df_clean['order_purchase_timestamp'].min().date(), 's.d.', df_clean['order_purchase_timestamp'].max().date())
print()
print('Data yang siap dipakai per metrik:')
print('- Low Review Rate          :', eligible_review.sum(), 'pesanan dengan ulasan valid')
print('- Late Delivery Rate       :', eligible_delivery.sum(), 'pesanan delivered')
print('- Seller Late Handover Rate:', eligible_handover.sum(), 'pesanan 1 seller')
````

| Langkah | Masalah | Tindakan | Alasan singkat |
|---|---|---|---|
| 3.2 | Kolom tanggal bertipe string; kolom bilangan bulat menjadi float | Cast ke `datetime` dan `Int64` | Durasi & keterlambatan bisa dihitung |
| 3.3 | Baris tergandakan akibat ulasan ganda | Simpan 1 ulasan terbaru per pesanan | 1 pesanan = 1 skor untuk Low Review Rate |
| 3.3 | `review_id` dipakai >1 pesanan (789 ID, pelanggan yang sama) | Tidak dipakai sebagai kunci deduplikasi | Bukan salah input; kunci tetap `order_id` + `order_item_id` |
| 3.4 | Tanggal proses pesanan kosong | Tidak diimputasi | Tahapan memang belum/tidak terjadi |
| 3.5 | Kategori kosong & tanpa terjemahan | Isi `'unknown'` & terjemahan manual | Pesanan tetap valid untuk analisis |
| 3.5 | Nama kategori salah ketik (4 nama) | Ejaan diperbaiki | Label grafik & dashboard rapi |
| 3.5 | Ulasan & pembayaran kosong | Dibiarkan | Tidak boleh dikarang; tidak memengaruhi metrik utama |
| 3.6 | Status vs tanggal tidak konsisten | Flag `is_status_inconsistent` | Dikecualikan dari analisis status |
| 3.6 | Urutan waktu terbalik | Flag `is_time_anomaly` | Dikecualikan dari handover & durasi per tahap |
| 3.6 | Batas serah ke kurir tahun 2020 | Flag `is_limit_anomaly` | Dikecualikan dari Seller Late Handover Rate |
| 3.6 | Ulasan dibuat sebelum pembelian | Flag `is_review_anomaly` | Dikecualikan dari Low Review Rate |
| 3.7 | Titik geolocation salah state / di luar Brasil | Dibuang sebelum median per kode pos dihitung ulang | Mencegah jarak yang tidak masuk akal |
| 3.7 | Koordinat kosong (kode pos tidak ada di geolocation) | Diperkirakan dari kode pos 4 digit yang sama; sisanya dibiarkan | Teruji akurat (median error ±2 km) |
| 3.7 | Kode pos, kota, dan state bisa saling bertentangan | Kode pos dijadikan acuan lokasi | Angka, tidak bisa salah ketik; 100% sesuai rentang resmi |
| 3.7 | State seller tidak sesuai kode pos (35 seller) | Diganti state menurut kode pos | Rute asal pengiriman akurat |
| 3.7 | Nama kota tidak seragam (typo, email, angka) | Diambil dari kode pos via tabel geolocation | Kota, state, dan koordinat dari sumber yang sama |
| 3.8 | Outlier harga, ongkir, durasi kirim | Dibiarkan | Nilai nyata; keterlambatan ekstrem adalah inti masalah |
| 3.9 | Periode data tipis di awal & akhir | Dibatasi Jan 2017 – Agu 2018 | Menghindari rate & tren palsu |
| 3.9 | Akhir Agustus 2018 tidak lengkap | Tetap dipakai, dicatat sebagai batasan | 20 hari pertama lengkap |

`df_clean` siap dipakai untuk tahap berikutnya, yaitu **Feature Engineering**. Di tahap ini dibuat kolom-kolom yang dibutuhkan analisis, misalnya lama pengiriman, status terlambat, status ulasan buruk, keterlambatan serah seller, dan jarak seller–pelanggan. Data juga disusun ulang menjadi level pesanan.

# 4. Feature Engineering

Data hasil cleaning (`df_clean`) belum bisa langsung dipakai untuk menjawab pertanyaan bisnis, karena metrik yang dibutuhkan (misalnya apakah pesanan terlambat atau apakah seller terlambat menyerahkan barang) belum tersedia sebagai kolom. Pada tahap ini dibuat kolom-kolom turunan (*features*) yang dibutuhkan oleh ketiga pilar analisis.

Urutan feature engineering mengikuti hierarki metrik di Business Understanding:
1. **hasil akhir** (ulasan buruk),
2. **faktor operasional utama** (keterlambatan pengiriman),
3. **sumber keterlambatan** (tahap seller vs pengiriman),
4. **faktor pendukung** (komposisi pesanan, geografi, dan waktu).

| Langkah | Fokus | Pilar / Metrik |
|---|---|---|
| 4.1 | Menentukan unit analisis (tabel level pesanan) | Semua pilar |
| 4.2 | Kepuasan pelanggan (ulasan buruk) | Low Review Rate |
| 4.3 | Ketepatan pengiriman | Late Delivery Rate (Pilar 1 & 2) |
| 4.4 | Durasi per tahap proses | Pilar 3 |
| 4.5 | Ketepatan seller menyerahkan barang | Seller Late Handover Rate (Pilar 3) |
| 4.6 | Komposisi pesanan | Pilar 1 |
| 4.7 | Geografi pengiriman | Pilar 2 |
| 4.8 | Waktu & finalisasi data | Pilar 2 & dashboard |

## 4.1 Menentukan Unit Analisis

Sesuai Business Understanding, **unit analisis utama adalah pesanan (`order_id`)**. Ulasan, status, tanggal pengiriman, dan estimasi melekat pada pesanan, bukan pada item. Sementara itu, `df_clean` masih berlevel item (1 baris = 1 item), sehingga pesanan dengan banyak item muncul di beberapa baris.

Jika metrik dihitung langsung dari `df_clean`, pesanan dengan banyak item akan terhitung berkali-kali. Contohnya, pesanan berisi 3 item yang terlambat akan terhitung sebagai 3 keterlambatan. Karena itu dibuat tabel baru **`df_order`** dengan **1 baris = 1 pesanan**.

Langkah pertama adalah memilih kolom yang memang bernilai sama untuk seluruh item dalam satu pesanan (kolom level pesanan):

````python
order_cols = [
    # identitas & status pesanan
    'order_id', 'customer_id', 'customer_unique_id', 'order_status',
    # tanggal proses pesanan
    'order_purchase_timestamp', 'order_approved_at', 'order_delivered_carrier_date',
    'order_delivered_customer_date', 'order_estimated_delivery_date',
    # lokasi pelanggan
    'customer_zip_code_prefix', 'customer_city', 'customer_state', 'customer_lat', 'customer_lng',
    # ulasan
    'review_id', 'review_score', 'review_comment_title', 'review_comment_message',
    'review_creation_date', 'review_answer_timestamp',
    # pembayaran (sudah diagregasi per pesanan saat merge)
    'payment_count', 'payment_types', 'payment_type_main', 'payment_installments_max', 'payment_value_total',
    # flag hasil cleaning yang berlaku per pesanan
    'is_status_inconsistent', 'is_time_anomaly', 'is_review_anomaly',
]
````

Sebelum tabel dibuat, perlu dipastikan bahwa kolom-kolom tersebut **benar-benar bernilai sama** di semua baris item dalam satu pesanan. Jika ada kolom yang nilainya berbeda antar item, mengambil satu baris per pesanan akan menghilangkan informasi.

````python
# jumlah nilai unik setiap kolom di dalam satu pesanan (nilai kosong ikut dihitung)
values_per_order = df_clean.groupby('order_id')[order_cols[1:]].nunique(dropna=False)

max_values_per_order = values_per_order.max()
print('Kolom yang nilainya berbeda di dalam satu pesanan:')
print(max_values_per_order[max_values_per_order > 1] if (max_values_per_order > 1).any() else '- tidak ada -')
````

Seluruh kolom tersebut bernilai sama dalam setiap pesanan, sehingga aman untuk mengambil **satu baris per pesanan**.

Kolom level item (misalnya `product_id`, `seller_id`, `price`, `freight_value`, `shipping_limit_date`) **tidak** dimasukkan ke `df_order`, karena satu pesanan bisa berisi beberapa produk dan beberapa seller. Informasi tersebut akan diringkas (diagregasi) per pesanan pada langkah 4.6, dan analisis seller dilakukan pada tabel terpisah di langkah 4.5.

````python
df_order = (
    df_clean[order_cols]
    .drop_duplicates('order_id')
    .reset_index(drop=True)
)

print('df_clean (level item)   :', df_clean.shape, '|', df_clean['order_id'].nunique(), 'pesanan')
print('df_order (level pesanan):', df_order.shape)
print('order_id unik di df_order:', df_order['order_id'].is_unique)
````

````python
df_order.head()
````

**Hasil:** `df_order` berisi **99.092 baris**, tepat sama dengan jumlah pesanan pada `df_clean`, dan setiap `order_id` hanya muncul satu kali. Seluruh feature pada langkah berikutnya ditambahkan ke tabel ini.

## 4.2 Kepuasan Pelanggan (Outcome)

Ulasan buruk adalah **hasil akhir (outcome)** yang ingin ditekan dalam project ini. Sesuai Business Understanding, ulasan dengan skor **1–2** digunakan sebagai indikator *customer dissatisfaction*, dan metrik utamanya adalah:

$$Low\ Review\ Rate = \frac{\text{Jumlah pesanan dengan skor ulasan 1–2}}{\text{Jumlah pesanan yang memiliki ulasan valid}} \times 100\%$$

Dua kolom dibuat:
- **`is_valid_review`**: pesanan memiliki ulasan **dan** ulasannya tidak termasuk anomali (ulasan yang dibuat sebelum pembelian, langkah 3.6). Kolom ini menjadi **penyebut** Low Review Rate.
- **`is_low_review`**: skor ulasan 1 atau 2. Hanya diisi untuk ulasan valid; selain itu dibiarkan kosong agar tidak ikut terhitung.

````python
df_order['is_valid_review'] = df_order['review_score'].notna() & ~df_order['is_review_anomaly']

df_order['is_low_review'] = (df_order['review_score'] <= 2).astype('boolean')
df_order.loc[~df_order['is_valid_review'], 'is_low_review'] = pd.NA

print('Pesanan dengan ulasan valid:', df_order['is_valid_review'].sum())
print('Pesanan tanpa ulasan valid :', (~df_order['is_valid_review']).sum())
print('Pesanan dengan low review  :', df_order['is_low_review'].sum())
````

### Distribusi Skor Ulasan

````python
score_dist = (
    df_order.loc[df_order['is_valid_review'], 'review_score']
    .value_counts().sort_index()
    .to_frame('jumlah_pesanan')
)
score_dist['persen'] = (score_dist['jumlah_pesanan'] / score_dist['jumlah_pesanan'].sum() * 100).round(2)
score_dist
````

````python
colors = ['#d62728' if score <= 2 else '#bbbbbb' for score in score_dist.index]

ax = score_dist['persen'].plot(kind='bar', color=colors, figsize=(8, 4), rot=0)
ax.set_title(f"Pola J: bintang 1 ({score_dist.loc[1, 'persen']:.1f}%) jauh lebih banyak daripada bintang 2 ({score_dist.loc[2, 'persen']:.1f}%)\n(distribusi skor ulasan, Jan 2017 – Agu 2018)")
ax.set_xlabel('Skor Ulasan (bintang)')
ax.set_ylabel('Persentase Pesanan (%)')
for i, value in enumerate(score_dist['persen']):
    ax.text(i, value + 1, f'{value:.1f}%', ha='center')
ax.set_ylim(0, score_dist['persen'].max() + 8)
plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **Merah** = ulasan buruk (bintang 1–2), yang dihitung dalam Low Review Rate
- **Pola J** = ulasan menumpuk di bintang 5 dan bintang 1, sedangkan bintang 2–3 sedikit, sehingga bentuknya mirip huruf J
- **Ulasan valid** = pesanan yang memiliki skor ulasan dan tanggal ulasannya tidak sebelum tanggal pembelian

````python
low_review_rate = df_order['is_low_review'].mean() * 100
print(f'Low Review Rate Olist: {low_review_rate:.2f}%')
````

**Temuan:**
- Mayoritas pesanan mendapat ulasan bintang 5 (±58%), dan rata-rata kepuasan pelanggan Olist tergolong tinggi.
- Namun **±14,6% pesanan** berakhir dengan ulasan buruk (bintang 1–2), atau sekitar **1 dari 7 pesanan**. Inilah angka acuan (*baseline*) Low Review Rate yang akan dibandingkan di setiap analisis berikutnya.
- Pola distribusinya berbentuk huruf **"J"**: setelah turun dari bintang 5 ke bintang 2, jumlahnya naik lagi di bintang 1. Bintang 1 (±11%) jauh lebih banyak daripada bintang 2 (±3%). Artinya pelanggan yang kecewa cenderung **langsung memberi skor terendah**, bukan setengah-setengah.

````python
# kelompok ulasan untuk visualisasi pada tahap analisis
df_order['review_group'] = pd.cut(
    df_order['review_score'], bins=[0, 2, 3, 5], labels=['Buruk (1-2)', 'Netral (3)', 'Baik (4-5)']
)
df_order.loc[~df_order['is_valid_review'], 'review_group'] = np.nan

df_order['review_group'].value_counts(dropna=False)
````

**Hasil langkah 4.2:** tiga kolom baru ditambahkan ke `df_order`:

| Kolom | Isi | Kegunaan |
|---|---|---|
| `is_valid_review` | Ulasan ada dan bukan anomali | Penyebut Low Review Rate |
| `is_low_review` | Skor 1–2 (kosong jika ulasan tidak valid) | Pembilang Low Review Rate; hasil akhir yang dibandingkan di Pilar 1 |
| `review_group` | Buruk (1–2) / Netral (3) / Baik (4–5) | Visualisasi |

**Baseline Low Review Rate Olist: 14,61%.**

## 4.3 Ketepatan Pengiriman

Keterlambatan pengiriman adalah faktor operasional utama yang akan diuji hubungannya dengan ulasan buruk (Pilar 1) dan dipetakan berdasarkan wilayah serta waktu (Pilar 2). Sesuai Business Understanding:

$$Late\ Delivery\ Rate = \frac{\text{Pesanan delivered yang diterima setelah estimated delivery date}}{\text{Total pesanan delivered}} \times 100\%$$

Kolom yang dibuat:

| Kolom | Rumus | Arti |
|---|---|---|
| `is_delivered` | status `delivered` dan tanggal diterima tersedia | Pesanan yang dapat dinilai ketepatan pengirimannya (penyebut) |
| `delivery_days` | tanggal diterima − tanggal pembelian | Lama pelanggan menunggu barang (hari) |
| `estimated_days` | tanggal estimasi − tanggal pembelian | Lama pengiriman yang **dijanjikan** saat checkout (hari) |
| `delay_days` | tanggal diterima − tanggal estimasi | Selisih terhadap janji; negatif = lebih cepat, positif = terlambat |
| `is_late` | `delay_days` > 0 | Pesanan terlambat (pembilang) |
| `delay_group` | pengelompokan `delay_days` | Tingkat keparahan keterlambatan |

````python
df_order['is_delivered'] = (
    (df_order['order_status'] == 'delivered') & df_order['order_delivered_customer_date'].notna()
)

delivered = df_order['is_delivered']
seconds_per_day = 24 * 60 * 60

df_order['delivery_days'] = (
    (df_order['order_delivered_customer_date'] - df_order['order_purchase_timestamp']).dt.total_seconds() / seconds_per_day
).where(delivered).round(2)

df_order['estimated_days'] = (
    (df_order['order_estimated_delivery_date'] - df_order['order_purchase_timestamp']).dt.total_seconds() / seconds_per_day
).round(2)

print('Pesanan delivered:', delivered.sum(), 'dari', df_order.shape[0])
````

Untuk menentukan terlambat atau tidak, perbandingan dilakukan pada **level tanggal**, bukan jam. Alasannya, `order_estimated_delivery_date` hanya berisi tanggal (jamnya selalu 00:00). Jika dibandingkan dengan jam, pesanan yang tiba **pada hari estimasi** (misal pukul 15.00) akan salah terhitung terlambat. Dengan perbandingan tanggal, pesanan yang tiba pada hari estimasi dianggap **tepat waktu**, sesuai definisi "melewati tanggal estimasi".

````python
df_order['delay_days'] = (
    df_order['order_delivered_customer_date'].dt.normalize() - df_order['order_estimated_delivery_date']
).dt.days.where(delivered).astype('Int64')

df_order['is_late'] = (df_order['delay_days'] > 0).astype('boolean')
df_order.loc[~delivered, 'is_late'] = pd.NA

late_delivery_rate = df_order['is_late'].mean() * 100
print(f'Pesanan terlambat     : {df_order["is_late"].sum()}')
print(f'Late Delivery Rate    : {late_delivery_rate:.2f}%')
````

### Pengecekan Hasil

Sebelum dipakai, nilai kolom baru dicek apakah masuk akal: lama pengiriman tidak boleh negatif, dan estimasi harus selalu setelah pembelian.

````python
df_order.loc[delivered, ['delivery_days', 'estimated_days', 'delay_days']].describe(
    percentiles=[0.05, 0.25, 0.5, 0.75, 0.95]
).round(1)
````

Nilai minimum `delivery_days` tidak negatif dan `estimated_days` selalu positif, sehingga kolom sudah sesuai. Terlihat pula bahwa **median estimasi pengiriman (±23 hari) jauh lebih lama daripada median pengiriman aktual (±10 hari)**: Olist cenderung memberikan estimasi yang longgar. Hal ini menarik, karena berarti pesanan yang tetap terlambat sudah melewati estimasi yang longgar tersebut.

### Kelompok Keterlambatan

`delay_days` dikelompokkan agar tingkat keparahan keterlambatan dapat dibandingkan (misalnya apakah telat 1–3 hari berdampak sama dengan telat lebih dari 2 minggu). Batas kelompok dipilih mengikuti satuan waktu yang mudah dipahami stakeholder: hari, minggu, dan dua minggu.

````python
delay_bins = [-np.inf, -8, -1, 0, 3, 7, 14, np.inf]
delay_labels = [
    'Lebih cepat >7 hari', 'Lebih cepat 1-7 hari', 'Tepat di hari estimasi',
    'Telat 1-3 hari', 'Telat 4-7 hari', 'Telat 8-14 hari', 'Telat >14 hari',
]

df_order['delay_group'] = pd.cut(df_order['delay_days'].astype('float'), bins=delay_bins, labels=delay_labels)

delay_group_dist = df_order['delay_group'].value_counts().reindex(delay_labels).to_frame('jumlah_pesanan')
delay_group_dist['persen'] = (delay_group_dist['jumlah_pesanan'] / delivered.sum() * 100).round(2)
delay_group_dist
````

**Hasil langkah 4.3:**

| Kolom | Keterangan |
|---|---|
| `is_delivered` | Penyebut Late Delivery Rate |
| `delivery_days`, `estimated_days` | Lama pengiriman aktual vs yang dijanjikan |
| `delay_days`, `is_late` | Selisih terhadap estimasi & status terlambat |
| `delay_group` | 7 kelompok tingkat keterlambatan |

**Baseline Late Delivery Rate Olist: 6,79%.** Sebagian besar pesanan justru tiba lebih dari seminggu sebelum tanggal estimasi, sedangkan pesanan terlambat tersebar mulai dari 1–3 hari hingga lebih dari 2 minggu.

## 4.4 Durasi per Tahap Proses

Lama pengiriman total (`delivery_days`) belum menjelaskan **di tahap mana** waktu paling banyak habis. Untuk menjawab pertanyaan Pilar 3 (apakah keterlambatan sudah terjadi di sisi seller atau baru terjadi saat transit), lama pengiriman dipecah menjadi tiga tahap:

```
checkout ──(1) approval──▶ pembayaran disetujui ──(2) proses seller──▶ diserahkan ke kurir ──(3) transit kurir──▶ diterima pelanggan
```

| Kolom | Rumus | Pihak yang bertanggung jawab |
|---|---|---|
| `approval_days` | `order_approved_at` − `order_purchase_timestamp` | Sistem pembayaran |
| `seller_processing_days` | `order_delivered_carrier_date` − `order_approved_at` | Seller |
| `carrier_transit_days` | `order_delivered_customer_date` − `order_delivered_carrier_date` | Kurir / logistik |

Durasi per tahap hanya dihitung untuk pesanan yang:
- berstatus delivered dan memiliki tanggal diterima (`is_delivered`),
- memiliki tanggal approval dan tanggal serah ke kurir,
- **tidak** memiliki anomali urutan waktu (`is_time_anomaly`, langkah 3.6), karena tanggal yang urutannya terbalik akan menghasilkan durasi negatif.

````python
valid_stage = (
    df_order['is_delivered']
    & df_order['order_approved_at'].notna()
    & df_order['order_delivered_carrier_date'].notna()
    & ~df_order['is_time_anomaly']
)
df_order['is_valid_stage'] = valid_stage

df_order['approval_days'] = (
    (df_order['order_approved_at'] - df_order['order_purchase_timestamp']).dt.total_seconds() / seconds_per_day
).where(valid_stage).round(2)

df_order['seller_processing_days'] = (
    (df_order['order_delivered_carrier_date'] - df_order['order_approved_at']).dt.total_seconds() / seconds_per_day
).where(valid_stage).round(2)

df_order['carrier_transit_days'] = (
    (df_order['order_delivered_customer_date'] - df_order['order_delivered_carrier_date']).dt.total_seconds() / seconds_per_day
).where(valid_stage).round(2)

print('Pesanan delivered                       :', df_order['is_delivered'].sum())
print('Pesanan dengan durasi per tahap yang valid:', valid_stage.sum())
print('Dikecualikan                             :', df_order['is_delivered'].sum() - valid_stage.sum())
````

### Pengecekan Hasil

Ada dua hal yang dicek:
1. Tidak boleh ada durasi negatif.
2. Ketiga tahap harus **menjumlah tepat** menjadi lama pengiriman total (`delivery_days`). Jika tidak, berarti ada kesalahan rumus.

````python
stage_cols = ['approval_days', 'seller_processing_days', 'carrier_transit_days']

df_order.loc[valid_stage, stage_cols + ['delivery_days']].describe(percentiles=[0.25, 0.5, 0.75, 0.95]).round(2)
````

````python
stage_total = df_order.loc[valid_stage, stage_cols].sum(axis=1)
difference = (stage_total - df_order.loc[valid_stage, 'delivery_days']).abs()

print('Durasi negatif per tahap :', (df_order.loc[valid_stage, stage_cols] < 0).sum().to_dict())
print('Selisih maksimal (hari) antara jumlah 3 tahap dan delivery_days:', round(difference.max(), 2))
````

Tidak ada durasi negatif, dan jumlah ketiga tahap sama dengan lama pengiriman total (selisih hanya karena pembulatan 2 desimal). Kolom sudah benar.

### Porsi Waktu per Tahap

````python
stage_median = df_order.loc[valid_stage, stage_cols].median()
stage_share = (df_order.loc[valid_stage, stage_cols].sum() / df_order.loc[valid_stage, stage_cols].sum().sum() * 100)

pd.DataFrame({
    'median_hari': stage_median.round(2),
    'porsi_dari_total_waktu_%': stage_share.round(1),
})
````

**Temuan awal:** dari **total waktu pengiriman seluruh pesanan** (jumlah hari semua pesanan), sekitar **74% habis di tahap transit kurir** (median per pesanan ±7 hari). Proses seller menyumbang ±22% (median ±1,8 hari), dan approval pembayaran hanya ±3% (median kurang dari 1 jam). Apakah pola ini berbeda antara pesanan yang tepat waktu dan yang terlambat akan dianalisis pada Pilar 3.

**Hasil langkah 4.4:** empat kolom baru ditambahkan ke `df_order`: `is_valid_stage` (penanda pesanan yang durasi per tahapnya bisa dipakai), `approval_days`, `seller_processing_days`, dan `carrier_transit_days`.

## 4.5 Ketepatan Seller Menyerahkan Barang (Handover)

Pilar 3 menanyakan seberapa sering keterlambatan **sudah terjadi di sisi seller**, sebelum barang diterima kurir. Setiap item memiliki `shipping_limit_date`, yaitu batas waktu seller menyerahkan barang ke kurir. Sesuai Business Understanding:

$$Seller\ Late\ Handover\ Rate = \frac{\text{Pesanan yang diserahkan ke kurir setelah shipping\_limit\_date}}{\text{Pesanan 1 seller yang memiliki tanggal serah ke kurir}} \times 100\%$$

**Kenapa hanya pesanan 1 seller?** Tanggal serah ke kurir (`order_delivered_carrier_date`) hanya dicatat **satu kali per pesanan**. Pada pesanan dengan beberapa seller, setiap seller mengirim barangnya sendiri, tetapi hanya ada satu tanggal serah, sehingga tidak dapat diketahui seller mana yang terlambat.

Langkah pertama adalah menandai pesanan 1 seller dan mengambil `seller_id` serta `shipping_limit_date`-nya.

````python
df_items_with_seller = df_clean.dropna(subset=['seller_id'])

sellers_per_order = df_items_with_seller.groupby('order_id')['seller_id'].nunique()
df_order['n_sellers'] = df_order['order_id'].map(sellers_per_order).fillna(0).astype(int)
df_order['is_single_seller'] = df_order['n_sellers'] == 1

print('Pesanan tanpa item (0 seller):', (df_order['n_sellers'] == 0).sum())
print('Pesanan 1 seller             :', df_order['is_single_seller'].sum())
print('Pesanan >1 seller            :', (df_order['n_sellers'] > 1).sum())
````

Pada pesanan 1 seller yang berisi beberapa item, setiap item memiliki `shipping_limit_date` sendiri. Sebelum mengambil satu nilai per pesanan, perlu dipastikan bahwa batas serah tersebut **sama untuk semua item** dalam pesanan.

````python
single_seller_items = df_items_with_seller[df_items_with_seller['order_id'].isin(df_order.loc[df_order['is_single_seller'], 'order_id'])]

limits_per_order = single_seller_items.groupby('order_id')['shipping_limit_date'].nunique()
print('Pesanan 1 seller dengan batas serah berbeda antar item:', (limits_per_order > 1).sum())
````

Batas serah selalu sama untuk semua item dalam pesanan 1 seller, sehingga aman diambil satu nilai per pesanan.

````python
single_seller_info = single_seller_items.groupby('order_id').agg(
    seller_id=('seller_id', 'first'),
    shipping_limit_date=('shipping_limit_date', 'first'),
    is_limit_anomaly=('is_limit_anomaly', 'any'),
)

df_order = df_order.merge(single_seller_info, on='order_id', how='left')
df_order['is_limit_anomaly'] = df_order['is_limit_anomaly'].fillna(False).astype(bool)

print('df_order:', df_order.shape)
print('Pesanan 1 seller yang memiliki seller_id:', df_order['seller_id'].notna().sum())
````

### Data yang Valid untuk Handover

Sebuah pesanan hanya dipakai untuk menghitung handover jika:
- merupakan pesanan **1 seller**,
- memiliki tanggal serah ke kurir,
- **tidak** memiliki anomali urutan waktu (`is_time_anomaly`, langkah 3.6), karena tanggal serah ke kurirnya tidak dapat dipercaya,
- **tidak** memiliki batas serah yang salah input di tahun 2020 (`is_limit_anomaly`, langkah 3.6).

````python
df_order['is_valid_handover'] = (
    df_order['is_single_seller']
    & df_order['order_delivered_carrier_date'].notna()
    & ~df_order['is_time_anomaly']
    & ~df_order['is_limit_anomaly']
)

valid_handover = df_order['is_valid_handover']

df_order['handover_delay_days'] = (
    (df_order['order_delivered_carrier_date'] - df_order['shipping_limit_date']).dt.total_seconds() / seconds_per_day
).where(valid_handover).round(2)

df_order['is_late_handover'] = (df_order['handover_delay_days'] > 0).astype('boolean')
df_order.loc[~valid_handover, 'is_late_handover'] = pd.NA

late_handover_rate = df_order['is_late_handover'].mean() * 100
print('Pesanan valid untuk handover :', valid_handover.sum())
print('Pesanan telat serah ke kurir :', df_order['is_late_handover'].sum())
print(f'Seller Late Handover Rate    : {late_handover_rate:.2f}%')
````

Berbeda dengan keterlambatan pengiriman (langkah 4.3), perbandingan handover dilakukan **sampai level jam**, karena `shipping_limit_date` dan `order_delivered_carrier_date` sama-sama memiliki keterangan jam.

### Pengecekan Hasil

````python
df_order.loc[valid_handover, 'handover_delay_days'].describe(percentiles=[0.05, 0.25, 0.5, 0.75, 0.95]).round(2).to_frame()
````

**Temuan:**
- Pada umumnya seller menyerahkan barang **±3 hari sebelum** batas serah (median `handover_delay_days` = −3,2 hari), sehingga sebagian besar seller patuh terhadap batas waktu.
- Namun **8.618 dari 94.774 pesanan (9,09%)** diserahkan ke kurir setelah batas serah. Ini adalah baseline **Seller Late Handover Rate**.
- Sebagian keterlambatan serah cukup ekstrem: ada pesanan yang baru diserahkan lebih dari 100 hari setelah batas serah.

**Hasil langkah 4.5:** kolom baru di `df_order`:

| Kolom | Keterangan |
|---|---|
| `n_sellers`, `is_single_seller` | Jumlah seller per pesanan & penanda pesanan 1 seller |
| `seller_id`, `shipping_limit_date` | Seller dan batas serah (hanya untuk pesanan 1 seller) |
| `is_limit_anomaly` | Flag batas serah salah input (dari langkah 3.6) |
| `is_valid_handover` | Penyebut Seller Late Handover Rate |
| `handover_delay_days` | Serah ke kurir − batas serah (positif = telat serah) |
| `is_late_handover` | Pembilang Seller Late Handover Rate |

Karena `seller_id` sudah tersedia di `df_order` untuk pesanan 1 seller, analisis per seller pada Pilar 3 dapat dilakukan langsung dari tabel ini.

## 4.6 Komposisi Pesanan

Pilar 1 membandingkan Low Review Rate berdasarkan karakteristik transaksi: **jumlah barang, jumlah seller, nilai pesanan, ongkos kirim, dan metode pembayaran utama**. Informasi barang masih berada di level item (`df_clean`), sehingga perlu diringkas menjadi satu nilai per pesanan.

| Kolom | Rumus | Arti |
|---|---|---|
| `n_items` | jumlah baris item | Jumlah barang dalam pesanan |
| `n_products` | jumlah `product_id` unik | Jumlah jenis produk berbeda |
| `order_value` | total `price` | Nilai barang (BRL), tanpa ongkir |
| `freight_total` | total `freight_value` | Total ongkos kirim (BRL) |
| `freight_ratio` | `freight_total` ÷ (`order_value` + `freight_total`) | Porsi ongkir dari total yang dibayar |
| `is_multi_seller` | `n_sellers` > 1 | Pesanan berisi barang dari lebih dari 1 seller |

````python
order_composition = df_items_with_seller.groupby('order_id').agg(
    n_items=('order_item_id', 'count'),
    n_products=('product_id', 'nunique'),
    order_value=('price', 'sum'),
    freight_total=('freight_value', 'sum'),
)

df_order = df_order.merge(order_composition, on='order_id', how='left')
df_order['n_items'] = df_order['n_items'].fillna(0).astype(int)
df_order['n_products'] = df_order['n_products'].fillna(0).astype(int)

df_order['freight_ratio'] = (df_order['freight_total'] / (df_order['order_value'] + df_order['freight_total'])).round(4)
df_order['is_multi_seller'] = df_order['n_sellers'] > 1

print('df_order:', df_order.shape)
print('Pesanan tanpa item (nilai pesanan kosong):', df_order['order_value'].isna().sum())
````

Pesanan tanpa item tidak diberi nilai pesanan 0, melainkan dibiarkan kosong. Nilai 0 akan terbaca sebagai "pesanan gratis", padahal pesanan tersebut memang tidak memiliki barang (umumnya berstatus `unavailable` atau `canceled`).

### Pengecekan Hasil

Untuk memastikan agregasi benar, nilai barang + ongkir dibandingkan dengan total pembayaran dari tabel pembayaran. Keduanya berasal dari tabel yang berbeda, sehingga jika hasilnya cocok, agregasi dapat dipercaya.

````python
has_item_and_payment = df_order['order_value'].notna() & df_order['payment_value_total'].notna()
payment_gap = (df_order['payment_value_total'] - (df_order['order_value'] + df_order['freight_total']))[has_item_and_payment]

print(f'Pesanan yang total pembayarannya cocok (selisih <= 1 BRL): {(payment_gap.abs() <= 1).mean() * 100:.2f}%')
print('Pesanan dengan selisih > 1 BRL                           :', (payment_gap.abs() > 1).sum())
````

Lebih dari 99% pesanan memiliki total pembayaran yang sama dengan nilai barang + ongkir, sehingga agregasi sudah benar. Selisih pada sebagian kecil pesanan kemungkinan berasal dari biaya cicilan atau penyesuaian lain yang tidak tercatat di dataset, dan tidak diubah karena tidak memengaruhi analisis.

````python
df_order.loc[df_order['n_items'] > 0, ['n_items', 'n_products', 'n_sellers', 'order_value', 'freight_total', 'freight_ratio']].describe(
    percentiles=[0.25, 0.5, 0.75, 0.95]
).round(2)
````

### Pengelompokan untuk Analisis Pilar 1

Agar setiap karakteristik dapat dibandingkan Low Review Rate-nya, nilai numerik dikelompokkan:
- **Jumlah barang**: 1, 2, 3, dan 4 barang atau lebih.
- **Jumlah seller**: tanpa barang, 1 seller, lebih dari 1 seller.
- **Nilai pesanan & ongkir**: dibagi menjadi **4 kuartil** (masing-masing ±25% pesanan). Kuartil dipilih karena distribusinya sangat miring (langkah 3.8), sehingga kelompok berbasis kuartil menjamin setiap kelompok memiliki jumlah pesanan yang seimbang.

````python
df_order['item_group'] = pd.cut(
    df_order['n_items'], bins=[0, 1, 2, 3, np.inf], labels=['1 barang', '2 barang', '3 barang', '4+ barang']
)

df_order['seller_group'] = np.select(
    [df_order['n_sellers'] == 0, df_order['n_sellers'] == 1],
    ['Tanpa barang', '1 seller'],
    default='>1 seller',
)

df_order['order_value_group'] = pd.qcut(df_order['order_value'], q=4,
                                         labels=['Q1 (terendah)', 'Q2', 'Q3', 'Q4 (tertinggi)'])
df_order['freight_group'] = pd.qcut(df_order['freight_total'], q=4,
                                     labels=['Q1 (terendah)', 'Q2', 'Q3', 'Q4 (tertinggi)'])

for col in ['item_group', 'seller_group', 'order_value_group', 'freight_group']:
    print(df_order[col].value_counts(dropna=False).sort_index().to_string(), '\n')
````

````python
# batas nilai setiap kuartil (BRL), untuk keperluan interpretasi
pd.DataFrame({
    'batas_nilai_pesanan': df_order['order_value'].quantile([0, 0.25, 0.5, 0.75, 1]).round(2).values,
    'batas_ongkir': df_order['freight_total'].quantile([0, 0.25, 0.5, 0.75, 1]).round(2).values,
}, index=['min', 'Q1', 'median', 'Q3', 'max'])
````

#### Kenapa Kuartil, Bukan Nilai Asli, Interval Tetap, atau Desil?

Ada beberapa cara untuk membandingkan Low Review Rate berdasarkan ongkir dan nilai pesanan. Pilihan cara pengelompokan ditentukan **dari distribusi datanya saja**, sebelum melihat Low Review Rate, agar kelompok tidak dipilih karena menghasilkan pola tertentu. Perbandingannya dicek pada ongkir:

````python
freight = df_order.loc[df_order['n_items'] > 0, 'freight_total']
orders_per_value = freight.value_counts()

print('Jumlah nilai ongkir unik          :', freight.nunique())
print('Median jumlah pesanan per nilai   :', orders_per_value.median())
print(f'Nilai ongkir dengan < 30 pesanan  : {(orders_per_value < 30).mean() * 100:.1f}%')
print(f'Skewness ongkir                   : {freight.skew():.2f}')

order_value = df_order.loc[df_order['n_items'] > 0, 'order_value']
print('Jumlah nilai pesanan unik         :', order_value.nunique())
print(f'Skewness nilai pesanan            : {order_value.skew():.2f}')

# opsi interval tetap: lebar kelompok sama (BRL 10), jumlah pesanannya?
fixed_width = pd.cut(freight, bins=[0, 10, 20, 30, 40, 50, 100, np.inf], include_lowest=True).value_counts().sort_index()
print('\nInterval tetap (BRL):')
print(fixed_width.to_string())

# opsi desil: jumlah pesanan seimbang, tetapi seberapa lebar rentang setiap kelompok?
decile_limits = freight.quantile(np.arange(0, 1.01, 0.1)).round(2).values
pd.DataFrame({
    'batas_bawah_BRL': decile_limits[:-1],
    'batas_atas_BRL': decile_limits[1:],
    'lebar_rentang_BRL': np.diff(decile_limits).round(2),
}, index=[f'Desil {i}' for i in range(1, 11)])
````

| Opsi | Hasil pengecekan | Keputusan |
|---|---|---|
| **Nilai asli** (tanpa dikelompokkan) | Ada **±8.000 nilai ongkir unik**, rata-rata hanya **±3 pesanan per nilai**, dan ±91% nilai memiliki kurang dari 30 pesanan. Rate dari 3 pesanan tidak stabil: 1 ulasan buruk saja sudah menjadi 33%. | Tidak dipakai |
| **Interval tetap** (per BRL 10) | Distribusi sangat miring (skewness ±12). Kelompok BRL 10–20 berisi **±53 ribu pesanan (±54%)**, sedangkan kelompok > BRL 100 hanya **±1.100 pesanan**. Rate kelompok kecil jauh kurang stabil dibanding kelompok besar, sehingga perbandingannya tidak adil. | Tidak dipakai |
| **Desil** (10 kelompok) | Jumlah pesanan seimbang, tetapi batas kelompok di bagian tengah **sangat rapat**: Desil 3–6 masing-masing hanya selebar **±BRL 1,1–1,7** (misalnya Desil 4: BRL 14,53–15,65). Perbedaan ongkir sekecil itu tidak bermakna bagi pelanggan maupun bisnis, dan 10 batang lebih sulit dibaca tanpa menambah informasi yang bisa ditindaklanjuti. | Tidak dipakai |
| **Kuartil** (4 kelompok) | Setiap kelompok berisi **±24.500 pesanan (25%)**, sehingga rate-nya sama stabil. Batasnya adalah **Q1, median, dan Q3** (BRL 13,85 / 17,17 / 24,03), statistik yang sama dengan `describe()`, boxplot, dan metode IQR di langkah 3.8. Interpretasinya mudah: *"25% pesanan dengan ongkir termahal (di atas ±BRL 24)"*. Jumlah per kelompok juga cukup besar untuk uji Chi-Square. | **Dipakai** |

Nilai pesanan memiliki karakteristik yang sama (±7.800 nilai unik, skewness ±9,8), sehingga dikelompokkan dengan cara yang sama. Kuartil tertinggi (Q4) juga dipakai sebagai "kelompok berisiko" pada ranking faktor di 5.1.7.

### Jarak Seller–Pelanggan

Jarak dihitung dengan **rumus Haversine**, yaitu rumus jarak terpendek antara dua titik di permukaan bumi (mengikuti lengkung bumi) berdasarkan latitude dan longitude:

$$d = 2R \cdot \arcsin\left(\sqrt{\sin^2\left(\frac{\Delta\varphi}{2}\right) + \cos\varphi_1 \cos\varphi_2 \sin^2\left(\frac{\Delta\lambda}{2}\right)}\right)$$

dengan $R$ = jari-jari bumi (6.371 km), $\varphi$ = latitude, dan $\lambda$ = longitude (dalam radian).

Catatan: ini adalah **jarak garis lurus**, bukan jarak jalan yang sebenarnya ditempuh kurir. Koordinat juga berasal dari titik tengah kode pos (langkah merge 8), bukan alamat persis. Karena itu jarak digunakan sebagai **gambaran kasar jauh-dekatnya pengiriman**, sesuai asumsi di Business Understanding.

````python
def haversine_km(lat1, lng1, lat2, lng2):
    lat1, lng1, lat2, lng2 = map(np.radians, [lat1, lng1, lat2, lng2])
    a = np.sin((lat2 - lat1) / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin((lng2 - lng1) / 2) ** 2
    return 2 * 6371 * np.arcsin(np.sqrt(a))

df_order['distance_km'] = haversine_km(
    df_order['seller_lat'], df_order['seller_lng'], df_order['customer_lat'], df_order['customer_lng']
).round(1)

print('Pesanan dengan jarak:', df_order['distance_km'].notna().sum())
print('Tanpa jarak (koordinat kosong atau multi-seller/tanpa barang):', df_order['distance_km'].isna().sum())
````

### Pengecekan Hasil

Jarak yang benar seharusnya: tidak negatif, tidak lebih dari ukuran wilayah Brasil (±4.000 km dari ujung ke ujung), dan pengiriman sama-state seharusnya jauh lebih dekat daripada beda-state.

````python
distance_check = df_order.groupby('same_state')['distance_km'].describe(percentiles=[0.5, 0.95]).round(1)
distance_check.index = distance_check.index.map({False: 'Beda-state', True: 'Sama-state'})
distance_check
````

Sisanya umumnya wajar karena terjadi di perbatasan state, misalnya dari Brasília (DF) ke kota-kota Goiás di sekitarnya, atau antar kota perbatasan MG–SP. Jumlahnya sangat kecil (±0,02% dari pengiriman beda-state) sehingga dibiarkan.

### Kelompok Jarak

| Kelompok | Gambaran |

````python
df_order['distance_group'] = pd.cut(
    df_order['distance_km'],
    bins=[-0.1, 100, 300, 700, 1500, np.inf],
    labels=['< 100 km', '100-300 km', '300-700 km', '700-1.500 km', '> 1.500 km'],
)

distance_group_dist = df_order['distance_group'].value_counts().sort_index().to_frame('jumlah_pesanan')
distance_group_dist['persen'] = (distance_group_dist['jumlah_pesanan'] / distance_group_dist['jumlah_pesanan'].sum() * 100).round(2)
distance_group_dist
````

| **Late Delivery Rate** | Pesanan delivered | ±1 dari 15 pesanan tiba setelah tanggal yang dijanjikan |
| **Seller Late Handover Rate** | Pesanan 1 seller dengan data handover valid | ±1 dari 11 pesanan diserahkan seller ke kurir setelah batas waktu |

# 5. Exploratory Data Analysis

## 5.1 Pilar 1 — WHAT: Faktor Apa yang Paling Berkaitan dengan Ulasan Buruk?

**Pertanyaan:** karakteristik pesanan apa yang paling membedakan pesanan dengan ulasan buruk (bintang 1–2) dari pesanan lainnya?

**Metode:**
1. Untuk setiap faktor, pesanan dikelompokkan, lalu **Low Review Rate** setiap kelompok dibandingkan dengan **baseline Olist (14,61%)**.
2. Besarnya perbedaan diukur dengan dua angka:
   - **selisih** terhadap baseline (dalam poin persen),
   - **rasio** terhadap baseline (misal rasio 2 = ulasan buruk 2× lebih sering daripada rata-rata).
3. Hubungan setiap faktor diuji dengan **uji Chi-Square** (kedua variabel bersifat kategorik), dengan tingkat signifikansi **α = 0,05**:
   - **Ho:** faktor tersebut **tidak berkaitan** dengan ulasan buruk (proporsi ulasan buruk sama di semua kelompok).
   - **Ha:** faktor tersebut **berkaitan** dengan ulasan buruk.
   - Jika p-value < 0,05 → Tolak Ho.

**Catatan penting:** karena jumlah data sangat besar (±98 ribu pesanan), perbedaan yang kecil pun akan menghasilkan p-value yang sangat kecil. Karena itu uji Chi-Square dipakai untuk memastikan hubungan **bukan kebetulan**, sedangkan **seberapa penting** suatu faktor dinilai dari **besar selisih/rasionya** dan **jumlah pesanan** yang terdampak.

Analisis hanya memakai pesanan dengan ulasan valid (`is_valid_review`). Pesanan dengan status yang bertentangan dengan tanggalnya (`is_status_inconsistent`, langkah 3.6a) **dikecualikan**, agar kelompok "sampai" dan "tidak sampai" terbaca sama di seluruh Pilar 1 (5.1.1, 5.1.3, dan 5.1.7).

**Alur Pilar 1:**

| Bagian | Pertanyaan |
|---|---|
| 5.1.1 | Apakah pesanan sampai ke pelanggan? |
| 5.1.2 | Apakah pesanan sampai tepat waktu, dan seberapa parah keterlambatannya? |
| 5.1.3 | Dari mana asal ulasan buruk (telat, tidak sampai, atau tepat waktu)? |
| 5.1.4–5.1.6 | Faktor pesanan lainnya di luar keterlambatan: kompleksitas, nilai & ongkir, metode pembayaran |
| 5.1.7 | Dari data pesanan, faktor mana yang paling penting? (ranking) |
| 5.1.8 | Apa yang sebenarnya dikeluhkan pelanggan? Kenapa pesanan tepat waktu tetap mendapat ulasan buruk? (isi komentar ulasan, breakdown pesanan tepat waktu, kategori produk) |

Prinsipnya: **semua faktor diuji lebih dulu dengan data pesanan** (5.1.1–5.1.6), lalu **diurutkan** (5.1.7). Setelah itu, **isi komentar pelanggan** dipakai sebagai konfirmasi dan untuk menjelaskan ulasan buruk yang tidak terjelaskan oleh data pesanan, terutama pada pesanan tepat waktu (5.1.8). Pilar 1 ditutup dengan kesimpulan.

Metode Low Review Rate + uji Chi-Square di atas dipakai di 5.1.1, 5.1.2, dan 5.1.4–5.1.6. Bagian 5.1.3 dan 5.1.8 melihat **porsi** (berapa bagian ulasan buruk yang berasal dari setiap sumber atau keluhan), sehingga metodenya dijelaskan di bagian masing-masing.

````python
from scipy.stats import chi2_contingency

# pesanan dengan status bertentangan dengan tanggalnya (3.6a) dikecualikan
excluded_status = df_order['is_valid_review'] & df_order['is_status_inconsistent']
df_review = df_order[df_order['is_valid_review'] & ~df_order['is_status_inconsistent']].copy()
print(f'Dikecualikan (status tidak konsisten): {excluded_status.sum()}')
baseline_low_review = df_review['is_low_review'].mean() * 100
print(f'Pesanan dengan ulasan valid: {df_review.shape[0]}')
print(f'Baseline Low Review Rate   : {baseline_low_review:.2f}%')
````

````python
review_share = df_review['is_low_review'].astype(bool).map({
    True: 'Ulasan buruk (bintang 1-2)', False: 'Bukan ulasan buruk (bintang 3-5)'
}).value_counts()
review_pct = review_share / review_share.sum() * 100

fig, ax = plt.subplots(figsize=(9, 2.6))
ax.barh(review_share.index, review_pct, color=['#9e9e9e' if 'Bukan' in label else '#d62728' for label in review_share.index])
for i, (pct, count) in enumerate(zip(review_pct, review_share)):
    ax.text(pct + 1, i, f'{pct:.1f}%  ({count:,} pesanan)'.replace(',', '.'), va='center')
ax.set_xlim(0, 115)
ax.set_xlabel('Persentase pesanan berulasan (%)')
ax.set_title(f'±1 dari 7 pesanan berulasan mendapat ulasan buruk ({baseline_low_review:.2f}%)\n(proporsi ulasan buruk dari seluruh pesanan berulasan)')
ax.invert_yaxis()
plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **Ulasan buruk** = ulasan bintang 1–2; bintang 3–5 = bukan ulasan buruk
- **Low Review Rate** = jumlah ulasan buruk ÷ jumlah pesanan berulasan valid
- **Pesanan berulasan valid** = pesanan dengan skor ulasan yang tanggal ulasannya tidak sebelum tanggal pembelian (pesanan berstatus tidak konsisten dikecualikan)

**Insight:** dari **98.280 pesanan** yang memiliki ulasan valid (9 pesanan dengan status tidak konsisten sudah dikecualikan), **14.357 pesanan (14,61%)** mendapat ulasan buruk (bintang 1–2), atau **±1 dari 7 pesanan**. Sisanya (85,4%) memberi bintang 3–5. Angka 14,61% inilah **baseline** yang menjadi pembanding seluruh faktor di Pilar 1.

(Bintang 3 dimasukkan ke kelompok "bukan ulasan buruk" karena definisi Low Review Rate hanya menghitung bintang 1–2, lihat bagian 1.5.)

Tiga fungsi bantu dibuat agar setiap faktor dianalisis dengan cara yang sama:

````python
def low_review_by(data, group_col):
    # Low Review Rate per kelompok, beserta selisih dan rasio terhadap baseline
    summary = data.groupby(group_col, observed=True).agg(
        jumlah_pesanan=('order_id', 'count'),
        jumlah_low_review=('is_low_review', 'sum'),
    )
    summary['low_review_rate_%'] = (summary['jumlah_low_review'] / summary['jumlah_pesanan'] * 100).round(2)
    summary['selisih_vs_baseline_pp'] = (summary['low_review_rate_%'] - baseline_low_review).round(2)
    summary['rasio_vs_baseline'] = (summary['low_review_rate_%'] / baseline_low_review).round(2)
    return summary


def chi_square_test(data, group_col, alpha=0.05):
    # uji Chi-Square: apakah kelompok berkaitan dengan ulasan buruk
    table = pd.crosstab(data[group_col], data['is_low_review'].astype(bool))
    stat, pvalue, dof, expected = chi2_contingency(table)
    pvalue_text = '< 1e-300' if pvalue == 0 else f'= {pvalue:.3g}'  # p-value 0 berarti terlalu kecil untuk ditampilkan
    if pvalue < alpha:
        print(f'p-value {pvalue_text} < {alpha}. Tolak Ho: {group_col} berkaitan dengan ulasan buruk.')
    else:
        print(f'p-value {pvalue_text} >= {alpha}. Gagal menolak Ho: {group_col} tidak terbukti berkaitan dengan ulasan buruk.')
    return pvalue


def plot_low_review(summary, title, xlabel):
    # bar chart horizontal Low Review Rate per kelompok, dengan garis baseline
    fig, ax = plt.subplots(figsize=(9, 0.6 * len(summary) + 1.5))
    colors = ['#d62728' if rate > baseline_low_review else '#9e9e9e' for rate in summary['low_review_rate_%']]
    ax.barh(summary.index.astype(str), summary['low_review_rate_%'], color=colors)
    ax.axvline(baseline_low_review, color='black', linestyle='--', linewidth=1)
    ax.text(baseline_low_review, len(summary) - 0.4, f' baseline {baseline_low_review:.1f}%', va='bottom')
    for i, (rate, n) in enumerate(zip(summary['low_review_rate_%'], summary['jumlah_pesanan'])):
        ax.text(rate + 0.8, i, f'{rate:.1f}%  (n={n:,})'.replace(',', '.'), va='center')
    ax.set_title(title)
    ax.set_xlabel('Low Review Rate (%)')
    ax.set_ylabel(xlabel)
    ax.set_xlim(0, max(summary['low_review_rate_%'].max() * 1.25, baseline_low_review * 1.5))
    ax.invert_yaxis()
    plt.tight_layout()
    plt.show()
````

### 5.1.1 Status Pesanan: Sampai vs Tidak Sampai

````python
low_review_by(df_review, 'order_status').sort_values('jumlah_pesanan', ascending=False)
````

Status selain `delivered` jumlahnya kecil-kecil, sehingga digabung menjadi satu kelompok **"Tidak sampai"** (pesanan yang dikirim tapi belum diterima, dibatalkan, tidak tersedia, atau masih diproses).

````python
df_review['delivery_status'] = np.where(df_review['order_status'] == 'delivered', 'Sampai (delivered)', 'Tidak sampai')

status_summary = low_review_by(df_review, 'delivery_status')
status_summary
````

````python
plot_low_review(status_summary, f"Pesanan yang tidak sampai {status_summary.loc['Tidak sampai', 'low_review_rate_%']:.1f}% mendapat ulasan buruk, "
                f"{status_summary.loc['Tidak sampai', 'rasio_vs_baseline']:.1f}× baseline\n(Low Review Rate: sampai vs tidak sampai)", 'Status Pesanan')
chi_square_test(df_review, 'delivery_status');
````

**Keterangan istilah:**

- **Low Review Rate** = jumlah ulasan buruk (bintang 1–2) ÷ jumlah pesanan berulasan valid
- **Baseline (garis putus-putus)** = Low Review Rate seluruh pesanan berulasan valid (14,61%)
- **Merah / abu-abu** = kelompok dengan Low Review Rate di atas / di bawah baseline; n = jumlah pesanan berulasan valid
- **Tidak sampai** = pesanan yang tidak tercatat sampai ke pelanggan (status selain delivered, misalnya canceled, unavailable, atau shipped)

**Insight:**
- Pesanan yang **tidak sampai** ke pelanggan memiliki Low Review Rate **78,3%**, atau **5,4× baseline**. Hampir 4 dari 5 pelanggan yang pesanannya tidak sampai memberi bintang 1–2. Ini kelompok dengan tingkat ulasan buruk tertinggi.
- Namun jumlahnya kecil: hanya **2.723 pesanan (±2,8%)** dari pesanan berulasan.
- Pesanan yang sampai memiliki Low Review Rate **12,8%**, sedikit di bawah baseline.
- Uji Chi-Square: **Tolak Ho**, status pesanan berkaitan dengan ulasan buruk.

**Implikasi:** pesanan yang tidak sampai hampir pasti berakhir dengan ulasan buruk, sehingga pencegahan pembatalan dan pesanan yang tidak tersedia tetap penting. Tetapi karena volumenya kecil, sumber ulasan buruk terbesar perlu dicari di dalam **pesanan yang sampai**, yaitu dari ketepatan waktunya.

### 5.1.2 Ketepatan Waktu: Telat vs Tepat Waktu

Analisis ini hanya memakai pesanan yang sampai (`is_delivered`), karena ketepatan waktu hanya bisa dinilai jika barang sudah diterima.

````python
df_review_delivered = df_review[df_review['is_delivered']].copy()
df_review_delivered['delivery_timeliness'] = np.where(df_review_delivered['is_late'], 'Telat', 'Tepat waktu')

timeliness_summary = low_review_by(df_review_delivered, 'delivery_timeliness')
timeliness_summary
````

````python
plot_low_review(timeliness_summary, f"Pesanan telat {timeliness_summary.loc['Telat', 'low_review_rate_%']:.1f}% mendapat ulasan buruk, "
                f"{timeliness_summary.loc['Telat', 'low_review_rate_%'] / timeliness_summary.loc['Tepat waktu', 'low_review_rate_%']:.1f}× pesanan tepat waktu\n(Low Review Rate: telat vs tepat waktu)", 'Ketepatan Waktu')
chi_square_test(df_review_delivered, 'delivery_timeliness');
````

**Keterangan istilah:**

- **Low Review Rate** = jumlah ulasan buruk (bintang 1–2) ÷ jumlah pesanan berulasan valid
- **Baseline (garis putus-putus)** = Low Review Rate seluruh pesanan berulasan valid (14,61%)
- **Merah / abu-abu** = kelompok dengan Low Review Rate di atas / di bawah baseline; n = jumlah pesanan berulasan valid
- **Telat** = pesanan delivered yang sampai setelah tanggal estimasi (dibandingkan per tanggal)
- **Tepat waktu** = sampai pada atau sebelum tanggal estimasi

**Insight:**
- Pesanan yang **telat** memiliki Low Review Rate **62,4%**, sedangkan pesanan **tepat waktu** hanya **9,2%**. Artinya, ulasan buruk pada pesanan telat muncul **±6,8× lebih sering** daripada pesanan tepat waktu.
- Uji Chi-Square: **Tolak Ho**, ketepatan waktu berkaitan dengan ulasan buruk.

Apakah semakin lama telat, semakin buruk ulasannya? Hal ini dicek dengan kelompok keterlambatan (`delay_group`, langkah 4.3):

````python
delay_summary = low_review_by(df_review_delivered, 'delay_group')
delay_summary
````

````python
fig, ax = plt.subplots(figsize=(11, 4.5))
colors = ['#d62728' if rate > baseline_low_review else '#9e9e9e' for rate in delay_summary['low_review_rate_%']]
ax.bar(delay_summary.index.astype(str), delay_summary['low_review_rate_%'], color=colors)
ax.axhline(baseline_low_review, color='black', linestyle='--', linewidth=1)
ax.text(len(delay_summary) - 0.5, baseline_low_review + 1.5, f'baseline {baseline_low_review:.1f}%', ha='right')
for i, rate in enumerate(delay_summary['low_review_rate_%']):
    ax.text(i, rate + 1.5, f'{rate:.1f}%', ha='center')
ax.set_title(f"Begitu melewati estimasi, ulasan buruk melonjak: {delay_summary.loc['Telat 1-3 hari', 'low_review_rate_%']:.0f}% (telat 1–3 hari) "
             f"hingga {delay_summary.loc['Telat 8-14 hari', 'low_review_rate_%']:.0f}% (telat 8–14 hari)\n(Low Review Rate per tingkat keterlambatan)")
ax.set_xlabel('Waktu Tiba Dibanding Tanggal Estimasi')
ax.set_ylabel('Low Review Rate (%)')
ax.set_ylim(0, delay_summary['low_review_rate_%'].max() + 10)
plt.xticks(rotation=20, ha='right')
plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **Low Review Rate** = jumlah ulasan buruk (bintang 1–2) ÷ jumlah pesanan berulasan valid
- **Baseline (garis putus-putus)** = Low Review Rate seluruh pesanan berulasan valid (14,61%); merah = di atas baseline
- **Telat / lebih cepat X hari** = selisih tanggal sampai dengan tanggal estimasi (positif = telat); sampai tepat di hari estimasi = tepat waktu

**Insight:**
- Selama pesanan tiba **sebelum atau tepat pada** tanggal estimasi, Low Review Rate tetap di bawah baseline (±9–12%).
- Begitu pesanan **melewati** tanggal estimasi, Low Review Rate langsung naik: **32% saat telat 1–3 hari**, **68% saat telat 4–7 hari**, dan sekitar **80% saat telat lebih dari 1 minggu**.
- Polanya bertingkat: semakin lama keterlambatan, semakin tinggi proporsi ulasan buruk. Pola ini memperkuat bahwa hubungan keterlambatan dengan ulasan buruk bukan kebetulan.
- Titik kritisnya ada pada **minggu pertama keterlambatan**: kenaikan terbesar terjadi dari telat 1–3 hari (32%) ke telat 4–7 hari (68%).

**Catatan interpretasi:** sesuai batasan project (1.9), survei kepuasan pada pesanan telat sering dikirim sebelum barang sampai. Karena itu, skor pada pesanan telat lebih mencerminkan pengalaman menunggu daripada kualitas produk.

**Cek catatan di atas dengan data.** Tanggal survei dikirim (`review_creation_date`) dibandingkan dengan tanggal barang diterima. Perbandingan dilakukan per tanggal, sama seperti definisi keterlambatan:

````python
df_review_delivered['survei_sebelum_sampai'] = (
    df_review_delivered['review_creation_date'].dt.normalize()
    < df_review_delivered['order_delivered_customer_date'].dt.normalize()
)
(df_review_delivered.groupby('delivery_timeliness')['survei_sebelum_sampai'].mean() * 100).round(1).to_frame('persen_survei_sebelum_sampai')
````

Agar urutan kejadiannya terlihat, dibuat **timeline** per kelompok pesanan: kapan pesanan dibuat (hari 0), kapan tanggal estimasi, kapan survei dikirim, kapan ulasan dijawab, dan kapan barang benar-benar sampai. Setiap titik adalah **median** hari sejak pembelian, karena durasi pengiriman miring ke kanan (4.1).

````python
timeline_order = ['Tepat waktu', 'Telat, survei setelah sampai', 'Telat, survei sebelum sampai']
df_review_delivered['kelompok_timeline'] = np.select(
    [~df_review_delivered['is_late'].astype(bool), df_review_delivered['survei_sebelum_sampai']],
    timeline_order[:1] + timeline_order[2:],
    default=timeline_order[1],
)

# median hari sejak pembelian untuk setiap kejadian
milestones = {
    'Estimasi sampai': 'order_estimated_delivery_date',
    'Survei dikirim': 'review_creation_date',
    'Ulasan dijawab': 'review_answer_timestamp',
    'Barang sampai': 'order_delivered_customer_date',
}
purchase_time = df_review_delivered['order_purchase_timestamp']
days_since_purchase = pd.DataFrame({
    name: (df_review_delivered[col] - purchase_time).dt.total_seconds() / seconds_per_day
    for name, col in milestones.items()
})
timeline = days_since_purchase.groupby(df_review_delivered['kelompok_timeline']).median().round(1).reindex(timeline_order)

group_stats = df_review_delivered.groupby('kelompok_timeline').agg(
    jumlah_pesanan=('order_id', 'count'),
    low_review_rate_pct=('is_low_review', 'mean'),
    median_telat_hari=('delay_days', 'median'),
)
timeline['median_telat_hari'] = group_stats['median_telat_hari']
timeline['jumlah_pesanan'] = group_stats['jumlah_pesanan']
timeline['low_review_rate_%'] = (group_stats['low_review_rate_pct'] * 100).round(1)
timeline
````

````python
milestone_style = {
    # nama kejadian: (warna, bentuk titik, posisi angka)
    'Estimasi sampai': ('#7f7f7f', 's', -16),
    'Survei dikirim': ('#ff7f0e', 'v', 10),
    'Ulasan dijawab': ('#1f77b4', 'D', -16),
    'Barang sampai': ('#d62728', 'o', 22),
}

fig, ax = plt.subplots(figsize=(12, 4.8))
for y, group in enumerate(timeline_order):
    row = timeline.loc[group]
    ax.plot([0, row[list(milestones)].max()], [y, y], color='#d9d9d9', linewidth=3, zorder=1)
    ax.scatter(0, y, color='black', marker='|', s=300, zorder=2, label='Pesanan dibuat (hari 0)' if y == 0 else None)
    for name, (color, marker, offset) in milestone_style.items():
        ax.scatter(row[name], y, color=color, marker=marker, s=110, zorder=3, label=name if y == 0 else None)
        ax.annotate(f'{row[name]:.0f}', (row[name], y), xytext=(0, offset), textcoords='offset points',
                    ha='center', fontsize=9, color=color)

labels = []
for group in timeline_order:
    n_orders = f"{timeline.loc[group, 'jumlah_pesanan']:,}".replace(',', '.')
    labels.append(f"{group}\n(n={n_orders}; Low Review Rate {timeline.loc[group, 'low_review_rate_%']:.1f}%)")
ax.set_yticks(range(len(timeline_order)), labels)
ax.set_ylim(-0.7, len(timeline_order) - 0.4)
ax.invert_yaxis()
ax.set_xlim(-1, timeline[list(milestones)].max().max() + 3)
ax.set_xlabel('Hari sejak pesanan dibuat (median)')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.17), ncol=5, frameon=False)

late_before = timeline.loc['Telat, survei sebelum sampai']
ax.set_title(f"Pesanan telat yang disurvei sebelum sampai: ulasan dijawab hari ke-{late_before['Ulasan dijawab']:.0f}, "
             f"barang baru sampai hari ke-{late_before['Barang sampai']:.0f} (Low Review Rate {late_before['low_review_rate_%']:.1f}%)\n"
             '(timeline median sejak pesanan dibuat: estimasi, survei, ulasan, dan barang sampai)')
plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **Estimasi sampai** = tanggal sampai yang dijanjikan kepada pelanggan saat pembelian
- **Survei dikirim** = tanggal survei kepuasan dikirim ke pelanggan (`review_creation_date`)
- **Ulasan dijawab** = waktu pelanggan mengisi survei (`review_answer_timestamp`)
- **Telat, survei sebelum sampai** = pesanan telat yang survei-nya dikirim sebelum tanggal barang diterima (dibandingkan per tanggal)
- **Telat, survei setelah sampai** = pesanan telat yang survei-nya dikirim pada atau setelah tanggal barang diterima
- **Angka di titik** = median hari sejak pesanan dibuat; setiap titik dihitung terpisah, sehingga selisih antar titik hanya gambaran kasar
- **Low Review Rate** = jumlah ulasan buruk (bintang 1–2) ÷ jumlah pesanan berulasan valid

**Insight:**
- **Pesanan tepat waktu:** barang sampai ±hari ke-10, survei dikirim di hari yang sama atau sesudahnya, dan ulasan dijawab ±hari ke-13. Pelanggan menilai **setelah** menerima barang.
- **Pesanan telat, survei sebelum sampai (4.749 pesanan, 74,5% pesanan telat):**
  - survei dikirim ±hari ke-25, ±2 hari setelah tanggal estimasi (hari ke-23) lewat;
  - ulasan dijawab ±hari ke-28;
  - barang baru sampai ±hari ke-35, artinya pelanggan menilai **±7 hari sebelum barangnya diterima**;
  - Low Review Rate **75,2%**.
- **Pesanan telat, survei setelah sampai (1.629 pesanan):** telatnya tipis (sampai ±hari ke-24, estimasi hari ke-21), dan Low Review Rate-nya **25,1%**, ±3× lebih rendah.
- **Catatan:** kelompok "survei sebelum sampai" juga telatnya jauh lebih lama (median telat ±9 vs ±2 hari setelah estimasi). Jadi selisih 75,2% vs 25,1% berkaitan dengan **dua hal sekaligus**: lamanya telat dan barang belum diterima saat menilai. Keduanya tidak dapat dipisahkan dengan data ini.

**Temuan:** pada **±74,5% pesanan telat**, survei sudah dikirim sebelum barangnya diterima, dibanding hanya **±0,2%** pada pesanan tepat waktu. Survei dikirim setelah tanggal estimasi lewat, sehingga pelanggan pesanan telat umumnya menilai saat barang belum diterima.

**Implikasi:** catatan interpretasi di atas didukung data. Jika keterlambatan tidak terhindarkan, pelanggan perlu **dikabari sebelum tanggal estimasi lewat**, karena pada saat itulah survei dikirim.

### 5.1.3 Dari Mana Asal Ulasan Buruk?

Low Review Rate menunjukkan **seberapa berisiko** suatu kelompok. Namun untuk prioritas bisnis, perlu dilihat juga **berapa banyak** ulasan buruk yang disumbang setiap kelompok, karena kelompok dengan rate tinggi belum tentu menyumbang ulasan buruk terbanyak.

````python
df_low_review = df_review[df_review['is_low_review'].astype(bool)].copy()
df_low_review['sumber'] = np.select(
    [~df_low_review['is_delivered'], df_low_review['is_late'].fillna(False).astype(bool)],
    ['Tidak sampai', 'Sampai, tapi telat'],
    default='Sampai tepat waktu',
)

source_summary = df_low_review['sumber'].value_counts().to_frame('jumlah_low_review')
source_summary['porsi_%'] = (source_summary['jumlah_low_review'] / source_summary['jumlah_low_review'].sum() * 100).round(1)

order_share = pd.Series({
    'Sampai tepat waktu': (df_review['is_delivered'] & ~df_review['is_late'].fillna(False).astype(bool)).mean() * 100,
    'Sampai, tapi telat': df_review['is_late'].fillna(False).astype(bool).mean() * 100,
    'Tidak sampai': (~df_review['is_delivered']).mean() * 100,
})
source_summary['porsi_dari_seluruh_pesanan_%'] = order_share.round(1)
source_summary
````

````python
plot_data = source_summary.loc[['Sampai tepat waktu', 'Sampai, tapi telat', 'Tidak sampai'], ['porsi_dari_seluruh_pesanan_%', 'porsi_%']]
late_row = source_summary.loc['Sampai, tapi telat']

fig, ax = plt.subplots(figsize=(10, 3.8))
y = np.arange(len(plot_data))
ax.barh(y - 0.2, plot_data['porsi_dari_seluruh_pesanan_%'], height=0.4, color='#bbbbbb', label='Porsi dari seluruh pesanan berulasan')
ax.barh(y + 0.2, plot_data['porsi_%'], height=0.4, color='#d62728', label='Porsi dari seluruh ulasan buruk')
for i, (orders, low) in enumerate(zip(plot_data['porsi_dari_seluruh_pesanan_%'], plot_data['porsi_%'])):
    ax.text(orders + 1, i - 0.2, f'{orders:.1f}%', va='center', fontsize=9)
    ax.text(low + 1, i + 0.2, f'{low:.1f}%', va='center', fontsize=9)
ax.set_yticks(y, plot_data.index)
ax.invert_yaxis()
ax.set_xlim(0, 105)
ax.set_xlabel('Persentase (%)')
ax.set_title(f"Pesanan telat hanya {late_row['porsi_dari_seluruh_pesanan_%']:.1f}% pesanan, tetapi menyumbang {late_row['porsi_%']:.1f}% ulasan buruk\n"
             '(porsi pesanan vs porsi ulasan buruk per sumber)')
ax.legend(loc='lower right')
plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **Ulasan buruk** = ulasan bintang 1–2
- **Sampai tepat waktu** = pesanan delivered yang sampai pada atau sebelum tanggal estimasi
- **Sampai, tapi telat** = pesanan delivered yang sampai setelah tanggal estimasi
- **Tidak sampai** = pesanan yang tidak tercatat sampai ke pelanggan (status selain delivered, misalnya canceled, unavailable, atau shipped)

**Insight:**
- Pesanan **telat** hanya **±6,5%** dari seluruh pesanan berulasan, tetapi menyumbang **±27,7%** dari seluruh ulasan buruk, lebih dari 4× porsinya.
- Pesanan **tidak sampai** (±2,8% pesanan) menyumbang **±14,8%** ulasan buruk.
- Pesanan yang **sampai tepat waktu** tetap menjadi sumber terbesar (**±57,4%**), karena jumlah pesanannya sangat besar (±91%), meskipun rate-nya rendah (9,2%).

**Implikasi:**
- Keterlambatan adalah faktor operasional dengan **risiko tertinggi per pesanan sekaligus jumlah pesanan yang cukup besar**, sehingga layak menjadi fokus utama Pilar 2 dan Pilar 3.
- Lebih dari separuh ulasan buruk berasal dari pesanan yang tepat waktu, sehingga **ada faktor lain di luar keterlambatan** yang juga perlu diperiksa. Karena itu, faktor-faktor pesanan lainnya diuji satu per satu di 5.1.4–5.1.6 (kompleksitas, nilai & ongkir, metode pembayaran). Seluruh faktor kemudian diurutkan di 5.1.7, lalu sisa ulasan buruk pada pesanan tepat waktu dijelaskan dengan **isi keluhan pelanggan** di 5.1.8.

(Grafik membandingkan porsi pesanan dengan porsi ulasan buruk setiap sumber, sehingga terlihat kelompok mana yang menyumbang ulasan buruk melebihi porsinya.)

### 5.1.4 Kompleksitas Pesanan: Jumlah Seller dan Jumlah Barang

````python
plot_low_review(seller_summary, f"Pesanan multi-seller {seller_summary.loc['>1 seller', 'low_review_rate_%']:.1f}% mendapat ulasan buruk, "
                f"{seller_summary.loc['>1 seller', 'low_review_rate_%'] / seller_summary.loc['1 seller', 'low_review_rate_%']:.1f}× pesanan 1 seller\n(Low Review Rate: 1 seller vs lebih dari 1 seller)", 'Jumlah Seller')
chi_square_test(df_review_items, 'seller_group');
````

- **Baseline (garis putus-putus)** = Low Review Rate seluruh pesanan berulasan valid (14,61%)
- **Merah / abu-abu** = kelompok dengan Low Review Rate di atas / di bawah baseline; n = jumlah pesanan berulasan valid
- **>1 seller (multi-seller)** = satu pesanan berisi barang dari lebih dari 1 seller

````python
item_summary = low_review_by(df_review_items, 'item_group')
plot_low_review(item_summary, f"Semakin banyak barang, semakin sering ulasan buruk: {item_summary.loc['1 barang', 'low_review_rate_%']:.1f}% (1 barang) "
                f"→ {item_summary.loc['4+ barang', 'low_review_rate_%']:.1f}% (4+ barang)\n(Low Review Rate per jumlah barang)", 'Jumlah Barang')
chi_square_test(df_review_items, 'item_group');
item_summary
````

**Keterangan istilah:**

- **Low Review Rate** = jumlah ulasan buruk (bintang 1–2) ÷ jumlah pesanan berulasan valid
- **Baseline (garis putus-putus)** = Low Review Rate seluruh pesanan berulasan valid (14,61%)
- **Merah / abu-abu** = kelompok dengan Low Review Rate di atas / di bawah baseline; n = jumlah pesanan berulasan valid

Pesanan multi-seller dan pesanan berisi banyak barang terlihat lebih sering mendapat ulasan buruk. Namun perlu dicek: **apakah ini hanya karena pesanan tersebut lebih sering telat?** Jika ya, maka faktor ini sebenarnya hanya "menumpang" efek keterlambatan.

Untuk mengujinya, perbandingan diulang **hanya pada pesanan yang sampai tepat waktu**, sehingga pengaruh keterlambatan dihilangkan.

````python
df_review_on_time = df_review_items[df_review_items['is_delivered'] & ~df_review_items['is_late'].fillna(False).astype(bool)]

complexity_check = pd.DataFrame({
    'late_delivery_rate_%': df_review_items[df_review_items['is_delivered']].groupby('seller_group')['is_late'].mean().mul(100).round(1),
    'low_review_rate_semua_%': seller_summary['low_review_rate_%'],
    'low_review_rate_tepat_waktu_%': low_review_by(df_review_on_time, 'seller_group')['low_review_rate_%'],
})
complexity_check
````

````python
# efek jumlah barang pada pesanan 1 seller yang tepat waktu (multi-seller dan keterlambatan dikeluarkan)
df_single_on_time = df_review_on_time[df_review_on_time['is_single_seller']]
low_review_by(df_single_on_time, 'item_group')[['jumlah_pesanan', 'low_review_rate_%']]
````

**Insight:**
- Pesanan **multi-seller** memiliki Low Review Rate **47,2%**, atau **3,4× lipat** pesanan 1 seller (13,7%). Uji Chi-Square: **Tolak Ho**.
- Menariknya, pesanan multi-seller justru **jarang telat** (Late Delivery Rate ±1%), dan Low Review Rate-nya **tetap ±47% meskipun pesanan sampai tepat waktu**. Artinya, ulasan buruk pada pesanan multi-seller **tidak dijelaskan oleh keterlambatan**, sehingga merupakan **faktor tersendiri**.
- **Jumlah barang** menunjukkan pola bertingkat: Low Review Rate naik dari **12,7% (1 barang)** menjadi **26,3% (2 barang)** hingga **35,3% (4+ barang)**. Pola ini **tetap ada** pada pesanan 1 seller yang tepat waktu (7,5% → 28,2%), sehingga juga bukan efek keterlambatan maupun multi-seller.
- Dugaan yang masuk akal: pada pesanan berisi banyak barang atau banyak seller, barang dapat tiba secara terpisah atau tidak lengkap, dan pelanggan mengisi survei sebelum seluruh barang diterima. Dugaan ini dicek dengan isi ulasan di 5.1.8.

**Implikasi:** kompleksitas pesanan adalah sumber ketidakpuasan kedua di luar keterlambatan. Namun volumenya kecil: pesanan multi-seller hanya ±1,3% dan pesanan 2+ barang ±9,8% dari pesanan berulasan.

Sesuai batasan project, hasil ini juga **tidak** berarti produk "terlalu mahal". Kedua faktor ini tidak perlu menjadi prioritas tersendiri.

### 5.1.7 Ranking Faktor: Mana yang Paling Penting?

Faktor-faktor yang diuji di 5.1.1–5.1.6 kini diurutkan untuk menentukan mana yang paling layak diprioritaskan.

Setiap faktor dinilai dari dua sisi:
1. **Risiko**: berapa kali lipat Low Review Rate kelompok berisiko dibanding kelompok pembandingnya.
2. **Dampak (volume)**: berapa banyak **ulasan buruk berlebih** yang muncul di kelompok berisiko. Ulasan buruk berlebih adalah selisih antara jumlah ulasan buruk yang terjadi dengan jumlah yang **akan terjadi jika** kelompok tersebut memiliki Low Review Rate yang sama dengan kelompok pembandingnya.

Contoh: ada 6.378 pesanan telat dengan Low Review Rate 62,4%. Jika pesanan telat memiliki rate yang sama dengan pesanan tepat waktu (9,2%), ulasan buruknya hanya ±590. Selisihnya adalah ulasan buruk berlebih yang berkaitan dengan keterlambatan.

````python
is_on_time = df_review['is_delivered'] & ~df_review['is_late'].fillna(False).astype(bool)
top_quartile = 'Q4 (tertinggi)'

factor_definitions = [
    # (faktor, kelompok berisiko, kelompok pembanding, data, p-value uji chi-square)
    ('Ketepatan waktu', 'Telat', 'Tepat waktu',
     df_review['is_late'].fillna(False).astype(bool), is_on_time),
    ('Status pesanan', 'Tidak sampai', 'Sampai',
     ~df_review['is_delivered'], df_review['is_delivered']),
    ('Ongkos kirim', 'Kuartil tertinggi', 'Kuartil 1-3',
     df_review['freight_group'] == top_quartile, df_review['freight_group'].notna() & (df_review['freight_group'] != top_quartile)),
    ('Jumlah barang', '2+ barang', '1 barang',
     df_review['n_items'] > 1, df_review['n_items'] == 1),
    ('Nilai pesanan', 'Kuartil tertinggi', 'Kuartil 1-3',
     df_review['order_value_group'] == top_quartile, df_review['order_value_group'].notna() & (df_review['order_value_group'] != top_quartile)),
    ('Jumlah seller', '>1 seller', '1 seller',
     df_review['n_sellers'] > 1, df_review['n_sellers'] == 1),
]

rows = []
for factor, risk_label, ref_label, risk_mask, ref_mask in factor_definitions:
    risk_rate = df_review.loc[risk_mask, 'is_low_review'].mean()
    ref_rate = df_review.loc[ref_mask, 'is_low_review'].mean()
    n_risk = risk_mask.sum()
    actual_low = df_review.loc[risk_mask, 'is_low_review'].sum()
    rows.append({
        'faktor': factor,
        'kelompok_berisiko': risk_label,
        'pembanding': ref_label,
        'jumlah_pesanan_berisiko': n_risk,
        'rate_berisiko_%': round(risk_rate * 100, 1),
        'rate_pembanding_%': round(ref_rate * 100, 1),
        'rasio_risiko': round(risk_rate / ref_rate, 1),
        'ulasan_buruk_berlebih': round(actual_low - n_risk * ref_rate),
    })

factor_ranking = pd.DataFrame(rows)
factor_ranking['porsi_dari_total_ulasan_buruk_%'] = (
    factor_ranking['ulasan_buruk_berlebih'] / df_review['is_low_review'].sum() * 100
).round(1)
factor_ranking = factor_ranking.sort_values('ulasan_buruk_berlebih', ascending=False).reset_index(drop=True)
factor_ranking
````

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5), sharey=True)
labels = factor_ranking['faktor'] + '\n(' + factor_ranking['kelompok_berisiko'] + ')'

axes[0].barh(labels, factor_ranking['rasio_risiko'], color='#ff9896')
axes[0].set_title('Risiko: Rasio Low Review Rate\n(kelompok berisiko ÷ pembanding)')
axes[0].set_xlabel('Kali lipat')
for i, v in enumerate(factor_ranking['rasio_risiko']):
    axes[0].text(v + 0.1, i, f'{v:.1f}×', va='center')

axes[1].barh(labels, factor_ranking['ulasan_buruk_berlebih'], color='#d62728')
axes[1].set_title('Dampak: Jumlah Ulasan Buruk Berlebih')
fig.suptitle(f"Pesanan {factor_ranking.iloc[0]['kelompok_berisiko'].lower()} adalah faktor dengan dampak terbesar: ±{factor_ranking.iloc[0]['ulasan_buruk_berlebih']:,} ulasan buruk berlebih "
             f"({factor_ranking.iloc[0]['rasio_risiko']:.1f}× risiko)".replace(',', '.'), fontsize=12)
axes[1].set_xlabel('Jumlah pesanan')
for i, v in enumerate(factor_ranking['ulasan_buruk_berlebih']):
    axes[1].text(v + 30, i, f'{v:,}'.replace(',', '.'), va='center')

axes[0].invert_yaxis()
plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **Kelompok berisiko vs pembanding** = Ketepatan waktu: Telat vs Tepat waktu; Status pesanan: Tidak sampai vs Sampai; Ongkos kirim: Kuartil tertinggi vs Kuartil 1-3; Jumlah barang: 2+ barang vs 1 barang; Nilai pesanan: Kuartil tertinggi vs Kuartil 1-3; Jumlah seller: >1 seller vs 1 seller
- **Rasio risiko** = Low Review Rate kelompok berisiko ÷ Low Review Rate kelompok pembanding
- **Ulasan buruk berlebih** = ulasan buruk kelompok berisiko − (jumlah pesanannya × Low Review Rate pembanding), yaitu kelebihan ulasan buruk dibanding jika kelompok itu setara pembanding
- **Ulasan buruk** = ulasan bintang 1–2

**Catatan membaca ranking:** angka ulasan buruk berlebih antar faktor **tidak dapat dijumlahkan**, karena satu pesanan bisa masuk beberapa kelompok sekaligus (misalnya pesanan telat yang juga ongkirnya tinggi). Angka ini juga hanya menunjukkan **keterkaitan**, bukan bukti sebab-akibat.

**Dari data pesanan, keterlambatan adalah faktor paling kuat**, baik dari sisi risiko maupun jumlah ulasan buruk berlebih. Namun ±57% ulasan buruk justru berasal dari pesanan yang sampai tepat waktu (5.1.3), dan data pesanan saja tidak bisa menjelaskan apa yang dirasakan pelanggan. **Apa yang dikeluhkan pelanggan sendiri?** Pertanyaan ini dijawab di 5.1.8 dengan membaca isi komentar ulasan.

Temuan di 5.1.8 (seller berisiko dan kategori produk berisiko) hanya berlaku **di dalam** pesanan tepat waktu, sehingga tidak dimasukkan ke ranking ini, tetapi dirangkum pada Kesimpulan Pilar 1.

### 5.1.8 Apa yang Dikeluhkan Pelanggan?

Ranking di 5.1.7 menunjukkan bahwa dari data pesanan, keterlambatan adalah faktor paling kuat. Namun data pesanan belum menjawab **apa yang sebenarnya dikeluhkan** pelanggan, terutama pada **±57% ulasan buruk dari pesanan yang sampai tepat waktu** (5.1.3). Bagian ini membaca **isi komentar** ulasan buruk sebagai konfirmasi:

| Bagian | Pertanyaan |
|---|---|
| a | Bagaimana isi komentar diberi label? |
| b | Apa keluhan utama di setiap sumber ulasan buruk (telat, tidak sampai, tepat waktu)? |
| c | Sampai mana pesanan saat ulasan ditulis? |
| d | Kenapa pesanan tepat waktu tetap mendapat ulasan buruk? (breakdown 5 kelompok + keluhan per kelompok) |
| e | Siapa yang disalahkan pelanggan? |
| f | Apakah ulasan buruk pesanan tepat waktu terkonsentrasi pada kategori produk tertentu? |

#### a. Metode: Label Isi Ulasan dengan Claude

| Aspek | Keterangan |
|---|---|
| Cakupan | Seluruh ulasan buruk (bintang 1–2) yang valid dan memiliki komentar. Komentar berbahasa Portugis. |
| Pelabel | **Claude** (model `claude-sonnet-5-5`, Anthropic), dijalankan di luar notebook dengan script Python terpisah, karena ±10 ribu komentar tidak mungkin dibaca manual satu per satu. |
| Input | Hanya **teks komentar** (judul + isi). Bintang ulasan **tidak** diberikan, sehingga label murni berdasarkan isi komentar. |
| Konteks untuk Claude | Problem statement project, deskripsi resmi dataset dari Kaggle, dan **codebook** berisi definisi, aturan prioritas, dan contoh setiap label. |
| Output | `Dataset/review_labels_llm.csv` (1 baris = 1 ulasan) |

Tiga label yang dipakai:

| Label | Pertanyaan | Nilai |
|---|---|---|
| **Pain Points** | Masalah utama apa yang dikeluhkan? | Delivery Time (belum sampai/telat), Wrong / Missing Item (barang salah, beda dari iklan, atau kurang), Product Quality (rusak, cacat, kualitas buruk), Seller Service (toko tidak merespons, stok habis/dibatalkan seller, retur sulit), Delivery Process (pelacakan, harus diambil sendiri, paket terpisah), Packaging, Cost & Billing, Others, Not Relevant (tidak ada keluhan) |
| **Review Stage** | Sampai mana pesanan saat ulasan ditulis? | Not Received Yet, Partially Received, Received, Received Not Used Yet, Used / Tested, Not Mentioned |
| **Blame Target** | Siapa yang disalahkan pelanggan? | Seller, Marketplace / Platform, Logistics / Carrier, Customer, Not Mentioned |

Jika beberapa masalah muncul sekaligus, dipilih **akar masalahnya**, misalnya "belum sampai dan toko tidak membalas" → Delivery Time.

**Kenapa memakai model bahasa, bukan daftar kata kunci?** Komentar ditulis informal, penuh singkatan dan salah ketik, sehingga daftar kata kunci mudah melewatkan keluhan yang ditulis dengan kata lain.

**Validasi manual (QA):** label dari Claude adalah hasil interpretasi model, sehingga akurasinya dicek manual oleh tim sebelum label dipakai. Metode dan hasil pengujiannya dijelaskan di bawah.

**Metode pengujian akurasi label (QA)**

| Aspek | Keterangan |
|---|---|
| Populasi | **10.813 ulasan buruk** valid yang berkomentar, yaitu seluruh ulasan yang diberi label oleh Claude |
| Sampel | **150 ulasan** diambil **acak** dan dibagi rata per bintang: **75 bintang 1** dan **75 bintang 2** (±1,4% dari populasi) |
| Cara mengecek | Setiap ulasan dibaca dalam teks aslinya (Portugis) dibantu terjemahan Google Translate. Ketiga label (Pain Points, Review Stage, Blame Target) ditandai **benar/salah** oleh anggota tim, ±50 ulasan per anggota |
| Ukuran yang dilaporkan | **Akurasi** = label benar ÷ label dicek, bersama **interval kepercayaan 95% (Wilson)** dan **margin of error**. Interval Wilson dipakai karena sampelnya kecil dan akurasinya mendekati 100% |

**Kenapa 150 sampel?**

- **Tujuan QA adalah mengukur akurasi label Claude**, bukan melabeli ulang seluruh ulasan. Akurasi adalah sebuah **proporsi** (label benar ÷ label dicek), sehingga ketelitiannya ditentukan oleh **jumlah sampel**, bukan jumlah populasi:

$$\text{margin of error} = 1{,}96 \times \sqrt{\frac{p(1-p)}{n}}$$

  Semakin besar n, semakin sempit rentang ketidakpastiannya. Ukuran populasi (±10 ribu ulasan) hampir tidak berpengaruh, karena sampel hanya ±1,4% dari populasi (faktor koreksi populasi terbatas = 0,993, hampir 1).
- **Angka 150 dipilih secara praktis.** Setiap ulasan dibaca manual dan dinilai pada 3 label, sehingga ±50 ulasan per anggota tim masih dapat dicek dengan teliti. Mengecek seluruh ulasan sama saja dengan melabeli manual dari awal, yang justru ingin dihindari dengan memakai Claude.
- **Konsekuensinya tidak disembunyikan.** Dengan n = 150, margin of error paling besar (saat akurasi 50%) adalah **±8 pp**. Agar margin terburuk turun ke ±5 pp dibutuhkan **±385 sampel** (future work). Karena itu setiap akurasi dilaporkan bersama interval kepercayaan dan margin of error-nya.

**Kenapa dibagi rata per bintang?** Di populasi, ±80% ulasan berkomentar adalah bintang 1 dan hanya ±20% bintang 2. Jika diambil acak biasa, ulasan bintang 2 hanya ±30 sampel sehingga akurasinya sulit dilihat. Karena pembagian rata ini tidak sesuai populasi, akurasi juga dihitung ulang dengan **bobot sesuai proporsi bintang di populasi**, untuk mengecek apakah pembagian rata membuat akurasi terlihat lebih tinggi atau lebih rendah.

**Hasil QA**

| Label | Benar / dicek | Akurasi | Interval kepercayaan 95% | Margin of error | Bintang 1 | Bintang 2 | Tertimbang sesuai populasi |
|---|---|---|---|---|---|---|---|
| **Pain Points** | 141 / 150 | **94,0%** | 89,0% – 96,8% | ±3,8 pp | 97,3% | 90,7% | 96,0% |
| **Review Stage** | 142 / 150 | **94,7%** | 89,8% – 97,3% | ±3,6 pp | 96,0% | 93,3% | 95,5% |
| **Blame Target** | 119 / 150 | **79,3%** | 72,2% – 85,0% | ±6,5 pp | 81,3% | 77,3% | 80,5% |

- **Pain Points** dan **Review Stage** cukup akurat untuk dipakai. Semua nilai label Pain Points (9) dan Review Stage (6) muncul di sampel. Label yang paling banyak dipakai di analisis hampir selalu benar: *Delivery Time* 48/49, *Wrong / Missing Item* 48/53, dan *Product Quality* 21/21. Kesalahan Pain Points terbanyak ada pada *Wrong / Missing Item* (5 dari 53).
- **Blame Target** paling rendah. 25 dari 31 kesalahannya ada pada label *Not Mentioned* (76/101 benar). Label ini paling bergantung pada tafsir, karena codebook mendefinisikan Blame Target sebagai **pihak yang disebut pelanggan secara eksplisit**, sehingga komentar yang tidak menyebut pihak mana pun mudah dinilai berbeda oleh pemeriksa. Label yang menyebut pihak tertentu tetap cukup akurat: *Seller* 27/32 dan *Logistics / Carrier* 7/7.
- **Ukuran sampel cukup untuk kesimpulan ini.** Batas bawah interval kepercayaan Pain Points dan Review Stage tetap di atas 89%, dan Blame Target tetap jelas paling rendah. Jadi kesimpulan tidak berubah walaupun memakai batas bawahnya.
- **Pembagian rata per bintang membuat akurasi yang dilaporkan sedikit lebih rendah (lebih hati-hati).** Akurasi pada bintang 2 lebih rendah, sehingga jika dibobot sesuai populasi (±80% bintang 1) akurasinya sedikit lebih tinggi. Angka yang dilaporkan (94,0% / 94,7% / 79,3%) tetap dipakai karena lebih hati-hati.
- Karena itu, Pain Points dan Review Stage dipakai sebagai temuan utama, sedangkan Blame Target hanya dipakai sebagai **gambaran persepsi pelanggan** dan dibaca dengan hati-hati.

````python
review_labels = pd.read_csv(DATA_PATH + 'review_labels_llm.csv')
label_cols = ['Pain Points', 'Review Stage', 'Blame Target']

df_low_labeled = df_low_review.merge(review_labels[['review_id'] + label_cols], on='review_id', how='left')
df_low_labeled['berkomentar'] = df_low_labeled['Pain Points'].notna()
df_comment = df_low_labeled[df_low_labeled['berkomentar']].copy()

print(f"Ulasan buruk               : {df_low_labeled.shape[0]}")
print(f"Ulasan buruk berkomentar   : {df_comment.shape[0]} ({df_low_labeled['berkomentar'].mean() * 100:.1f}%)")
df_low_labeled.groupby('sumber')['berkomentar'].mean().mul(100).round(1).to_frame('persen_berkomentar')
````

Sekitar 3 dari 4 ulasan buruk memiliki komentar, dan porsinya mirip di ketiga sumber, sehingga hasil di bawah cukup mewakili.

#### b. Keluhan Utama (Pain Points) per Sumber Ulasan Buruk

````python
pain_order = ['Delivery Time', 'Wrong / Missing Item', 'Product Quality', 'Seller Service', 'Delivery Process',
              'Packaging', 'Cost & Billing', 'Others', 'Not Relevant']
source_order = ['Sampai tepat waktu', 'Sampai, tapi telat', 'Tidak sampai']
main_pain = ['Delivery Time', 'Wrong / Missing Item', 'Product Quality', 'Seller Service']


def pain_share(data, group_col, group_order, total_label=None):
    # persen Pain Points per kelompok (baris = kelompok, kolom = keluhan)
    table = pd.crosstab(data[group_col], data['Pain Points'], normalize='index').mul(100)
    if total_label:
        table.loc[total_label] = data['Pain Points'].value_counts(normalize=True).mul(100)
        group_order = [total_label] + group_order
    return table.reindex(index=group_order, columns=pain_order).fillna(0).round(1)


def plot_pain_share(table, title):
    # 100% stacked bar: 4 keluhan utama + gabungan sisanya
    plot_data = table[main_pain].assign(Lainnya=100 - table[main_pain].sum(axis=1))
    ax = plot_data.plot(kind='barh', stacked=True, figsize=(11, 0.6 * len(plot_data) + 1.6), width=0.7,
                        color=['#d62728', '#ff7f0e', '#1f77b4', '#9467bd', '#c7c7c7'])
    ax.invert_yaxis()
    for container, col in zip(ax.containers, plot_data.columns):
        ax.bar_label(container, labels=[f'{v:.0f}%' if v >= 6 else '' for v in plot_data[col]],
                     label_type='center', fontsize=9, color='black' if col == 'Lainnya' else 'white')
    ax.set_title(title)
    ax.set_xlabel('Persen ulasan buruk berkomentar')
    ax.set_ylabel('')
    ax.set_xlim(0, 100)
    ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.25), ncol=5, frameon=False)
    plt.tight_layout()
    plt.show()


pain_by_source = pain_share(df_comment, 'sumber', source_order, total_label='Semua ulasan buruk')
plot_pain_share(pain_by_source, 'Pesanan telat dikeluhkan soal waktu, pesanan tepat waktu soal barang kurang/salah dan kualitas\n(keluhan utama ulasan buruk berkomentar, per sumber)')
pain_by_source
````

**Keterangan istilah:**

- **Ulasan buruk berkomentar** = ulasan bintang 1–2 yang memiliki komentar teks
- **Keluhan utama (Pain Points)** = label isi komentar dari Claude, 1 label per ulasan (akar masalahnya)
- **Delivery Time** = belum sampai atau telat
- **Wrong / Missing Item** = barang salah, beda dari iklan, atau kurang
- **Product Quality** = barang rusak, cacat, atau kualitasnya buruk
- **Seller Service** = toko tidak merespons, stok habis/dibatalkan seller, atau retur sulit
- **Lainnya** = Delivery Process, Packaging, Cost & Billing, Others, dan Not Relevant
- **Sampai tepat waktu** = pesanan delivered yang sampai pada atau sebelum tanggal estimasi
- **Sampai, tapi telat** = pesanan delivered yang sampai setelah tanggal estimasi
- **Tidak sampai** = pesanan yang tidak tercatat sampai ke pelanggan (status selain delivered, misalnya canceled, unavailable, atau shipped)

**Insight:**
- Secara keseluruhan, tiga keluhan mendominasi: **keterlambatan (±35%)**, **barang kurang/salah (±30%)**, dan **kualitas produk (±16%)**. Ketiganya mencakup ±81% ulasan buruk berkomentar.
- Pada pesanan **telat**, ±78% keluhan adalah soal waktu (Delivery Time), sesuai dugaan. Pada pesanan **tidak sampai**, selain waktu (±61%), ±22% mengeluhkan **layanan seller**, misalnya stok habis atau pesanan dibatalkan seller.
- Pada pesanan **tepat waktu**, isi keluhan berubah total: **barang kurang/salah (±47%)** dan **kualitas produk (±25%)** mendominasi, sedangkan keluhan waktu hanya ±11%. Artinya, ulasan buruk pada pesanan tepat waktu umumnya soal **barang yang diterima**, bukan soal pengiriman.

#### c. Posisi Pesanan saat Ulasan Ditulis (Review Stage)

````python
stage_order = ['Not Received Yet', 'Partially Received', 'Received', 'Received, Not Used Yet', 'Used / Tested', 'Not Mentioned']

pd.crosstab(df_comment['sumber'], df_comment['Review Stage'], normalize='index').mul(100).reindex(
    index=source_order, columns=stage_order
).fillna(0).round(1)
````

**Insight:**
- Pada ulasan buruk pesanan **telat**, **±77%** ditulis saat barang **belum diterima**. Angka ini sejalan dengan temuan 5.1.2, bahwa pada ±74,5% pesanan telat survei sudah dikirim sebelum barang tiba. Pelanggan menilai **pengalaman menunggu**, bukan produknya.
- Pada pesanan **tepat waktu**, **±24%** ulasan buruk ditulis saat barang **baru diterima sebagian**, padahal status pesanannya sudah `delivered`. Hal ini dicek lebih lanjut pada pesanan kompleks di bawah.

#### d. Kenapa Pesanan Tepat Waktu Tetap Mendapat Ulasan Buruk?

Kelompok terbesar pada grafik di 5.1.3 adalah **8.245 ulasan buruk dari pesanan yang sampai tepat waktu (57,4%)**. Karena keterlambatan tidak bisa menjelaskan kelompok ini, kelompok tersebut dipecah lagi menggunakan **data operasional pesanan** (tanpa melihat isi komentar). Setelah itu, isi keluhan di setiap kelompok dibaca dengan label ulasan.

Setiap pesanan tepat waktu dimasukkan ke **tepat satu kelompok** (tidak tumpang tindih), dicek berurutan dari atas:

| Urutan | Kelompok | Kriteria | Alasan dicek |
|---|---|---|---|
| 1 | **Multi-seller** | Barang berasal dari > 1 seller | Pesanan dikirim terpisah oleh beberapa seller, sehingga berisiko diterima tidak bersamaan |
| 2 | **1 seller, 2+ barang** | 1 seller, jumlah barang ≥ 2 | Lebih banyak barang berarti lebih banyak peluang ada barang kurang, rusak, atau salah |
| 3 | **1 barang, seller berisiko** | Seller yang Low Review Rate pesanan tepat waktunya **signifikan lebih tinggi** dari rata-rata (kriteria di bawah) | Ulasan buruk terkonsentrasi pada seller tertentu walaupun pengirimannya tepat waktu |
| 4 | **1 barang, pengiriman > 3 minggu** | `delivery_days` > 21 hari | Tepat waktu menurut estimasi, tetapi waktu tunggunya tetap lama (±2× median waktu pengiriman) |
| 5 | **1 barang, tanpa sinyal operasional** | Sisanya | Tidak ada tanda masalah di data operasional |

Urutan dimulai dari kompleksitas pesanan karena risikonya paling tinggi (lihat 5.1.4), lalu seller karena seller adalah pihak yang bisa langsung ditindaklanjuti oleh tim Seller Management.

**Kriteria seller berisiko:**
- Seller memiliki **minimal 55 pesanan tepat waktu** (pesanan 1 seller). Batas ini mengikuti syarat z-test proporsi, yaitu **n × p ≥ 5**. Rata-rata Low Review Rate pesanan tepat waktu ±9,25%, sehingga dibutuhkan n ≥ 5 ÷ 0,0925 ≈ 55 pesanan agar z-test dapat dipakai.
- Low Review Rate seller tersebut diuji dengan **one sample z-test proporsi satu sisi** (`proportions_ztest`), yaitu membandingkan proporsi ulasan buruk seller dengan rata-rata pesanan tepat waktu.
  - **Ho:** Low Review Rate seller = rata-rata pesanan tepat waktu.
  - **Ha:** Low Review Rate seller > rata-rata pesanan tepat waktu.
- Tingkat signifikansi yang dipakai adalah **α = 0,01** (lebih ketat dari 0,05), karena ratusan seller diuji sekaligus: dengan α = 0,05, sekitar 5% seller akan lolos hanya karena kebetulan.

````python
from statsmodels.stats.proportion import proportions_ztest

df_on_time = df_review[df_review['is_delivered'] & ~df_review['is_late'].fillna(False).astype(bool)].copy()
df_on_time['is_low_review'] = df_on_time['is_low_review'].astype(bool)
on_time_rate = df_on_time['is_low_review'].mean()

seller_on_time = df_on_time[df_on_time['n_sellers'] == 1].groupby('seller_id').agg(
    jumlah_pesanan=('order_id', 'count'),
    jumlah_low_review=('is_low_review', 'sum'),
)
# minimal pesanan agar syarat z-test terpenuhi: n × p ≥ 5
min_on_time_orders = int(np.ceil(5 / on_time_rate))
seller_on_time = seller_on_time[seller_on_time['jumlah_pesanan'] >= min_on_time_orders].copy()
seller_on_time['low_review_rate_%'] = (seller_on_time['jumlah_low_review'] / seller_on_time['jumlah_pesanan'] * 100).round(1)
seller_on_time['p_value'] = [
    proportions_ztest(count=low, nobs=n, value=on_time_rate, alternative='larger')[1]
    for low, n in zip(seller_on_time['jumlah_low_review'], seller_on_time['jumlah_pesanan'])
]
risky_sellers = seller_on_time[seller_on_time['p_value'] < 0.01].index

print(f'Low Review Rate pesanan tepat waktu : {on_time_rate * 100:.2f}%')
print(f'Minimal pesanan tepat waktu (5 ÷ {on_time_rate:.4f}): {min_on_time_orders}')
print(f'Seller dengan >= {min_on_time_orders} pesanan tepat waktu:', seller_on_time.shape[0])
print('Seller berisiko (p-value < 0,01)       :', len(risky_sellers))
print(f"Low Review Rate terendah di antara seller berisiko: {seller_on_time.loc[risky_sellers, 'low_review_rate_%'].min():.1f}%")
````

````python
segment_order = [
    'Multi-seller', '1 seller, 2+ barang', '1 barang, seller berisiko',
    '1 barang, pengiriman > 3 minggu', '1 barang, tanpa sinyal operasional',
]
df_on_time['segmen'] = np.select(
    [
        df_on_time['n_sellers'] > 1,
        df_on_time['n_items'] > 1,
        df_on_time['seller_id'].isin(risky_sellers),
        df_on_time['delivery_days'] > 21,
    ],
    segment_order[:4],
    default=segment_order[4],
)

on_time_breakdown = low_review_by(df_on_time, 'segmen').reindex(segment_order)
on_time_breakdown['porsi_ulasan_buruk_%'] = (
    on_time_breakdown['jumlah_low_review'] / on_time_breakdown['jumlah_low_review'].sum() * 100
).round(1)
on_time_breakdown['porsi_pesanan_%'] = (on_time_breakdown['jumlah_pesanan'] / on_time_breakdown['jumlah_pesanan'].sum() * 100).round(1)
on_time_breakdown[['jumlah_pesanan', 'porsi_pesanan_%', 'jumlah_low_review', 'porsi_ulasan_buruk_%', 'low_review_rate_%']]
````

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5), sharey=True)
colors = ['#9e9e9e' if seg == segment_order[4] else '#d62728' for seg in segment_order]

axes[0].barh(segment_order, on_time_breakdown['jumlah_low_review'], color=colors)
axes[0].set_title(f"Jumlah Ulasan Buruk (total {on_time_breakdown['jumlah_low_review'].sum():,})".replace(',', '.'))
axes[0].set_xlabel('Jumlah pesanan dengan ulasan buruk')
axes[0].set_xlim(0, on_time_breakdown['jumlah_low_review'].max() * 1.35)
for i, (count, share) in enumerate(zip(on_time_breakdown['jumlah_low_review'], on_time_breakdown['porsi_ulasan_buruk_%'])):
    axes[0].text(count + 50, i, f'{count:,} ({share:.1f}%)'.replace(',', '.'), va='center')

axes[1].barh(segment_order, on_time_breakdown['low_review_rate_%'], color=colors)
axes[1].axvline(on_time_rate * 100, color='black', linestyle='--', linewidth=1)
axes[1].text(on_time_rate * 100 + 0.5, 3.5, f'rata-rata tepat waktu {on_time_rate * 100:.1f}%', va='center', fontsize=9)
axes[1].set_title('Low Review Rate per Kelompok')
axes[1].set_xlabel('Low Review Rate (%)')
axes[1].set_xlim(0, on_time_breakdown['low_review_rate_%'].max() + 8)
for i, rate in enumerate(on_time_breakdown['low_review_rate_%']):
    axes[1].text(rate + 0.5, i, f'{rate:.1f}%', va='center')

axes[0].invert_yaxis()
fig.suptitle(f"±{100 - on_time_breakdown.loc[segment_order[4], 'porsi_ulasan_buruk_%']:.0f}% ulasan buruk pesanan tepat waktu memiliki sinyal operasional, "
             'terbesar dari pesanan kompleks\n(breakdown ulasan buruk pada pesanan yang sampai tepat waktu)', fontsize=12)
plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **Kelompok** = setiap pesanan tepat waktu masuk ke kelompok pertama yang cocok, berurutan dari atas
- **Multi-seller** = satu pesanan berisi barang dari lebih dari 1 seller
- **Pesanan kompleks** = Multi-seller dan 1 seller, 2+ barang
- **Seller berisiko** = seller yang Low Review Rate pesanan tepat waktunya signifikan lebih tinggi dari rata-rata (9,2%); z-test proporsi satu sisi, p < 0,01, min. 55 pesanan tepat waktu (30 seller)
- **Pengiriman > 3 minggu** = lama pengiriman lebih dari 21 hari sejak pembelian, walaupun tidak melewati tanggal estimasi
- **Tanpa sinyal operasional** = pesanan 1 barang yang tidak masuk kelompok di atas (abu-abu); 4 kelompok lainnya = memiliki sinyal operasional (merah)
- **Rata-rata tepat waktu (garis putus-putus)** = Low Review Rate seluruh pesanan yang sampai tepat waktu
- **Ulasan buruk** = ulasan bintang 1–2

````python
# sensitivitas urutan: ulasan buruk yang memenuhi kriteria seller berisiko sekaligus pengiriman > 3 minggu
one_item = df_on_time['n_items'] == 1
both_criteria = (one_item & df_on_time['seller_id'].isin(risky_sellers)
                 & (df_on_time['delivery_days'] > 21) & df_on_time['is_low_review'])
low_by_segment = df_on_time[df_on_time['is_low_review']]['segmen'].value_counts()

print('Ulasan buruk yang memenuhi kedua kriteria          :', both_criteria.sum())
print('Jika urutan 3 & 4 ditukar -> seller berisiko       :', low_by_segment['1 barang, seller berisiko'] - both_criteria.sum())
print('Jika urutan 3 & 4 ditukar -> pengiriman > 3 minggu :', low_by_segment['1 barang, pengiriman > 3 minggu'] + both_criteria.sum())
````

**Insight:**
- **Kompleksitas pesanan** menjadi sumber terbesar yang terlihat di data operasional. Pesanan **multi-seller** (584 ulasan buruk, 7,1%) dan **1 seller dengan 2+ barang** (1.629, 19,8%) bersama-sama menyumbang **±27% ulasan buruk pesanan tepat waktu**, padahal hanya ±10% dari pesanan tepat waktu. Low Review Rate-nya **46,9%** dan **21,0%**, jauh di atas rata-rata pesanan tepat waktu (9,2%).
- **30 seller berisiko** menyumbang **1.268 ulasan buruk (15,4%)** dari pesanan 1 barang yang tepat waktu, dengan Low Review Rate **15,2%** (±1,6× rata-rata). Artinya, sebagian ulasan buruk pada pesanan tepat waktu **terkonsentrasi pada seller tertentu**, bukan tersebar merata.
- Pesanan 1 barang yang tepat waktu tetapi **pengirimannya lebih dari 3 minggu** menyumbang **622 ulasan buruk (7,5%)** dengan Low Review Rate **11,9%**. Pesanan dianggap tepat waktu karena estimasinya panjang, tetapi pelanggan tetap menunggu lama.
- Sisanya, **4.142 ulasan buruk (50,2%)**, berasal dari pesanan 1 barang **tanpa sinyal operasional apa pun**. Kelompok ini mencakup ±75% pesanan tepat waktu, tetapi Low Review Rate-nya hanya **6,2%**, di bawah rata-rata.

**Implikasi:**
- Sekitar **separuh (±50%)** ulasan buruk pada pesanan tepat waktu memiliki sinyal yang dapat ditindaklanjuti dari data operasional:
  - kompleksitas pesanan → sudah terbukti sebagai faktor tersendiri di 5.1.4, dan isi keluhannya dicek di bawah,
  - seller berisiko → bahan evaluasi tim **Seller Management** (misalnya kualitas produk atau kesesuaian deskripsi listing),
  - waktu tunggu yang lama → estimasi yang panjang membuat pesanan "tepat waktu" di sistem, tetapi tetap terasa lama bagi pelanggan.
- Separuh sisanya tidak terlihat dari data operasional. Kemungkinan berkaitan dengan produk itu sendiri (kualitas, tidak sesuai harapan), sehingga hanya dapat ditelusuri dari **isi ulasan pelanggan**, yang dibaca di bawah.

**Catatan:**
- Seller berisiko ditentukan dari Low Review Rate seller itu sendiri. Jadi temuan ini menunjukkan **konsentrasi** ulasan buruk pada seller tertentu (keterkaitan), bukan bukti penyebabnya.
- Pembagian kelompok bergantung pada urutan pengecekan. Ada 215 ulasan buruk yang memenuhi kriteria seller berisiko sekaligus pengiriman > 3 minggu. Jika urutan langkah 3 dan 4 ditukar, kelompok seller berisiko menjadi 1.053 dan pengiriman > 3 minggu menjadi 837. Kesimpulan tidak berubah: kedua kelompok tetap jauh lebih kecil daripada kompleksitas pesanan dan kelompok tanpa sinyal operasional.

Breakdown di atas membagi ulasan buruk pesanan tepat waktu ke 5 kelompok berdasarkan data operasional. Label isi ulasan kini dapat menjawab apa yang dikeluhkan di setiap kelompok, termasuk kelompok **"tanpa sinyal operasional"** yang belum terjelaskan:

````python
on_time_low = df_on_time[df_on_time['is_low_review']].merge(review_labels[['review_id'] + label_cols], on='review_id')

pain_by_segment = pain_share(on_time_low, 'segmen', segment_order)
plot_pain_share(pain_by_segment, 'Pesanan kompleks dikeluhkan karena barang kurang/salah, seller berisiko karena kualitas produk\n(keluhan utama ulasan buruk pesanan tepat waktu, per kelompok)')
pain_by_segment.insert(0, 'jumlah_berkomentar', on_time_low['segmen'].value_counts().reindex(segment_order))
pain_by_segment
````

- **Ulasan buruk berkomentar** = ulasan bintang 1–2 yang memiliki komentar teks

- **Multi-seller** = satu pesanan berisi barang dari lebih dari 1 seller
- **Pesanan kompleks** = Multi-seller dan 1 seller, 2+ barang
- **Seller berisiko** = seller yang Low Review Rate pesanan tepat waktunya signifikan lebih tinggi dari rata-rata (9,2%); z-test proporsi satu sisi, p < 0,01, min. 55 pesanan tepat waktu (30 seller)
- **Pengiriman > 3 minggu** = lama pengiriman lebih dari 21 hari sejak pembelian, walaupun tidak melewati tanggal estimasi

**Insight:**

- **Seller berisiko:** keluhannya adalah **kualitas produk (±41%)** dan **barang salah/tidak sesuai (±37%)**. Jadi ulasan buruk yang terkonsentrasi pada seller ini berasal dari **produk dan listing-nya**, bukan dari pengiriman.
- **Pengiriman > 3 minggu:** keluhan waktu paling tinggi di antara kelompok tepat waktu (±33%). Walaupun tidak melewati estimasi, pelanggan tetap merasakan waktu tunggu yang lama.
- **Tanpa sinyal operasional** (±50% ulasan buruk pesanan tepat waktu): didominasi **barang salah/tidak sesuai (±36%)** dan **kualitas produk (±32%)**. Inilah jawaban pertanyaan dari breakdown di atas: kelompok ini tidak terlihat di data operasional karena masalahnya ada pada **produk yang diterima**, misalnya tidak sesuai deskripsi atau kualitasnya buruk.

**Cek kelengkapan pesanan kompleks (5.1.4) dengan isi ulasan.** Tabel di atas sudah menunjukkan bahwa keluhan pesanan kompleks didominasi **barang kurang/salah** (multi-seller ±72%, 1 seller 2+ barang ±78%). Yang belum terjawab adalah dugaan di 5.1.4: apakah barangnya memang **baru diterima sebagian** saat ulasan ditulis?

Hal ini dicek dengan label *Review Stage* = *Partially Received* pada ulasan buruk pesanan **tepat waktu** yang berkomentar, per jumlah barang dan per jumlah seller.

````python
on_time_comment = df_review_on_time[df_review_on_time['is_low_review'].astype(bool)].merge(
    review_labels[['review_id'] + label_cols], on='review_id'
)
on_time_comment['barang_kurang_salah'] = on_time_comment['Pain Points'] == 'Wrong / Missing Item'
on_time_comment['baru_diterima_sebagian'] = on_time_comment['Review Stage'] == 'Partially Received'


def complaint_share(group_col):
    grouped = on_time_comment.groupby(group_col, observed=True)
    return pd.DataFrame({
        'jumlah_berkomentar': grouped.size(),
        'persen_baru_diterima_sebagian': grouped['baru_diterima_sebagian'].mean().mul(100).round(1),
    })


pd.concat([complaint_share('item_group'), complaint_share('seller_group')])
````

````python
# contoh komentar pesanan multi-barang yang dilabel barang kurang/salah
on_time_comment.loc[
    on_time_comment['barang_kurang_salah'] & (on_time_comment['n_items'] > 1),
    ['n_items', 'n_sellers', 'review_score', 'Review Stage', 'review_comment_message']
].head(5)
````

**Temuan:**
- Pola **baru diterima sebagian** sangat tegas: hanya **±5%** pada pesanan 1 barang, tetapi **±70–74%** pada pesanan 2+ barang, dan sedikit naik seiring jumlah barang.
- Pada pesanan **multi-seller**, **±78%** ulasan buruk ditulis saat barang baru diterima sebagian, dibanding ±20% pada pesanan 1 seller.
- Contoh komentar di atas memperlihatkan polanya, misalnya pelanggan yang membeli 2 mini bike tetapi baru menerima satu.

**Kesimpulan:** dugaan bahwa pesanan kompleks sering diterima **tidak lengkap** didukung kuat oleh isi ulasan. Kemungkinan barang dikirim dalam beberapa paket, tetapi status pesanan sudah tercatat `delivered` ketika sebagian barang sampai, sehingga pesanan terlihat tepat waktu di sistem dan pelanggan menilai sebelum seluruh barang diterima. Konfirmasi penuh tetap memerlukan data status per paket (lihat Future Work).

#### f. Kategori Produk pada Pesanan Tepat Waktu

Keluhan kualitas dan ketidaksesuaian produk mendominasi ulasan buruk pesanan tepat waktu. Pertanyaan lanjutannya: apakah ulasan buruk ini terkonsentrasi pada **kategori produk** tertentu, dan apakah seller berisiko (bagian d) hanya kebetulan menjual kategori yang rawan?

**Metode:**
- Hanya memakai **pesanan 1 barang yang tepat waktu**, sehingga setiap pesanan memiliki tepat satu kategori, dan pengaruh keterlambatan serta kompleksitas pesanan sudah dikeluarkan (sama seperti pengecekan ongkir di 5.1.5).
- Hanya kategori dengan **minimal 500 pesanan** yang dibandingkan, agar rate tidak dipengaruhi segelintir pesanan.
- Uji **Chi-Square** untuk melihat apakah Low Review Rate berbeda antar kategori, lalu **one sample z-test proporsi satu sisi (α = 0,01)** per kategori terhadap rata-rata, sama seperti kriteria seller berisiko di bagian d.
  - **Ho:** Low Review Rate kategori = rata-rata pesanan 1 barang yang tepat waktu.
  - **Ha:** Low Review Rate kategori > rata-rata.

````python
# pesanan 1 barang yang tepat waktu: setiap pesanan memiliki tepat satu kategori
order_category = df_clean.drop_duplicates('order_id').set_index('order_id')['product_category_name_english']
df_one_item = df_on_time[df_on_time['n_items'] == 1].copy()
df_one_item['kategori'] = df_one_item['order_id'].map(order_category)
one_item_rate = df_one_item['is_low_review'].mean()

category_summary = df_one_item.groupby('kategori').agg(
    jumlah_pesanan=('order_id', 'count'),
    jumlah_low_review=('is_low_review', 'sum'),
)
category_summary['low_review_rate_%'] = (category_summary['jumlah_low_review'] / category_summary['jumlah_pesanan'] * 100).round(2)

min_orders = 500
category_tested = category_summary[category_summary['jumlah_pesanan'] >= min_orders].copy()
category_tested['p_value'] = [
    proportions_ztest(count=low, nobs=n, value=one_item_rate, alternative='larger')[1]
    for low, n in zip(category_tested['jumlah_low_review'], category_tested['jumlah_pesanan'])
]
risky_categories = category_tested[category_tested['p_value'] < 0.01].sort_values('low_review_rate_%', ascending=False).index

tested_orders = df_one_item[df_one_item['kategori'].isin(category_tested.index)]
stat, pvalue, dof, expected = chi2_contingency(pd.crosstab(tested_orders['kategori'], tested_orders['is_low_review']))

print(f"Pesanan 1 barang yang tepat waktu        : {df_one_item.shape[0]} (Low Review Rate {one_item_rate * 100:.2f}%)")
print(f"Kategori dengan >= {min_orders} pesanan        : {category_tested.shape[0]} dari {category_summary.shape[0]} kategori "
      f"({category_tested['jumlah_pesanan'].sum() / category_summary['jumlah_pesanan'].sum() * 100:.1f}% pesanan)")
print(f'Chi-Square kategori vs ulasan buruk      : p-value = {pvalue:.3g}')
print(f'Kategori berisiko (z-test, p < 0,01)     : {len(risky_categories)}')

risky_cat = category_tested.loc[risky_categories]
risky_cat_excess = risky_cat['jumlah_low_review'].sum() - risky_cat['jumlah_pesanan'].sum() * one_item_rate
print(f"Kategori berisiko: {risky_cat['jumlah_pesanan'].sum() / df_one_item.shape[0] * 100:.1f}% pesanan, "
      f"{risky_cat['jumlah_low_review'].sum() / df_one_item['is_low_review'].sum() * 100:.1f}% ulasan buruk, "
      f'ulasan buruk berlebih {risky_cat_excess:.0f}')
category_tested.sort_values('low_review_rate_%', ascending=False).round({'p_value': 4})
````

````python
plot_data = category_tested.sort_values('low_review_rate_%')
colors = ['#d62728' if category in risky_categories else '#9e9e9e' for category in plot_data.index]

fig, ax = plt.subplots(figsize=(10, 8))
ax.barh(plot_data.index, plot_data['low_review_rate_%'], color=colors)
ax.axvline(one_item_rate * 100, color='black', linestyle='--', linewidth=1)
ax.text(one_item_rate * 100, len(plot_data) - 0.4, f' rata-rata {one_item_rate * 100:.1f}%', va='bottom', fontsize=9)
for i, (rate, n) in enumerate(zip(plot_data['low_review_rate_%'], plot_data['jumlah_pesanan'])):
    ax.text(rate + 0.2, i, f'{rate:.1f}%  (n={n:,})'.replace(',', '.'), va='center', fontsize=8)
ax.set_xlim(0, plot_data['low_review_rate_%'].max() * 1.35)
ax.set_title(f"{len(risky_categories)} kategori memiliki ulasan buruk signifikan lebih tinggi, tertinggi {risky_categories[0]} "
             f"({category_tested.loc[risky_categories[0], 'low_review_rate_%']:.1f}%)\n"
             '(Low Review Rate per kategori, pesanan 1 barang tepat waktu; merah = z-test proporsi p < 0,01)')
ax.set_xlabel('Low Review Rate (%)')
ax.set_ylabel('Kategori produk')
plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **Kategori berisiko (merah)** = kategori dengan Low Review Rate signifikan lebih tinggi dari rata-rata; z-test proporsi satu sisi, p < 0,01
- **Rata-rata (garis putus-putus)** = Low Review Rate seluruh pesanan 1 barang yang sampai tepat waktu
- **Cakupan** = hanya pesanan 1 barang yang tepat waktu dan kategori dengan min. 500 pesanan, agar keterlambatan dan jumlah barang tidak ikut memengaruhi; n = jumlah pesanan

````python
# apakah seller berisiko hanya menjual kategori yang rawan?
# ekspektasi = Low Review Rate kategori dari setiap pesanan, dirata-rata per kelompok seller
category_rate = df_one_item.groupby('kategori')['is_low_review'].mean()
df_one_item['ekspektasi_kategori'] = df_one_item['kategori'].map(category_rate)
df_one_item['kelompok_seller'] = np.where(df_one_item['seller_id'].isin(risky_sellers), 'Seller berisiko', 'Seller lain')

seller_vs_category = df_one_item.groupby('kelompok_seller').agg(
    jumlah_pesanan=('order_id', 'count'),
    rate_aktual=('is_low_review', 'mean'),
    rate_ekspektasi_kategori=('ekspektasi_kategori', 'mean'),
)
seller_vs_category[['rate_aktual', 'rate_ekspektasi_kategori']] = (seller_vs_category[['rate_aktual', 'rate_ekspektasi_kategori']] * 100).round(1)
seller_vs_category['rasio_aktual_vs_ekspektasi'] = (seller_vs_category['rate_aktual'] / seller_vs_category['rate_ekspektasi_kategori']).round(2)
print('Low Review Rate aktual vs ekspektasi berdasarkan kategori yang dijual (%):')
print(seller_vs_category.to_string())

# keluhan utama ulasan buruk pada kategori berisiko vs kategori lain
category_low = df_one_item[df_one_item['is_low_review']].merge(review_labels[['review_id', 'Pain Points']], on='review_id')
category_low['kelompok_kategori'] = np.where(category_low['kategori'].isin(risky_categories), 'Kategori berisiko', 'Kategori lain')
category_pain = pd.crosstab(category_low['kelompok_kategori'], category_low['Pain Points'], normalize='index').mul(100).round(1)
category_pain.insert(0, 'jumlah_berkomentar', category_low['kelompok_kategori'].value_counts())
category_pain[['jumlah_berkomentar', 'Product Quality', 'Wrong / Missing Item', 'Delivery Time']]
````

**Insight:**
- Low Review Rate pesanan 1 barang yang tepat waktu **berbeda antar kategori** (Chi-Square: **Tolak Ho**, p ≈ 2e-46). Di antara 26 kategori yang dibandingkan, rate-nya berkisar dari **4,8%** (stationery) sampai **14,2%** (office_furniture), dengan rata-rata 7,5%.
- **6 kategori berisiko** (signifikan lebih tinggi dari rata-rata): **office_furniture (14,2%)**, **unknown (11,0%)**, **telephony (9,5%)**, **bed_bath_table (9,3%)**, **computers_accessories (9,3%)**, dan **watches_gifts (8,6%)**. Keenam kategori ini mencakup ±29% pesanan 1 barang yang tepat waktu, tetapi menyumbang ±36% ulasan buruknya (±450 ulasan buruk berlebih dibanding rata-rata).
- Kategori `unknown` adalah produk **tanpa kategori** di data produk (3.5). Produk tanpa kategori memiliki Low Review Rate lebih tinggi, namun penyebab hubungan tersebut **tidak dapat ditentukan** dari data yang tersedia.
- Pada kategori berisiko, keluhan **kualitas produk** lebih dominan (**40,5%** vs 29,3% di kategori lain), sedangkan keluhan keterlambatan lebih kecil (10,6% vs 13,6%).
- **Seller berisiko tidak dijelaskan oleh kategori yang dijualnya.** Berdasarkan kategori produknya, Low Review Rate seller berisiko seharusnya sekitar **9,0%**, tetapi kenyataannya **15,2%** (±1,7× perkiraan). Seller lain justru sedikit di bawah perkiraannya (6,6% vs 7,3%).

**Implikasi:** ulasan buruk pada pesanan tepat waktu berkaitan dengan **dua hal yang berbeda**:
- **kategori produk tertentu**, yaitu masalah umum di kategori tersebut, terutama kualitas;
- **seller tertentu**, yaitu masalah khusus seller yang tidak dijelaskan oleh kategori yang dijualnya. Keduanya ditindaklanjuti di Rekomendasi 3: evaluasi produk dimulai dari seller berisiko, sedangkan standar kualitas dan kelengkapan listing ditetapkan per kategori berisiko.

**Catatan:** kategori adalah atribut produk, sehingga temuan ini hanya menunjukkan **keterkaitan** dan tidak membuktikan bahwa kategori tersebut lebih buruk kualitasnya. Hasil ini juga tidak menilai harga produk.

#### g. Dari Label ke Tindakan

Bagian b–f membaca label per sumber, per kelompok, dan per kategori. Bagian ini membalik sudut pandangnya: **untuk setiap label Pain Points**, dari pesanan seperti apa keluhan itu datang, dan siapa yang dapat memperbaikinya. Profil setiap label dilihat dari:
- **sumber pesanan**: tepat waktu, telat, atau tidak sampai;
- **jenis pesanan**: porsi pesanan kompleks (2+ barang atau > 1 seller);
- **posisi barang saat ulasan ditulis** (Review Stage): belum diterima atau baru diterima sebagian;
- **pihak yang disalahkan** (Blame Target), hanya dari keluhan yang menyebut pihak tertentu. Label ini hanya gambaran persepsi pelanggan (akurasi QA 79%);
- **konsentrasi seller**: porsi keluhan yang berasal dari 10% seller dengan keluhan terbanyak (pesanan 1 seller), serta dari 30 seller berisiko (d).

````python
named_party = ~df_comment['Blame Target'].isin(['Not Mentioned', 'Not Relevant'])
grouped = df_comment.groupby('Pain Points')


def top_seller_share(group):
    # porsi keluhan dari 10% seller dengan keluhan terbanyak (hanya pesanan 1 seller)
    counts = group.loc[group['n_sellers'] == 1, 'seller_id'].value_counts()
    top_n = int(np.ceil(len(counts) * 0.1))
    return counts.head(top_n).sum() / counts.sum() * 100 if len(counts) else np.nan


label_profile = pd.DataFrame({
    'jumlah_ulasan': grouped.size(),
    'porsi_%': grouped.size() / len(df_comment) * 100,
    'dari_tepat_waktu_%': grouped['sumber'].apply(lambda s: (s == 'Sampai tepat waktu').mean() * 100),
    'dari_telat_%': grouped['sumber'].apply(lambda s: (s == 'Sampai, tapi telat').mean() * 100),
    'dari_tidak_sampai_%': grouped['sumber'].apply(lambda s: (s == 'Tidak sampai').mean() * 100),
    'pesanan_kompleks_%': grouped.apply(lambda g: ((g['n_items'] > 1) | (g['n_sellers'] > 1)).mean() * 100),
    'belum_diterima_%': grouped['Review Stage'].apply(lambda s: (s == 'Not Received Yet').mean() * 100),
    'baru_diterima_sebagian_%': grouped['Review Stage'].apply(lambda s: (s == 'Partially Received').mean() * 100),
    'dari_10%_seller_teratas_%': grouped.apply(top_seller_share),
    'dari_seller_berisiko_%': grouped['seller_id'].apply(lambda s: s.isin(risky_sellers).mean() * 100),
})
blame_by_label = pd.crosstab(df_comment.loc[named_party, 'Pain Points'], df_comment.loc[named_party, 'Blame Target'], normalize='index').mul(100)
label_profile['menyalahkan_seller_%'] = blame_by_label['Seller']
label_profile['menyalahkan_kurir_%'] = blame_by_label['Logistics / Carrier']
label_profile = label_profile.reindex(pain_order).round(1)
label_profile
````

Dua label perlu dipecah lebih lanjut karena isinya campuran:
- **Wrong / Missing Item** mencakup barang yang **kurang** (pesanan belum lengkap) dan barang yang **salah atau tidak sesuai deskripsi**. Keduanya membutuhkan tindakan yang berbeda.
- **Seller Service** banyak berasal dari pesanan tidak sampai, sehingga perlu dilihat status pesanannya.

````python
wrong_missing = df_comment[df_comment['Pain Points'] == 'Wrong / Missing Item'].copy()
wrong_missing['jenis_pesanan'] = np.where((wrong_missing['n_items'] > 1) | (wrong_missing['n_sellers'] > 1),
                                          'Pesanan kompleks (2+ barang / > 1 seller)', 'Pesanan 1 barang')
received_stages = ['Received', 'Received, Not Used Yet', 'Used / Tested']
wm_split = wrong_missing.groupby('jenis_pesanan').agg(
    jumlah_ulasan=('review_id', 'count'),
    baru_diterima_sebagian_pct=('Review Stage', lambda s: (s == 'Partially Received').mean() * 100),
    sudah_diterima_pct=('Review Stage', lambda s: s.isin(received_stages).mean() * 100),
)
wm_split.insert(1, 'porsi_%', wm_split['jumlah_ulasan'] / wm_split['jumlah_ulasan'].sum() * 100)
print('Wrong / Missing Item per jenis pesanan:')
display(wm_split.round(1))

seller_service_status = (
    df_comment.loc[(df_comment['Pain Points'] == 'Seller Service') & (df_comment['sumber'] == 'Tidak sampai'), 'order_status']
    .value_counts().to_frame('jumlah_ulasan')
)
seller_service_status['porsi_%'] = (seller_service_status['jumlah_ulasan'] / seller_service_status['jumlah_ulasan'].sum() * 100).round(1)
print('Seller Service pada pesanan tidak sampai, per status pesanan:')
seller_service_status
````

**Estimasi dampak (berbasis skenario).** Dua label terbesar di luar keterlambatan dapat dihubungkan ke kelompok pesanan dari bagian d:
- **Product Quality** terkonsentrasi pada seller berisiko. Skenario: Low Review Rate pesanan 1 barang tepat waktu dari 30 seller berisiko turun ke rata-rata pesanan tepat waktu.
- **Wrong / Missing Item** terkonsentrasi pada pesanan kompleks. Skenario: Low Review Rate pesanan kompleks yang tepat waktu turun ke rata-rata pesanan tepat waktu. Angka ini adalah **batas atas**, karena tidak semua keluhan pesanan kompleks dapat dicegah dengan pengiriman yang lengkap.

> Ulasan buruk berkurang = jumlah pesanan kelompok × (Low Review Rate kelompok − rata-rata pesanan tepat waktu)

````python
scenario_groups = {
    'Seller berisiko (Product Quality)': ['1 barang, seller berisiko'],
    'Pesanan kompleks (Wrong / Missing Item, batas atas)': ['Multi-seller', '1 seller, 2+ barang'],
}
label_impact = []
for name, segments in scenario_groups.items():
    rows = on_time_breakdown.loc[segments]
    reduction = (rows['jumlah_pesanan'] * (rows['low_review_rate_%'] / 100 - on_time_rate)).sum()
    label_impact.append({
        'skenario': name,
        'jumlah_pesanan': rows['jumlah_pesanan'].sum(),
        'ulasan_buruk_saat_ini': rows['jumlah_low_review'].sum(),
        'target_low_review_rate_%': round(on_time_rate * 100, 2),
        'ulasan_buruk_berkurang': round(reduction),
    })
pd.DataFrame(label_impact).set_index('skenario')
````

**Insight (profil setiap label):**
- **Delivery Time (35,2%)**: 81% berasal dari pesanan telat atau tidak sampai, dan 87% ditulis saat barang belum diterima. Ini masalah logistik yang sudah ditangani Pilar 2–3 dan Rekomendasi 1–2.
- **Wrong / Missing Item (29,7%)** sebenarnya berisi **dua masalah berbeda**:
  - **47%** berasal dari pesanan kompleks, dan **88%** di antaranya ditulis saat barang **baru diterima sebagian**. Barang **kurang** karena tiba bertahap.
  - **53%** berasal dari pesanan 1 barang, dan **86%** di antaranya barangnya sudah diterima. Barang yang diterima **salah atau tidak sesuai deskripsi**.
- **Product Quality (15,9%)**: hampir semuanya dari pesanan tepat waktu (96,5%), dan merupakan keluhan yang paling terkonsentrasi. **10% seller teratas menyumbang 50%** keluhan kualitas, dan 27,9% berasal dari 30 seller berisiko saja.
- **Seller Service (5,6%)**: 54,6% berasal dari pesanan tidak sampai, terutama berstatus **unavailable (42%)** dan **canceled (27%)**. Artinya stok tidak akurat atau pesanan dibatalkan. Seller paling banyak disalahkan (84%).
- **Delivery Process (5,4%)**: satu-satunya keluhan yang paling banyak **menyalahkan kurir (56,5%)**, misalnya soal pelacakan atau barang harus diambil sendiri.
- **Packaging dan Cost & Billing** (masing-masing 1,2%) volumenya kecil. **Others dan Not Relevant** (5,7%) tidak spesifik, sehingga tidak ditindaklanjuti tersendiri.
- **Potensi (skenario):** jika Low Review Rate seller berisiko turun ke rata-rata pesanan tepat waktu (9,25%), **±498 ulasan buruk** dapat dicegah. Untuk pesanan kompleks, batas atasnya **±1.380**.

## 5.2 Pilar 2 — WHERE & WHEN: Di Mana dan Kapan Keterlambatan Terjadi?

**Pertanyaan:** di rute, jarak, wilayah, dan bulan apa risiko keterlambatan pengiriman paling tinggi?

**Metode:** sama seperti Pilar 1, tetapi metrik yang dibandingkan adalah **Late Delivery Rate** terhadap **baseline 6,79%**, dengan uji Chi-Square (α = 0,05):
- **Ho:** kelompok tidak berkaitan dengan keterlambatan.
- **Ha:** kelompok berkaitan dengan keterlambatan.

**Data:**
- Analisis waktu dan wilayah pelanggan memakai seluruh **pesanan delivered** (`is_delivered`).
- Analisis rute dan jarak memakai **pesanan delivered 1 seller** yang memiliki lokasi seller (langkah 4.7), karena hanya pesanan ini yang memiliki satu rute pengiriman yang jelas. State dan koordinat seller sudah dikoreksi dengan **kode pos sebagai acuan** (3.7), sehingga asal pengiriman setiap rute sesuai dengan lokasi seller yang sebenarnya.

````python
df_delivered = df_order[df_order['is_delivered']].copy()
df_route = df_delivered[df_delivered['seller_state'].notna()].copy()
baseline_late = df_delivered['is_late'].mean() * 100

print(f'Pesanan delivered                 : {df_delivered.shape[0]}')
print(f'Pesanan delivered dengan rute      : {df_route.shape[0]}')
print(f'Baseline Late Delivery Rate        : {baseline_late:.2f}%')
````

````python
def late_rate_by(data, group_col):
    # Late Delivery Rate per kelompok, beserta median lama pengiriman dan Low Review Rate
    summary = data.groupby(group_col, observed=True).agg(
        jumlah_pesanan=('order_id', 'count'),
        jumlah_telat=('is_late', 'sum'),
        median_lama_kirim_hari=('delivery_days', 'median'),
        low_review_rate_pct=('is_low_review', 'mean'),
    )
    summary['late_rate_%'] = (summary['jumlah_telat'] / summary['jumlah_pesanan'] * 100).round(2)
    summary['rasio_vs_baseline'] = (summary['late_rate_%'] / baseline_late).round(2)
    summary['low_review_rate_%'] = (summary.pop('low_review_rate_pct') * 100).round(1)
    summary['median_lama_kirim_hari'] = summary['median_lama_kirim_hari'].round(1)
    return summary


def chi_square_late(data, group_col, alpha=0.05):
    # uji Chi-Square: apakah kelompok berkaitan dengan keterlambatan
    table = pd.crosstab(data[group_col], data['is_late'].astype(bool))
    stat, pvalue, dof, expected = chi2_contingency(table)
    pvalue_text = '< 1e-300' if pvalue == 0 else f'= {pvalue:.3g}'
    if pvalue < alpha:
        print(f'p-value {pvalue_text} < {alpha}. Tolak Ho: {group_col} berkaitan dengan keterlambatan.')
    else:
        print(f'p-value {pvalue_text} >= {alpha}. Gagal menolak Ho: {group_col} tidak terbukti berkaitan dengan keterlambatan.')
    return pvalue


def plot_late_rate(summary, title, ylabel):
    # bar chart horizontal Late Delivery Rate per kelompok, dengan garis baseline
    fig, ax = plt.subplots(figsize=(9, 0.6 * len(summary) + 1.5))
    colors = ['#d62728' if rate > baseline_late else '#9e9e9e' for rate in summary['late_rate_%']]
    ax.barh(summary.index.astype(str), summary['late_rate_%'], color=colors)
    ax.axvline(baseline_late, color='black', linestyle='--', linewidth=1)
    ax.text(baseline_late, len(summary) - 0.4, f' baseline {baseline_late:.1f}%', va='bottom')
    for i, (rate, n) in enumerate(zip(summary['late_rate_%'], summary['jumlah_pesanan'])):
        ax.text(rate + 0.2, i, f'{rate:.1f}%  (n={n:,})'.replace(',', '.'), va='center')
    ax.set_title(title)
    ax.set_xlabel('Late Delivery Rate (%)')
    ax.set_ylabel(ylabel)
    ax.set_xlim(0, max(summary['late_rate_%'].max() * 1.3, baseline_late * 1.5))
    ax.invert_yaxis()
    plt.tight_layout()
    plt.show()
````

Selain Late Delivery Rate, setiap tabel juga menampilkan **median lama pengiriman** dan **Low Review Rate**, agar terlihat apakah kelompok yang sering telat juga lebih sering mendapat ulasan buruk (menghubungkan Pilar 2 dengan Pilar 1).

### 5.2.1 Sama-State vs Beda-State

````python
df_route['jenis_rute'] = np.where(df_route['same_state'], 'Sama-state', 'Beda-state')

same_state_summary = late_rate_by(df_route, 'jenis_rute')
plot_late_rate(same_state_summary, f"Pengiriman beda-state lebih sering telat: {same_state_summary.loc['Beda-state', 'late_rate_%']:.1f}% "
               f"vs {same_state_summary.loc['Sama-state', 'late_rate_%']:.1f}%\n(Late Delivery Rate: sama-state vs beda-state)", 'Jenis Rute')
chi_square_late(df_route, 'jenis_rute');
same_state_summary
````

**Keterangan istilah:**

- **Late Delivery Rate** = pesanan yang sampai setelah tanggal estimasi ÷ pesanan delivered (dibandingkan per tanggal)
- **Baseline (garis putus-putus)** = Late Delivery Rate seluruh pesanan delivered (6,79%)
- **Merah / abu-abu** = kelompok dengan Late Delivery Rate di atas / di bawah baseline; n = jumlah pesanan delivered
- **Sama-state / beda-state** = seller dan pelanggan berada di state yang sama / berbeda; hanya pesanan 1 seller yang lokasi seller-nya diketahui

**Insight:**
- Pengiriman **beda-state** memiliki Late Delivery Rate **8,2%**, hampir **1,8× lipat** pengiriman sama-state (**4,6%**).
- Median lama pengiriman beda-state (±12,8 hari) juga **hampir dua kali** sama-state (±6,5 hari), dan Low Review Rate-nya lebih tinggi (13,7% vs 10,0%).
- Pengiriman beda-state mencakup **±64%** pesanan, sehingga masalah ini menyangkut mayoritas transaksi.
- Uji Chi-Square: **Tolak Ho**.

### 5.2.2 Jarak Seller–Pelanggan

````python
distance_summary = late_rate_by(df_route, 'distance_group')
plot_late_rate(distance_summary, f"Risiko telat naik seiring jarak, tertinggi di atas 1.500 km ({distance_summary['late_rate_%'].iloc[-1]:.1f}%)\n"
               '(Late Delivery Rate per jarak seller–pelanggan)', 'Jarak')
chi_square_late(df_route.dropna(subset=['distance_group']), 'distance_group');
distance_summary
````

**Keterangan istilah:**

- **Late Delivery Rate** = pesanan yang sampai setelah tanggal estimasi ÷ pesanan delivered (dibandingkan per tanggal)
- **Baseline (garis putus-putus)** = Late Delivery Rate seluruh pesanan delivered (6,79%)
- **Merah / abu-abu** = kelompok dengan Late Delivery Rate di atas / di bawah baseline; n = jumlah pesanan delivered
- **Jarak** = jarak garis lurus (Haversine) antara koordinat kode pos seller dan pelanggan, bukan jarak tempuh jalan

**Insight:**
- Late Delivery Rate naik seiring jarak: dari **4,5% (< 100 km)** menjadi **11,7% (> 1.500 km)**, atau **±2,6× lipat**. Median lama pengiriman juga naik dari ±5 hari menjadi ±18 hari.
- Lonjakan paling tajam terjadi pada jarak **> 1.500 km**: dari ±7,5% (700–1.500 km) ke **11,7%**. Pada jarak di bawah 1.500 km, Late Delivery Rate masih relatif dekat dengan baseline.
- Low Review Rate ikut naik (10,0% → 15,7%), sejalan dengan temuan Pilar 1.
- Uji Chi-Square: **Tolak Ho**.

**Implikasi:** estimasi pengiriman untuk jarak sangat jauh tampaknya **belum cukup memperhitungkan** lamanya perjalanan, sehingga pengiriman > 1.500 km paling sering melewati janji.

### 5.2.3 Wilayah Tujuan dan Rute Antar-Region

````python
region_summary = late_rate_by(df_delivered, 'customer_region').sort_values('late_rate_%', ascending=False)
plot_late_rate(region_summary, f"{region_summary.index[0]} adalah tujuan yang paling sering telat ({region_summary['late_rate_%'].iloc[0]:.1f}%)\n"
               '(Late Delivery Rate per region pelanggan)', 'Region Pelanggan')
chi_square_late(df_delivered, 'customer_region');
region_summary
````

**Keterangan istilah:**

- **Late Delivery Rate** = pesanan yang sampai setelah tanggal estimasi ÷ pesanan delivered (dibandingkan per tanggal)
- **Baseline (garis putus-putus)** = Late Delivery Rate seluruh pesanan delivered (6,79%)
- **Merah / abu-abu** = kelompok dengan Late Delivery Rate di atas / di bawah baseline; n = jumlah pesanan delivered
- **Region** = 5 wilayah resmi Brasil (IBGE): Norte, Nordeste, Centro-Oeste, Sudeste, dan Sul

Rute antar-region hanya ditampilkan jika memiliki **minimal 500 pesanan**, agar persentase tidak didominasi rute dengan pesanan sangat sedikit (sesuai asumsi minimum jumlah pesanan di Business Understanding).

````python
MIN_ORDERS_REGION_ROUTE = 500

region_route_summary = late_rate_by(df_route, 'route_region')
region_route_summary = region_route_summary[region_route_summary['jumlah_pesanan'] >= MIN_ORDERS_REGION_ROUTE].sort_values('late_rate_%', ascending=False)

print(f'Rute antar-region dengan >= {MIN_ORDERS_REGION_ROUTE} pesanan: {len(region_route_summary)} rute, '
      f'mencakup {region_route_summary["jumlah_pesanan"].sum() / df_route.shape[0] * 100:.1f}% pesanan')
plot_late_rate(region_route_summary, f"Rute menuju Nordeste paling sering telat ({region_route_summary.index[0]} {region_route_summary['late_rate_%'].iloc[0]:.1f}%)\n"
               f'(Late Delivery Rate per rute antar-region, min. {MIN_ORDERS_REGION_ROUTE} pesanan)', 'Rute (region seller → region pelanggan)')
region_route_summary
````

**Keterangan istilah:**

- **Late Delivery Rate** = pesanan yang sampai setelah tanggal estimasi ÷ pesanan delivered (dibandingkan per tanggal)
- **Baseline (garis putus-putus)** = Late Delivery Rate seluruh pesanan delivered (6,79%)
- **Merah / abu-abu** = kelompok dengan Late Delivery Rate di atas / di bawah baseline; n = jumlah pesanan delivered
- **Rute antar-region** = region seller → region pelanggan; hanya rute dengan min. 500 pesanan

**Insight:**
- Region **Nordeste (Timur Laut)** adalah tujuan dengan Late Delivery Rate tertinggi: **12,8%**, hampir **1,9× baseline**, dengan median lama pengiriman ±17 hari dan Low Review Rate **16,8%** (tertinggi di antara region).
- Rute **Sudeste → Nordeste** paling relevan secara bisnis: Late Delivery Rate **13,0%** dan volumenya besar (**±7.500 pesanan**). Rute Sul → Nordeste memiliki rate serupa (±13,3%) tetapi volumenya jauh lebih kecil.
- Sebaliknya, rute di dalam wilayah selatan (Sul → Sul) dan dari Sul/Centro-Oeste ke Sudeste memiliki Late Delivery Rate jauh di bawah baseline (3–5%).
- Uji Chi-Square (region pelanggan): **Tolak Ho**.

**Level state pelanggan.** Region masih cukup luas, sehingga Late Delivery Rate juga dihitung per state pelanggan. Selain rate, dihitung **pesanan telat berlebih** = jumlah pesanan × (Late Delivery Rate state − baseline). Angka ini menunjukkan berapa pesanan telat yang hilang jika state tersebut setara baseline nasional, sehingga state kecil dengan rate tinggi tidak otomatis menjadi prioritas.

````python
state_summary = late_rate_by(df_delivered, 'customer_state')
state_summary['telat_berlebih'] = (state_summary['jumlah_pesanan'] * (state_summary['late_rate_%'] - baseline_late) / 100).round(1)

plot_late_rate(state_summary.sort_values('late_rate_%', ascending=False),
               f"Rate tertinggi di {state_summary['late_rate_%'].idxmax()} ({state_summary['late_rate_%'].max():.1f}%), "
               f"tetapi pesanan telat berlebih terbesar di {state_summary['telat_berlebih'].idxmax()}\n(Late Delivery Rate per state pelanggan)", 'State Pelanggan')
chi_square_late(df_delivered, 'customer_state');
state_summary.sort_values('telat_berlebih', ascending=False).head(8)
````

**Keterangan istilah:**

- **Late Delivery Rate** = pesanan yang sampai setelah tanggal estimasi ÷ pesanan delivered (dibandingkan per tanggal)
- **Baseline (garis putus-putus)** = Late Delivery Rate seluruh pesanan delivered (6,79%)
- **Merah / abu-abu** = kelompok dengan Late Delivery Rate di atas / di bawah baseline; n = jumlah pesanan delivered
- **Pesanan telat berlebih** = jumlah pesanan × (Late Delivery Rate state − baseline), yaitu kelebihan pesanan telat dibanding jika state itu setara baseline

**Insight:**
- **Berdasarkan rate**, state dengan Late Delivery Rate tertinggi ada di Nordeste: **AL (21,5%)** dan **MA (17,5%)**, sejalan dengan temuan region.
- **Berdasarkan pesanan telat berlebih**, prioritas utama justru **RJ (±659 pesanan)**: rate-nya 12,1% (±1,8× baseline) dengan volume sangat besar (12.310 pesanan). Angka ini hampir 4× state berikutnya, BA (±175).
- AL, yang rate-nya tertinggi, hanya menyumbang ±58 pesanan telat berlebih karena pesanannya sedikit (396).

**Implikasi:** RJ adalah tujuan dengan potensi penurunan pesanan telat terbesar. Hal ini sejalan dengan rute **SP → RJ** pada analisis berikutnya.

### 5.2.4 Rute State → State: Risiko vs Dampak

Pada level state, rute dinilai dari dua sisi:
- **Risiko**: Late Delivery Rate rute tersebut.
- **Dampak**: **jumlah pesanan telat** pada rute tersebut. Rute dengan rate sedang tetapi volumenya sangat besar dapat menyumbang keterlambatan lebih banyak daripada rute kecil dengan rate tinggi.

Hanya rute dengan **minimal 200 pesanan** yang dianalisis.

**Kenapa rute dihitung per state, bukan per kota?**

Nama kota seller dan pelanggan sudah dibersihkan dengan kode pos sebagai acuan (3.7d), sehingga rute sebenarnya bisa dihitung lebih rinci, yaitu kota asal → kota tujuan. Sebelum memilih, kedua unit rute dibandingkan dari sisi **seberapa banyak pesanan yang bisa dianalisis**. Nama kota diberi kode state agar kota bernama sama di state berbeda tidak tertukar:

````python
seller_city_map = df_clean.dropna(subset=['seller_id']).drop_duplicates('seller_id').set_index('seller_id')['seller_city']
route_city = (
    df_route['seller_id'].map(seller_city_map) + ' (' + df_route['seller_state'] + ') → '
    + df_route['customer_city'] + ' (' + df_route['customer_state'] + ')'
)

unit_check = []
for unit, routes in [('Kota → kota', route_city), ('State → state', df_route['route_state'])]:
    orders_per_route = routes.value_counts()
    row = {
        'unit_rute': unit,
        'jumlah_rute': orders_per_route.shape[0],
        'persen_rute_1_pesanan': round((orders_per_route == 1).mean() * 100, 1),
    }
    for min_orders in [30, 200]:
        kept = orders_per_route[orders_per_route >= min_orders].index
        row[f'rute_min_{min_orders}_pesanan'] = len(kept)
        row[f'cakupan_pesanan_min_{min_orders}_%'] = round(routes.isin(kept).mean() * 100, 1)
        row[f'cakupan_telat_min_{min_orders}_%'] = round(df_route.loc[routes.isin(kept), 'is_late'].sum() / df_route['is_late'].sum() * 100, 1)
    unit_check.append(row)

pd.DataFrame(unit_check).set_index('unit_rute').astype(object).T
````

**Temuan:**
- Rute per kota pecah menjadi **34.442 pasangan kota**, dan **±73%** di antaranya hanya berisi **1 pesanan**. Persentase telat dari satu atau segelintir pesanan tidak stabil: 1 pesanan telat dari 3 pesanan sudah terbaca 33%.
- Dengan batas minimal yang sama (200 pesanan), rute per kota hanya menyisakan **26 rute** yang mencakup **±14,5% pesanan** (±13,5% pesanan telat). Rute per state menyisakan **49 rute** yang mencakup **±91,5% pesanan** (±91,1% pesanan telat).
- Bahkan dengan batas yang jauh lebih longgar (30 pesanan), rute per kota hanya mencakup ±32% pesanan.

**Keputusan:** rute dianalisis **per state** (state seller → state pelanggan).

**Alasan:**
1. **Datanya cukup.** Setiap rute memiliki pesanan yang cukup untuk menghasilkan rate yang stabil dan dapat diuji, dan hampir seluruh pesanan serta pesanan telat ikut dianalisis. Jika dihitung per kota, ±85% pesanan telat tidak akan masuk ke dalam prioritas.
2. **Sesuai cara paket bergerak.** Paket dikumpulkan di hub dan diangkut antar-wilayah sebelum didistribusikan ke kota tujuan. Hambatan yang ditemukan (transit, jarak, tujuan Nordeste dan RJ) berada pada jalur antar-wilayah.
3. **Sesuai kebutuhan bisnis.** Tim Operasional & Logistik membutuhkan daftar rute yang pendek dan bervolume cukup untuk ditindaklanjuti, misalnya evaluasi mitra logistik atau penyesuaian estimasi, bukan ratusan pasangan kota kecil.

Rute per kota tetap berguna untuk **mendalami** rute yang sudah diprioritaskan, misalnya mencari kota asal mana yang paling sering telat di dalam rute SP → RJ. Hal ini dicatat sebagai future work (6.4).

````python
MIN_ORDERS_STATE_ROUTE = 200

state_route_summary = late_rate_by(df_route, 'route_state')
state_route_summary = state_route_summary[state_route_summary['jumlah_pesanan'] >= MIN_ORDERS_STATE_ROUTE]

print(f'Rute state dengan >= {MIN_ORDERS_STATE_ROUTE} pesanan: {len(state_route_summary)} rute, '
      f'mencakup {state_route_summary["jumlah_pesanan"].sum() / df_route.shape[0] * 100:.1f}% pesanan')
print('\nTop 10 rute berdasarkan RISIKO (Late Delivery Rate):')
state_route_summary.sort_values('late_rate_%', ascending=False).head(10)
````

````python
print('Top 10 rute berdasarkan DAMPAK (jumlah pesanan telat):')
state_route_summary.sort_values('jumlah_telat', ascending=False).head(10)
````

Kedua sudut pandang digabung dalam satu grafik kuadran. Garis horizontal adalah baseline Late Delivery Rate, sedangkan garis vertikal adalah median jumlah pesanan antar rute. **Kuadran kanan atas** berisi rute yang **sering telat sekaligus bervolume besar**, yaitu rute prioritas.

````python
fig, ax = plt.subplots(figsize=(11, 6.5))
volume_median = state_route_summary['jumlah_pesanan'].median()
is_priority = (state_route_summary['late_rate_%'] > baseline_late) & (state_route_summary['jumlah_pesanan'] > volume_median)

ax.scatter(state_route_summary['jumlah_pesanan'], state_route_summary['late_rate_%'],
           s=state_route_summary['jumlah_telat'] / 3 + 10,
           c=np.where(is_priority, '#d62728', '#9e9e9e'), alpha=0.7, edgecolor='white')
ax.axhline(baseline_late, color='black', linestyle='--', linewidth=1)
ax.axvline(volume_median, color='black', linestyle=':', linewidth=1)
ax.set_xscale('log')

for route, row in state_route_summary.iterrows():
    if row['late_rate_%'] > 12 or row['jumlah_telat'] > 200:
        ax.annotate(route, (row['jumlah_pesanan'], row['late_rate_%']), fontsize=8,
                    xytext=(4, 4), textcoords='offset points')

ax.set_title(f'{is_priority.sum()} rute prioritas sering telat sekaligus bervolume besar, terutama SP → RJ\n'
             '(risiko vs volume rute state → state; ukuran titik = jumlah pesanan telat; merah = rute prioritas)')
ax.set_xlabel('Jumlah pesanan (skala log)')
ax.set_ylabel('Late Delivery Rate (%)')
plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **Late Delivery Rate** = pesanan yang sampai setelah tanggal estimasi ÷ pesanan delivered
- **Rute prioritas (merah)** = rute dengan Late Delivery Rate di atas baseline DAN jumlah pesanan di atas median rute (635 pesanan)
- **Garis putus-putus / titik-titik** = baseline Late Delivery Rate (6,79%) / median jumlah pesanan per rute
- **Cakupan** = rute state seller → state pelanggan dengan min. 200 pesanan

````python
# pesanan telat berlebih = jumlah pesanan × (rate rute − baseline), sama seperti analisis state pelanggan
priority_routes = state_route_summary[is_priority].copy()
priority_routes['telat_berlebih'] = (priority_routes['jumlah_pesanan'] * (priority_routes['late_rate_%'] - baseline_late) / 100).round(0)
priority_routes = priority_routes.sort_values('telat_berlebih', ascending=False)
print(f'Rute prioritas: {len(priority_routes)} rute, '
      f'{priority_routes["jumlah_telat"].sum()} pesanan telat '
      f'({priority_routes["jumlah_telat"].sum() / df_route["is_late"].sum() * 100:.1f}% dari seluruh pesanan telat pada data rute)')
print(f"Pesanan telat berlebih pada rute prioritas: {priority_routes['telat_berlebih'].sum():.0f}")
priority_routes
````

Agar lokasi rute prioritas terlihat langsung, ke-11 rute ditampilkan di **peta Brasil**. Setiap rute digambar sebagai garis dari state seller ke state pelanggan:
- **tebal garis** = jumlah pesanan telat berlebih (dasar urutan rute prioritas),
- **warna garis** = Late Delivery Rate rute,
- rute lain (beda state, min. 200 pesanan) ditampilkan tipis abu-abu sebagai pembanding.

Titik setiap state adalah **median koordinat pelanggan** di state tersebut, sehingga mewakili lokasi pesanan, bukan titik tengah wilayah. Batas state diambil dari data resmi IBGE (bagian Referensi).

````python
import json
from matplotlib.cm import ScalarMappable
from matplotlib.lines import Line2D
from matplotlib.patches import Circle, FancyArrowPatch, Polygon, Rectangle

# batas state Brasil (API malha IBGE), hanya dipakai sebagai latar peta
with open(DATA_PATH + 'brazil_uf_ibge.geojson', encoding='utf-8') as f:
    brazil_states = json.load(f)

# titik setiap state = median koordinat pelanggan di state tersebut
state_point = df_order.groupby('customer_state')[['customer_lng', 'customer_lat']].median()

# rute beda state dengan min. 200 pesanan; rute sama state tidak bisa digambar sebagai garis
map_routes = state_route_summary.copy()
map_routes[['asal', 'tujuan']] = map_routes.index.to_series().str.split(' → ', expand=True)
map_routes = map_routes[map_routes['asal'] != map_routes['tujuan']]
map_routes['prioritas'] = map_routes.index.isin(priority_routes.index)
map_routes['telat_berlebih'] = priority_routes['telat_berlebih'].reindex(map_routes.index)

print('Rute beda state yang digambar:', len(map_routes), f"({map_routes['prioritas'].sum()} rute prioritas)")
print('Rute prioritas dari SP        :', priority_routes.index.str.startswith('SP →').sum())
print('Rute prioritas menuju RJ      :', priority_routes.index.str.endswith('→ RJ').sum())
````

````python
cmap = plt.cm.Reds
norm = plt.Normalize(0, priority_routes['late_rate_%'].max())
max_excess = priority_routes['telat_berlebih'].max()
zoom_lng, zoom_lat = (-52, -39), (-28.5, -18)  # area Sudeste–Sul yang diperbesar
route_states = sorted(set(map_routes['asal']) | set(map_routes['tujuan']))


def route_width(excess):
    return 1.5 + 10 * excess / max_excess


def draw_route(ax, origin, destination, width, color, zorder):
    # garis lengkung + panah; lengkungan membuat rute bolak-balik (SP → RJ dan RJ → SP) tidak menumpuk
    ax.add_patch(FancyArrowPatch(
        state_point.loc[origin].values, state_point.loc[destination].values,
        connectionstyle='arc3,rad=0.25', arrowstyle='-|>', mutation_scale=6 + 1.5 * width,
        linewidth=width, color=color, shrinkA=4, shrinkB=4, zorder=zorder,
    ))


def draw_map(ax, label_size, label_offset):
    # latar: batas state
    for feature in brazil_states['features']:
        geometry = feature['geometry']
        polygons = geometry['coordinates'] if geometry['type'] == 'MultiPolygon' else [geometry['coordinates']]
        for polygon in polygons:
            ax.add_patch(Polygon(polygon[0], closed=True, facecolor='#f0f0f0', edgecolor='white', linewidth=1))
    # rute lain sebagai pembanding
    for _, row in map_routes[~map_routes['prioritas']].iterrows():
        draw_route(ax, row['asal'], row['tujuan'], 0.7, '#c4c4c4', 2)
    # rute prioritas: tebal = pesanan telat berlebih, warna = Late Delivery Rate (yang tebal digambar dulu)
    for _, row in map_routes[map_routes['prioritas']].sort_values('telat_berlebih', ascending=False).iterrows():
        draw_route(ax, row['asal'], row['tujuan'], route_width(row['telat_berlebih']), cmap(norm(row['late_rate_%'])), 3)
    # titik dan label state
    ax.scatter(state_point.loc[route_states, 'customer_lng'], state_point.loc[route_states, 'customer_lat'],
               s=18, color='black', zorder=4)
    for state in route_states:
        lng, lat = state_point.loc[state]
        ax.text(lng + label_offset, lat + label_offset, state, fontsize=label_size, fontweight='bold', zorder=5, clip_on=True)
    ax.set_aspect(1 / np.cos(np.radians(15)))  # koreksi skala lintang agar bentuk Brasil tidak gepeng
    ax.axis('off')


fig, (ax_map, ax_zoom) = plt.subplots(1, 2, figsize=(17, 9), gridspec_kw={'width_ratios': [1.15, 1]})

# kiri: seluruh Brasil
draw_map(ax_map, 9, 0.4)
ax_map.set_xlim(-74.5, -33.5)
ax_map.set_ylim(-34.5, 6)
ax_map.add_patch(Rectangle((zoom_lng[0], zoom_lat[0]), zoom_lng[1] - zoom_lng[0], zoom_lat[1] - zoom_lat[0],
                           fill=False, edgecolor='black', linestyle=':', linewidth=1.2, zorder=6))
ax_map.set_title('Seluruh Brasil (kotak titik-titik = area yang diperbesar)')
width_examples = [20, 100, max_excess]
handles = [Line2D([], [], color='#d62728', linewidth=route_width(v), label=f'±{v:.0f} pesanan telat berlebih') for v in width_examples]
handles.append(Line2D([], [], color='#c4c4c4', linewidth=0.7, label=f'Rute lain (min. {MIN_ORDERS_STATE_ROUTE} pesanan)'))
ax_map.legend(handles=handles, loc='lower left', title='Tebal garis (rute prioritas)', frameon=False,
              labelspacing=1.3, handlelength=4, handletextpad=1)

# kanan: Sudeste dan Sul diperbesar
draw_map(ax_zoom, 11, 0.15)
ax_zoom.set_xlim(*zoom_lng)
ax_zoom.set_ylim(*zoom_lat)
ax_zoom.add_patch(Rectangle((0, 0), 1, 1, transform=ax_zoom.transAxes, fill=False, edgecolor='black',
                            linestyle=':', linewidth=1.2, clip_on=False))
ax_zoom.set_title('Diperbesar: Sudeste dan Sul')

# lingkari rute terbesar
top_route = priority_routes.index[0]
top = priority_routes.iloc[0]
total_excess = priority_routes['telat_berlebih'].sum()
middle = state_point.loc[top_route.split(' → ')].mean().values
ax_zoom.add_patch(Circle(middle, radius=1.3, fill=False, edgecolor='black', linestyle='--', linewidth=1.3, zorder=6))
ax_zoom.annotate(f"{top_route}: Late Delivery Rate {top['late_rate_%']:.1f}%\n"
                 f"±{top['telat_berlebih']:.0f} dari ±{total_excess:,.0f} pesanan telat berlebih".replace(',', '.'),
                 xy=(middle[0] + 0.9, middle[1] - 1.0), xytext=(-46.9, -27.7), fontsize=11, zorder=7,
                 bbox=dict(boxstyle='round', facecolor='white', edgecolor='black'),
                 arrowprops=dict(arrowstyle='-', color='black', linewidth=1))
colorbar = fig.colorbar(ScalarMappable(norm=norm, cmap=cmap), ax=ax_zoom, shrink=0.6, pad=0.02)
colorbar.set_label('Late Delivery Rate rute prioritas (%)')

fig.suptitle(f"{priority_routes.index.str.startswith('SP →').sum()} dari {len(priority_routes)} rute prioritas berangkat dari SP, "
             f"dan {top_route} paling tebal (±{top['telat_berlebih'] / total_excess * 100:.0f}% pesanan telat berlebih)\n"
             '(peta rute prioritas state → state; tebal garis = pesanan telat berlebih, warna = Late Delivery Rate)', fontsize=13)
plt.tight_layout(rect=(0, 0, 1, 0.94))
plt.show()
````

**Keterangan istilah:**

- **Rute prioritas** = rute state → state dengan Late Delivery Rate di atas baseline (6,79%) DAN jumlah pesanan di atas median rute (635 pesanan); dipilih dari grafik kuadran di atas
- **Pesanan telat berlebih (tebal garis)** = jumlah pesanan × (Late Delivery Rate rute − baseline), yaitu kelebihan pesanan telat dibanding jika rute itu setara baseline
- **Late Delivery Rate (warna garis)** = pesanan yang sampai setelah tanggal estimasi ÷ pesanan delivered
- **Arah panah** = dari state seller ke state pelanggan; garis dibuat melengkung agar rute bolak-balik tidak menumpuk
- **Titik state** = median koordinat pelanggan di state tersebut (lokasi pesanan, bukan titik tengah wilayah)
- **Rute lain (abu-abu)** = rute beda state dengan min. 200 pesanan yang bukan rute prioritas; rute sama state (misalnya SP → SP) tidak digambar

**Insight:**
- **Pusatnya di SP.** 8 dari 11 rute prioritas berangkat dari SP dan menyebar ke 5 region: Sudeste (RJ, ES), Sul (SC), Centro-Oeste (GO), Nordeste (BA, CE, PE), dan Norte (PA).
- **RJ adalah tujuan masalah utama.** 3 rute prioritas masuk ke RJ (dari SP, PR, dan MG), dengan total ±675 pesanan telat berlebih. SP → RJ sendiri adalah garis paling tebal: ±606 dari ±1.132 pesanan telat berlebih (±54%).
- **Rute ke Nordeste hampir semerah SP → RJ, tetapi lebih tipis.** SP → CE (13,9%) dan SP → BA (13,1%) memiliki Late Delivery Rate yang mendekati SP → RJ (14,4%). Namun volumenya lebih kecil, sehingga pesanan telat berlebihnya hanya ±143 (BA) dan ±68 (CE).

**Implikasi:** perbaikan rute (Rekomendasi 1) dapat difokuskan pada dua jalur:
1. **Jalur masuk ke RJ** (SP, PR, MG → RJ): ±675 pesanan telat berlebih.
2. **Jalur panjang SP → Nordeste** (BA, CE, PE): ±245 pesanan telat berlebih, dengan risiko per pesanan yang tinggi.

````python
# apakah SP → RJ telat karena jaraknya? bandingkan dengan pengiriman berjarak mirip ke tujuan selain RJ
sp_rj = df_route['route_state'] == 'SP → RJ'
similar_distance = df_route['distance_km'].between(300, 700) & (df_route['customer_state'] != 'RJ')
sp_rj_rate = df_route.loc[sp_rj, 'is_late'].mean() * 100
similar_rate = df_route.loc[similar_distance, 'is_late'].mean() * 100
print(f"Median jarak SP → RJ                          : {df_route.loc[sp_rj, 'distance_km'].median():.0f} km")
print(f'Late Delivery Rate SP → RJ                    : {sp_rj_rate:.1f}%')
print(f'Late Delivery Rate jarak 300–700 km selain RJ : {similar_rate:.1f}% ({similar_distance.sum()} pesanan)')
print(f'Rasio                                         : {sp_rj_rate / similar_rate:.1f}x')
````

**Insight:**
- **SP → RJ** adalah rute dengan **dampak terbesar** setelah SP → SP: **±8.000 pesanan** dengan Late Delivery Rate **14,4%** (±2,1× baseline), menghasilkan **±1.150 pesanan telat**. Low Review Rate pada rute ini juga tinggi (**19,9%**).
- Menariknya, median jarak SP → RJ hanya **±383 km**. Pengiriman lain pada jarak 300–700 km (selain tujuan RJ) memiliki Late Delivery Rate hanya **±5,1%**, hampir 3× lebih rendah. Artinya, keterlambatan ke RJ **tidak dijelaskan oleh jarak**, melainkan kemungkinan berkaitan dengan kondisi pengiriman di tujuan tersebut (data tidak dapat menjelaskan penyebab pastinya).
- **SP → SP** menyumbang pesanan telat terbanyak secara jumlah (±1.400), tetapi Late Delivery Rate-nya rendah (4,7%). Banyaknya keterlambatan hanya karena volumenya sangat besar, sehingga **bukan** rute prioritas.
- Rute dengan risiko tertinggi adalah dari SP ke state di Nordeste (**AL 24%, MA 19%, SE 17%, PI 17%**), tetapi volume masing-masing kecil.
- Terdapat **11 rute prioritas** (kuadran kanan atas) yang secara bersama menyumbang **±40% dari seluruh pesanan telat** pada data rute, padahal hanya mencakup ±23% pesanan. Rute ini didominasi pengiriman **dari SP** ke **RJ, BA, ES, SC, CE**, dan state lain.
- Daftar rute prioritas diurutkan berdasarkan **pesanan telat berlebih** (jumlah pesanan × (rate − baseline)), sama seperti analisis state pelanggan. **SP → RJ** sendiri menyumbang **±606** dari ±1.130 pesanan telat berlebih pada 11 rute prioritas (lebih dari separuh). Sebaliknya, **MG → RJ, RJ → SP, dan SP → GO** hanya ±12–17 pesanan di atas baseline, sehingga berada di urutan terbawah dan cukup dipantau.

Peta panas berikut merangkum Late Delivery Rate seluruh rute state → state dengan minimal 200 pesanan, sehingga pola asal–tujuan terlihat sekaligus (sel kosong = pesanan kurang dari 200):

````python
route_pairs = df_route.groupby(['seller_state', 'customer_state']).agg(
    jumlah_pesanan=('order_id', 'count'),
    late_rate=('is_late', 'mean'),
)
heat_table = (
    route_pairs.loc[route_pairs['jumlah_pesanan'] >= MIN_ORDERS_STATE_ROUTE, 'late_rate']
    .astype(float).mul(100).round(0)
    .unstack()
)

fig, ax = plt.subplots(figsize=(15, 4))
sns.heatmap(heat_table, annot=True, fmt='.0f', cmap='Reds', cbar_kws={'label': 'Late Delivery Rate (%)'}, ax=ax)
ax.set_title(f'Rute paling merah adalah dari SP ke Nordeste dan rute yang masuk ke RJ\n'
             f'(Late Delivery Rate per rute state → state, %, min. {MIN_ORDERS_STATE_ROUTE} pesanan)')
ax.set_xlabel('State pelanggan')
ax.set_ylabel('State seller')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **Angka di sel** = Late Delivery Rate rute tersebut (%), yaitu pesanan yang sampai setelah tanggal estimasi ÷ pesanan delivered
- **Sel kosong** = rute dengan kurang dari 200 pesanan, tidak ditampilkan agar rate-nya tidak menyesatkan

**Insight:**
- Sel paling merah adalah rute **dari SP ke Nordeste**: SP → AL (24%), SP → MA (19%), SP → SE dan SP → PI (17%).
- **Masalah RJ tidak hanya dari SP.** Rute lain menuju RJ juga di atas baseline: PR → RJ (12%), SC → RJ (9%), dan MG → RJ (8%). Sebaliknya, pengiriman di dalam RJ sendiri (RJ → RJ) hanya ±5%. Ini mendukung bahwa kendalanya ada pada pengiriman **masuk ke RJ** dari state lain, bukan pada jaraknya (lihat insight SP → RJ di atas).

Lonjakan Februari–Maret 2018 diperiksa lebih rinci **per minggu barang diserahkan ke kurir**. Tanggal serah ke kurir menunjukkan **kapan paket berada di perjalanan**, sehingga terlihat kapan tepatnya transit melambat dan kapan pulih kembali.

````python
in_window = df_delivered['order_delivered_carrier_date'].between('2018-01-01', '2018-04-29 23:59:59')
df_window = df_delivered[in_window & df_delivered['is_valid_stage']].copy()
df_window['minggu_serah'] = df_window['order_delivered_carrier_date'].dt.to_period('W').dt.start_time

weekly = df_window.groupby('minggu_serah').agg(
    jumlah_pesanan=('order_id', 'count'),
    median_transit_hari=('carrier_transit_days', 'median'),
    late_rate_pct=('is_late', 'mean'),
)
weekly['late_rate_%'] = (weekly.pop('late_rate_pct') * 100).round(1)
weekly['median_transit_hari'] = weekly['median_transit_hari'].round(1)
weekly.index = weekly.index.strftime('%d %b %Y')

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(13, 7), sharex=True)
ax1.plot(weekly.index, weekly['median_transit_hari'], color='#d62728', marker='o')
ax1.set_ylabel('Median transit kurir (hari)')
ax1.set_ylim(0, weekly['median_transit_hari'].max() + 3)

ax2.bar(weekly.index, weekly['late_rate_%'], color='#ff9896')
ax2.axhline(baseline_late, color='black', linestyle='--', linewidth=1)
ax2.set_ylabel('Late Delivery Rate (%)')
ax2.set_xlabel('Minggu barang diserahkan ke kurir (tanggal awal minggu)')
plt.setp(ax2.get_xticklabels(), rotation=45, ha='right')

fig.suptitle(f"Transit melambat sejak Februari 2018, dan Late Delivery Rate memuncak di {weekly['late_rate_%'].max():.1f}% "
             f"(paket diserahkan minggu {weekly['late_rate_%'].idxmax()})\n"
             '(median transit dan Late Delivery Rate per minggu serah ke kurir, Jan–Apr 2018)', fontsize=12)
plt.tight_layout()
plt.show()

weekly
````

**Keterangan istilah:**

- **Median transit kurir** = median hari dari barang diserahkan ke kurir sampai diterima pelanggan
- **Late Delivery Rate** = pesanan yang sampai setelah tanggal estimasi ÷ pesanan delivered; garis putus-putus = baseline
- **Minggu serah ke kurir** = minggu saat seller menyerahkan barang ke kurir, bukan minggu pembelian

Pada Maret 2018, transit melambat **dan** estimasi memendek. Untuk melihat mana yang lebih berpengaruh, Late Delivery Rate dihitung ulang dengan asumsi estimasi diperpanjang 2 dan 4 hari (±4 hari = selisih median estimasi Januari dan Maret). Perbandingan tetap dilakukan per tanggal, sesuai definisi keterlambatan:

````python
sensitivity = {}
for month in ['2018-02', '2018-03']:
    data = df_delivered[df_delivered['purchase_month'] == month]
    arrived = data['order_delivered_customer_date'].dt.normalize()
    promised = data['order_estimated_delivery_date'].dt.normalize()
    sensitivity[month] = {
        f'estimasi +{extra} hari': ((arrived > promised + pd.Timedelta(days=extra)).mean() * 100).round(1)
        for extra in [0, 2, 4]
    }

pd.DataFrame(sensitivity).T.rename_axis('bulan pembelian')
````

**Insight: penjelasan tiga perubahan drastis**

**1. Oktober → November 2017 (4,2% → 12,4%): bertepatan dengan lonjakan volume Black Friday**
- Volume pesanan naik **±63%** (4.478 → 7.288 pesanan). Black Friday jatuh pada 24 November 2017.
- Semua tahap ikut melambat: seller telat serah naik **7,9% → 12,9%**, median proses seller **2,0 → 2,7 hari**, dan median transit **7,0 → 8,8 hari**.
- Estimasi tidak berubah (±22 hari), sementara waktu di setiap tahap bertambah, sehingga selisih antara janji dan waktu kirim sebenarnya menyempit.
- Saat volume turun di Desember (5.513 pesanan) dan estimasi diperpanjang (27,4 hari), rate turun ke 7,5%.

**2. Januari → Februari → Maret 2018 (5,7% → 14,1% → 19,0%): transit melambat, estimasi justru memendek**
- **Volume relatif stabil:** jumlah pesanan ±6.500–7.000 per bulan.
- **Sisi seller relatif stabil:** median proses seller tetap ±2,1–2,2 hari, dan seller telat serah pada Maret justru turun (8,6%).
- **Perubahan terbesar yang teramati terjadi pada transit.** Mulai paket yang diserahkan ke kurir pada minggu 5 Februari, median transit naik dari ±8 menjadi ±11 hari. Puncaknya ada pada paket yang diserahkan **26 Februari – 18 Maret**: median transit ±11–12 hari dan Late Delivery Rate **23–27%**.
- **Bersifat menyeluruh:** pengiriman sama-state yang jaraknya dekat pun ikut naik, dari 2,9% menjadi 8,5%.
- **Estimasi tidak menyesuaikan, bahkan memendek** dari 25,6 hari (Januari) menjadi 21,3 hari (Maret). Uji sensitivitas menunjukkan bahwa jika estimasi Maret diperpanjang 4 hari, Late Delivery Rate hanya turun ke **13,3%**, masih jauh di atas bulan normal. Artinya, **perubahan terbesar yang teramati terjadi pada transit**, sedangkan estimasi yang memendek berkaitan dengan sekitar **±40%** dari kenaikan (5,7 dari 13,3 poin persen kenaikan sejak Januari, berdasarkan uji sensitivitas).
- **Konteks eksternal (tidak terdapat di data):** periode ini bertepatan dengan libur Karnaval (10–14 Februari 2018) dan mogok singkat pegawai layanan pos nasional pada 12–13 Maret 2018. Karena dataset tidak memiliki data kurir, keterkaitannya **tidak dapat dibuktikan** dan hanya menjadi dugaan (lihat Future Work).

**3. Maret → April 2018 (19,0% → 4,5%): transit pulih**
- Paket yang diserahkan ke kurir mulai awal April kembali memiliki median transit **±7 hari**, bahkan sedikit lebih cepat daripada Januari.
- Seller juga lebih cepat (median proses 1,5 hari, telat serah 6,4%) dan estimasi sedikit lebih panjang (23,4 hari), sehingga jarak antara janji dan waktu kirim sebenarnya kembali lebar.
- Pemulihan yang cepat ini **cocok dengan dugaan adanya gangguan sementara pada tahap transit**, bukan perubahan permanen pada seller atau rute.

**Implikasi:** kedua lonjakan memiliki pola yang berbeda. November 2017 **bertepatan dengan volume puncak** (indikasi tekanan kapasitas), yang dapat diantisipasi karena tanggalnya sudah diketahui. Februari–Maret 2018 **bertepatan dengan perlambatan transit yang tidak diikuti penyesuaian estimasi**. Keduanya mendukung **Rekomendasi 2**: persiapan kapasitas menjelang periode sibuk dan estimasi yang **mengikuti kondisi transit terkini**, bukan angka tetap.

#### Seberapa Besar Persiapan yang Dibutuhkan pada Periode Lonjakan?

Ketiga lonjakan memiliki sifat berbeda, sehingga persiapannya juga berbeda:

| Lonjakan | Sifat | Pertanyaan persiapan |
|---|---|---|
| **November 2017** (Black Friday) | Volume pesanan melonjak | a. Berapa kapasitas kurir tambahan yang dibutuhkan? b. Seller mana yang perlu bersiap? |
| **Februari 2018** (Karnaval) dan **Maret 2018** (mogok layanan pos) | Volume stabil, transit melambat | c. Berapa buffer estimasi yang dibutuhkan? d. Kapan tanda bahaya perlu memicu tindakan? |

**Catatan:** dataset hanya berisi ±100 ribu pesanan publik Olist, belum tentu seluruh pesanan. Karena itu, jumlah paket di bawah adalah **jumlah dalam dataset**. Yang dapat langsung dipakai untuk kondisi sebenarnya adalah **persentase dan rasionya**.

##### a. Kapasitas Kurir saat Black Friday

Kapasitas kurir tidak tercatat di data. Sebagai gantinya, dilihat **seberapa besar lonjakan volume yang selama ini masih tertangani** tanpa menaikkan keterlambatan:
- **Volume** = jumlah paket yang diserahkan ke kurir per minggu.
- **Rasio lonjakan** = volume minggu itu ÷ rata-rata volume 4 minggu sebelumnya. Rasio dipakai, bukan jumlah paket, karena volume Olist terus tumbuh sepanjang 2017–2018.
- **Batas aman** = rasio lonjakan tertinggi pada minggu yang Late Delivery Rate-nya **tidak melebihi baseline**.

````python
# volume paket yang diserahkan ke kurir per minggu
df_carrier = df_delivered[df_delivered['order_delivered_carrier_date'].notna()].copy()
df_carrier['minggu_serah'] = df_carrier['order_delivered_carrier_date'].dt.to_period('W').dt.start_time

weekly_volume = df_carrier.groupby('minggu_serah').agg(
    jumlah_paket=('order_id', 'count'),
    late_rate_pct=('is_late', 'mean'),
    median_transit_hari=('carrier_transit_days', 'median'),
).loc['2017-02-01':'2018-08-13']  # minggu paling awal dan paling akhir datanya belum lengkap
weekly_volume['late_rate_%'] = (weekly_volume.pop('late_rate_pct') * 100).round(1)
weekly_volume['median_transit_hari'] = weekly_volume['median_transit_hari'].round(1)
weekly_volume['rata_4_minggu_sebelumnya'] = weekly_volume['jumlah_paket'].shift(1).rolling(4).mean().round(0)
weekly_volume['rasio_lonjakan'] = (weekly_volume['jumlah_paket'] / weekly_volume['rata_4_minggu_sebelumnya']).round(2)

# batas aman = lonjakan tertinggi yang masih tertangani (Late Delivery Rate minggu itu <= baseline)
safe_ratio = weekly_volume.loc[weekly_volume['late_rate_%'] <= baseline_late, 'rasio_lonjakan'].max()
print(f'Batas aman lonjakan: {safe_ratio:.2f}x rata-rata 4 minggu sebelumnya')

safe_text = f'{safe_ratio:.2f}'.replace('.', ',')
surge_band = pd.cut(weekly_volume['rasio_lonjakan'], [0, 1.1, 1.3, safe_ratio, np.inf],
                    labels=['<= 1,1x', '1,1-1,3x', f'1,3-{safe_text}x', f'> {safe_text}x'])
weekly_volume.groupby(surge_band, observed=True).agg(
    jumlah_minggu=('jumlah_paket', 'count'),
    median_late_rate_pct=('late_rate_%', 'median'),
    median_transit_hari=('median_transit_hari', 'median'),
)
````

````python
# periode Black Friday: volume normal = rata-rata 4 minggu sebelum minggu Black Friday (23 Okt - 19 Nov 2017)
WORKDAYS_PER_WEEK = 5  # serah ke kurir hampir seluruhnya terjadi Senin-Jumat
normal_volume = weekly_volume.loc['2017-10-23':'2017-11-13', 'jumlah_paket'].mean()
safe_capacity = safe_ratio * normal_volume

bf_plan = weekly_volume.loc['2017-11-20':'2017-12-04', ['jumlah_paket', 'late_rate_%', 'median_transit_hari']].copy()
bf_plan['rasio_vs_normal'] = (bf_plan['jumlah_paket'] / normal_volume).round(2)
bf_plan['kapasitas_aman'] = round(safe_capacity)
bf_plan['kekurangan_paket'] = (bf_plan['jumlah_paket'] - safe_capacity).clip(lower=0).round(0)
bf_plan['tambahan_kapasitas_%'] = (bf_plan['kekurangan_paket'] / safe_capacity * 100).round(0)
bf_plan['tambahan_per_hari_kerja'] = (bf_plan['kekurangan_paket'] / WORKDAYS_PER_WEEK).round(0)

weekday_share = (df_carrier['order_delivered_carrier_date'].dt.dayofweek < 5).mean() * 100
print(f'Paket diserahkan Senin-Jumat     : {weekday_share:.1f}%')
print(f'Volume normal (paket/minggu)     : {normal_volume:.0f}')
print(f'Kapasitas aman (paket/minggu)    : {safe_ratio:.2f} x {normal_volume:.0f} = {safe_capacity:.0f}')
bf_plan
````

````python
plot_weeks = weekly_volume.loc['2017-10-02':'2017-12-25']
peak_week = bf_plan['kekurangan_paket'].idxmax()
peak = bf_plan.loc[peak_week]

fig, ax = plt.subplots(figsize=(12, 4.8))
labels = plot_weeks.index.strftime('%d %b')
colors = ['#d62728' if volume > round(safe_capacity) else '#9e9e9e' for volume in plot_weeks['jumlah_paket']]
ax.bar(labels, plot_weeks['jumlah_paket'], color=colors)
ax.axhline(safe_capacity, color='black', linestyle='--', linewidth=1,
           label=f'Kapasitas aman ±{safe_capacity:,.0f} paket ({safe_text}x volume normal)'.replace(',', '.', 1))
ax.axhline(normal_volume, color='black', linestyle=':', linewidth=1,
           label=f'Volume normal ±{normal_volume:,.0f} paket (rata-rata 23 Okt-19 Nov)'.replace(',', '.', 1))
ax.legend(loc='upper left')
for i, (volume, rate) in enumerate(zip(plot_weeks['jumlah_paket'], plot_weeks['late_rate_%'])):
    ax.text(i, volume + 40, f'telat\n{rate:.0f}%', ha='center', fontsize=8)
ax.set_ylim(0, plot_weeks['jumlah_paket'].max() * 1.2)
ax.set_xlabel('Minggu barang diserahkan ke kurir (tanggal awal minggu)')
ax.set_ylabel('Jumlah paket')
gap_text = f"{peak['kekurangan_paket']:,.0f}".replace(',', '.')
ax.set_title(f"Minggu {peak_week:%d %b %Y} melewati kapasitas aman sebesar ±{gap_text} paket "
             f"(+{peak['tambahan_kapasitas_%']:.0f}%), dan Late Delivery Rate naik ke {peak['late_rate_%']:.0f}%\n"
             '(jumlah paket yang diserahkan ke kurir per minggu, Okt-Des 2017)')
plt.setp(ax.get_xticklabels(), rotation=45, ha='right')
plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **Volume normal (garis titik-titik)** = rata-rata paket yang diserahkan ke kurir per minggu pada 4 minggu sebelum minggu Black Friday (23 Okt – 19 Nov 2017)
- **Kapasitas aman (garis putus-putus)** = batas aman (1,59×) × volume normal, yaitu volume tertinggi yang selama ini masih tertangani tanpa menaikkan Late Delivery Rate di atas baseline
- **Merah** = minggu dengan volume di atas kapasitas aman
- **Telat X%** = Late Delivery Rate paket yang diserahkan ke kurir pada minggu tersebut

**Cara menghitung kebutuhan kapasitas dan armada tambahan:**

| Langkah | Rumus | Perhitungan (minggu puncak, 27 Nov 2017) |
|---|---|---|
| 1. Kapasitas aman | batas aman × volume normal | 1,59 × 1.018 = **±1.619 paket/minggu** |
| 2. Kekurangan kapasitas | volume puncak − kapasitas aman | 2.839 − 1.619 = **±1.220 paket/minggu** |
| 3. Tambahan kapasitas (%) | kekurangan ÷ kapasitas aman | 1.220 ÷ 1.619 = **+75%** |
| 4. Tambahan per hari kerja | kekurangan ÷ 5 hari kerja | 1.220 ÷ 5 = **±244 paket/hari** |
| 5. Armada tambahan | tambahan per hari kerja ÷ kapasitas 1 armada per hari | lihat tabel di bawah |

Kapasitas 1 armada per hari tidak tersedia di data, sehingga langkah 5 dicoba dengan beberapa angka asumsi:

````python
# armada tambahan = tambahan paket per hari kerja / kapasitas 1 armada per hari
# kapasitas 1 armada tidak ada di data, sehingga dicoba beberapa angka asumsi
extra_per_day = bf_plan.loc[peak_week, 'tambahan_per_hari_kerja']
fleet_plan = pd.DataFrame({'kapasitas_1_armada_paket_per_hari (asumsi)': [50, 100, 150]})
fleet_plan['armada_tambahan'] = np.ceil(extra_per_day / fleet_plan.iloc[:, 0]).astype(int)
print(f'Tambahan paket per hari kerja pada minggu puncak: {extra_per_day:.0f}')
fleet_plan
````

**Insight (kapasitas kurir):**
- Selama ini lonjakan hingga **1,59×** rata-rata 4 minggu sebelumnya masih tertangani. Pada 75 minggu dengan lonjakan di bawah batas itu, median Late Delivery Rate hanya **3,8–5,1%** dan median transit ±7 hari.
- Satu-satunya minggu yang melewati batas adalah **minggu setelah Black Friday (27 Nov 2017)**: volume **2,79×** normal, Late Delivery Rate **16%**, dan median transit naik ke **9,9 hari**.
- Kekurangan kapasitas tidak berhenti di minggu puncak. Minggu berikutnya (4 Des) masih kekurangan **±364 paket (+22%)** karena sisa tumpukan paket, dan Late Delivery Rate-nya masih 13,9%. Jadi kapasitas tambahan perlu disiapkan selama **±2 minggu** setelah Black Friday.
- **Untuk kondisi sebenarnya, yang dipakai adalah persentasenya:** kapasitas kurir perlu ditambah **±75%** dari kapasitas normal pada minggu puncak dan **±22%** pada minggu berikutnya. Contoh: jika 1 armada mengantar 100 paket/hari, volume dalam dataset membutuhkan **±3 armada tambahan** pada minggu puncak.
- **Asumsi:** batas aman 1,59× dianggap sebagai kapasitas maksimum armada normal, dan kapasitas naik sebanding dengan jumlah armada. Batas ini adalah batas yang **teramati** di data, bukan kapasitas pasti mitra logistik.

##### b. Kesiapan Seller saat Black Friday

Pada November 2017, seller telat serah juga naik (7,9% → 12,9%). Dicek apakah seller yang pesanannya melonjak paling tinggi juga paling sering telat serah. **Lonjakan seller** = pesanan seller pada minggu Black Friday (20–26 Nov 2017) ÷ rata-rata pesanan mingguannya pada 4 minggu sebelumnya. Hanya pesanan 1 seller dengan tanggal serah valid (sama seperti Pilar 3).

````python
df_bf_seller = df_delivered[df_delivered['is_valid_handover']].copy()
df_bf_seller['is_late_handover'] = df_bf_seller['is_late_handover'].astype(bool)

orders_before = df_bf_seller[df_bf_seller['order_purchase_timestamp'].between('2017-10-23', '2017-11-19 23:59:59')].groupby('seller_id')['order_id'].count() / 4
bf_week = df_bf_seller[df_bf_seller['order_purchase_timestamp'].between('2017-11-20', '2017-11-26 23:59:59')].copy()
seller_surge_ratio = bf_week.groupby('seller_id')['order_id'].count() / orders_before

surge_labels = ['<= 1,6x', '1,6-3x', '3-5x', '> 5x']
bf_week['lonjakan_seller'] = pd.cut(bf_week['seller_id'].map(seller_surge_ratio), [0, 1.6, 3, 5, np.inf], labels=surge_labels)
bf_week['lonjakan_seller'] = bf_week['lonjakan_seller'].cat.add_categories('Seller baru').fillna('Seller baru')

seller_surge = bf_week.groupby('lonjakan_seller', observed=True).agg(
    jumlah_seller=('seller_id', 'nunique'),
    jumlah_pesanan=('order_id', 'count'),
    telat_serah_pct=('is_late_handover', 'mean'),
)
seller_surge['telat_serah_%'] = (seller_surge.pop('telat_serah_pct') * 100).round(1)

normal_handover = df_bf_seller.loc[df_bf_seller['purchase_month'] == '2017-10', 'is_late_handover'].mean() * 100
top_sellers = bf_week['seller_id'].value_counts()
sellers_for_half = (top_sellers.cumsum() / top_sellers.sum() < 0.5).sum() + 1
print(f'Seller telat serah Oktober 2017 (normal)           : {normal_handover:.1f}%')
print(f'Seller yang menyumbang 50% pesanan minggu Black Friday: {sellers_for_half} dari {len(top_sellers)}')

bf_week['lonjakan_di_atas_5x'] = bf_week['lonjakan_seller'] == '> 5x'
chi_table = pd.crosstab(bf_week['lonjakan_di_atas_5x'], bf_week['is_late_handover'])
stat, pvalue, dof, expected = chi2_contingency(chi_table)
print(f"Chi-Square lonjakan > 5x vs lainnya: p-value = {pvalue:.3g}. {'Tolak Ho' if pvalue < 0.05 else 'Gagal menolak Ho'}.")
seller_surge
````

**Insight (kesiapan seller):**
- Seller yang pesanannya melonjak **lebih dari 5×** (90 seller, 814 pesanan) telat serah **21,0%**. Angka ini hampir **3× kondisi normal** Oktober 2017 (7,9%) dan jauh di atas seller dengan lonjakan lebih kecil (11,9–13,8%). Uji Chi-Square: **Tolak Ho**.
- **Seller baru** yang belum punya pesanan pada 4 minggu sebelumnya (131 seller) juga cukup sering telat serah (16,1%).
- Pesanan Black Friday **terkonsentrasi**: hanya **59 dari 617 seller** yang menyumbang 50% pesanan pada minggu itu.

**Implikasi:** persiapan seller dapat difokuskan pada dua kelompok kecil:
1. **±60 seller bervolume terbesar**, karena separuh pesanan Black Friday berasal dari mereka.
2. **Seller yang diperkirakan lonjakannya > 5×**, misalnya yang ikut promo besar. Mereka perlu menyiapkan stok dan tenaga packing sebelum Black Friday, dan penyerahan ke kurirnya dipantau setiap hari.

##### c. Buffer Estimasi Tanggal Sampai

Untuk setiap bulan lonjakan, Late Delivery Rate dihitung ulang dengan asumsi estimasi diperpanjang beberapa hari (cara yang sama dengan uji sensitivitas di atas). Dicari buffer terkecil agar Late Delivery Rate kembali **tidak melebihi baseline**.

````python
buffer_rows = []
for month in spike_months:
    data = df_delivered[df_delivered['purchase_month'] == month]
    arrived = data['order_delivered_customer_date'].dt.normalize()
    promised = data['order_estimated_delivery_date'].dt.normalize()
    rate_by_buffer = pd.Series({extra: (arrived > promised + pd.Timedelta(days=extra)).mean() * 100 for extra in range(0, 15)})
    buffer_rows.append({
        'bulan_pembelian': month,
        'median_estimasi_hari': round(data['estimated_days'].median(), 1),
        'late_rate_%': round(rate_by_buffer[0], 1),
        'buffer_3_hari_%': round(rate_by_buffer[3], 1),
        'buffer_5_hari_%': round(rate_by_buffer[5], 1),
        'buffer_agar_<=_baseline_hari': rate_by_buffer[rate_by_buffer <= baseline_late].index.min(),
    })

buffer_table = pd.DataFrame(buffer_rows).set_index('bulan_pembelian')
buffer_table
````

**Insight (buffer estimasi):**
- Agar Late Delivery Rate kembali ke baseline, buffer yang dibutuhkan cukup panjang: **+7 hari** (Nov 2017), **+8 hari** (Feb 2018), dan **+10 hari** (Mar 2018). Angka ini ±31–47% lebih lama dari median estimasi bulan tersebut (21–24 hari).
- Estimasi sepanjang itu berisiko **menurunkan minat beli** (6.3), sehingga buffer **tidak cukup jika dipakai sendirian**.
- Buffer moderat **+3 hari** sudah menurunkan Late Delivery Rate sekitar 3–5 poin persen:
  - Nov 2017: 12,4% → 9,0%
  - Feb 2018: 14,1% → 11,0%
  - Mar 2018: 19,0% → 14,4%
- Sisa keterlambatan ditangani lewat kapasitas kurir dan kesiapan seller (a, b), serta tanda bahaya (d).
- Maret 2018 membutuhkan buffer paling panjang karena estimasinya justru paling pendek (21,3 hari). Minimal, estimasi **tidak boleh diperpendek** saat transit sedang melambat.

##### d. Tanda Bahaya dari Kecepatan Transit

Lonjakan Februari–Maret tidak didahului kenaikan volume, sehingga perlu tanda bahaya lain. Median transit mingguan (dari 5.2.5 di atas) dikelompokkan untuk melihat **pada median transit berapa hari** Late Delivery Rate mulai naik:

````python
transit_band = pd.cut(weekly_volume['median_transit_hari'], [0, 8, 9, 10, np.inf],
                      labels=['<= 8 hari', '8-9 hari', '9-10 hari', '> 10 hari'])
transit_trigger = weekly_volume.groupby(transit_band, observed=True).agg(
    jumlah_minggu=('jumlah_paket', 'count'),
    median_late_rate_pct=('late_rate_%', 'median'),
)
transit_trigger['contoh_minggu'] = weekly_volume.groupby(transit_band, observed=True).apply(
    lambda g: ', '.join(g.index[-3:].strftime('%d %b %y'))
)
transit_trigger
````

**Insight (tanda bahaya):**
- Selama median transit mingguan **≤ 8 hari** (61 dari 80 minggu), median Late Delivery Rate hanya **3,4%**. Begitu melewati 8 hari, Late Delivery Rate naik bertahap: **7,7%** (8–9 hari), **10,6%** (9–10 hari), dan **12,1%** (> 10 hari).
- Jadi **median transit mingguan > 8 hari** dapat dipakai sebagai tanda bahaya.
- Pada lonjakan 2018, tanda ini sudah muncul pada paket yang diserahkan minggu **29 Jan** (8,8 hari). Itu sekitar **4 minggu sebelum** puncak keterlambatan (paket yang diserahkan 26 Feb – 18 Mar).

**Implikasi:** begitu tanda bahaya muncul, langkah yang dapat diambil adalah:
1. tambahkan buffer estimasi untuk pesanan baru,
2. kirim notifikasi proaktif ke pelanggan yang pesanannya berisiko telat,
3. alihkan sebagian pengiriman ke kurir cadangan.

Kejadian terjadwal seperti libur Karnaval dapat disiapkan **sebelum** tanggalnya, tanpa menunggu tanda bahaya.

#### Apakah Ada Pola Hari atau Tanggal Tertentu?

Analisis bulanan dan mingguan di atas belum menjawab apakah permintaan selalu memuncak pada **hari** atau **tanggal** tertentu. Di Indonesia, misalnya, belanja online cenderung ramai setelah gajian di awal bulan. Di Brasil ada aturan serupa:
- gaji bulanan wajib dibayar **paling lambat hari kerja ke-5 bulan berikutnya** (CLT Pasal 459 §1);
- **gaji ke-13** dibayar setengahnya antara Februari dan November, dan sisanya paling lambat 20 Desember (Lei 4.749/1965).

**Cara mengukur:**
- **Indeks harian** = jumlah pesanan pada hari itu ÷ rata-rata pesanan per hari pada bulan yang sama. Indeks 1,2 berarti 20% lebih ramai dari hari rata-rata bulan tersebut. Indeks dipakai, bukan jumlah pesanan, karena volume Olist terus tumbuh dan tanggal 29–31 tidak ada di setiap bulan.
- Hari Black Friday dikeluarkan dari uji karena merupakan pencilan. Agustus 2018 tidak dipakai karena datanya terpotong (3.9).
- Perbedaan antar kelompok diuji dengan **One Way ANOVA** (membandingkan rata-rata indeks 3 kelompok atau lebih).

````python
DAY_NAMES = ['Senin', 'Selasa', 'Rabu', 'Kamis', 'Jumat', 'Sabtu', 'Minggu']

# jumlah pesanan per tanggal pembelian; hari tanpa pesanan dihitung 0
purchase_date = df_order['order_purchase_timestamp'].dt.normalize()
daily_orders = (
    purchase_date.value_counts()
    .reindex(pd.date_range('2017-01-01', '2018-07-31'), fill_value=0)
    .sort_index().to_frame('jumlah_pesanan')
)
daily_orders['bulan'] = daily_orders.index.to_period('M')
daily_orders['indeks'] = daily_orders['jumlah_pesanan'] / daily_orders.groupby('bulan')['jumlah_pesanan'].transform('mean')
daily_orders['hari'] = daily_orders.index.dayofweek
daily_orders['tanggal'] = daily_orders.index.day
daily_orders['periode_tanggal'] = pd.cut(daily_orders['tanggal'], [0, 7, 14, 21, 31],
                                         labels=['Tgl 1-7', 'Tgl 8-14', 'Tgl 15-21', 'Tgl 22-akhir'])
is_weekday = daily_orders['hari'] < 5
daily_orders['hari_kerja_ke'] = is_weekday.groupby(daily_orders['bulan']).cumsum().where(is_weekday)

black_friday = pd.Timestamp('2017-11-24')
bf_orders = daily_orders.loc[black_friday, 'jumlah_pesanan']
normal_day_orders = daily_orders.loc['2017-11-01':'2017-11-23', 'jumlah_pesanan'].median()
df_pattern = daily_orders.drop(black_friday)

print(f'Pesanan pada Black Friday (24 Nov 2017)      : {bf_orders}')
print(f'Median pesanan per hari, 1-23 Nov 2017      : {normal_day_orders:.0f}')
print(f"Indeks Black Friday                         : {daily_orders.loc[black_friday, 'indeks']:.2f}")
print(f'Hari yang dianalisis (tanpa Black Friday)   : {len(df_pattern)}')
````

````python
bf_window = daily_orders.loc['2017-11-01':'2017-12-10']

fig, ax = plt.subplots(figsize=(13, 4.2))
colors = ['#d62728' if day == black_friday else '#9e9e9e' for day in bf_window.index]
ax.bar(bf_window.index.strftime('%d %b'), bf_window['jumlah_pesanan'], color=colors)
ax.axhline(normal_day_orders, color='black', linestyle='--', linewidth=1,
           label=f'Median hari biasa 1-23 Nov ({normal_day_orders:.0f} pesanan)')
ax.text(bf_window.index.get_loc(black_friday), bf_orders + 20, f'{bf_orders:,}'.replace(',', '.'), ha='center', fontsize=10)
ax.set_ylim(0, bf_orders * 1.12)
ax.set_xlabel('Tanggal pembelian')
ax.set_ylabel('Jumlah pesanan')
ax.legend(loc='upper right')
ax.set_title(f"Pada Black Friday masuk {bf_orders:,} pesanan dalam sehari, ±{bf_orders / normal_day_orders:.0f}x hari biasa di November\n".replace(',', '.', 1)
             + '(jumlah pesanan per tanggal pembelian, 1 Nov - 10 Des 2017)')
plt.setp(ax.get_xticklabels(), rotation=90)
plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **Merah** = hari Black Friday (24 Nov 2017)
- **Garis putus-putus** = median pesanan per hari pada 1–23 Nov 2017, sebelum Black Friday
- **Jumlah pesanan** = seluruh pesanan yang dibuat pada tanggal tersebut (semua status pesanan)

````python
weekday_index = df_pattern.groupby('hari')['indeks'].mean()
weekday_index.index = DAY_NAMES
date_index = df_pattern.groupby('tanggal')['indeks'].mean()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 4.5), gridspec_kw={'width_ratios': [1, 2]}, sharey=True)
ax1.bar(weekday_index.index, weekday_index, color=['#d62728' if v > 1 else '#9e9e9e' for v in weekday_index])
ax1.axhline(1, color='black', linestyle='--', linewidth=1)
for i, value in enumerate(weekday_index):
    ax1.text(i, value + 0.02, f'{value:.2f}', ha='center', fontsize=9)
ax1.set_title('Per hari dalam seminggu')
ax1.set_ylabel('Indeks pesanan harian (1 = rata-rata bulan)')

ax2.axvspan(0.5, 7.5, color='#fde0dd', label='Tanggal 1-7 (sekitar masa gajian)')
ax2.bar(date_index.index, date_index, color='#9e9e9e')
ax2.axhline(1, color='black', linestyle='--', linewidth=1)
ax2.set_xticks(range(1, 32))
ax2.set_xlim(0.5, 31.5)
ax2.set_title('Per tanggal dalam sebulan')
ax2.set_xlabel('Tanggal')
ax2.legend(loc='upper left')
ax1.set_ylim(0, max(weekday_index.max(), date_index.max()) + 0.2)

fig.suptitle(f"Pesanan paling ramai pada Senin-Selasa (±{weekday_index.iloc[:2].mean():.2f}x rata-rata) dan paling sepi pada Sabtu "
             f"({weekday_index['Sabtu']:.2f}x), sedangkan awal bulan tidak lebih ramai\n"
             '(indeks pesanan harian per hari dan per tanggal, Jan 2017 - Jul 2018, tanpa Black Friday)', fontsize=12)
plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **Indeks pesanan harian** = jumlah pesanan pada hari itu ÷ rata-rata pesanan per hari pada bulan yang sama; 1 = sama dengan rata-rata bulan tersebut (garis putus-putus)
- **Merah (kiri)** = hari dengan indeks di atas rata-rata
- **Area merah muda (kanan)** = tanggal 1–7, sekitar masa gajian (gaji bulanan dibayar paling lambat hari kerja ke-5)
- **Masa gajian (uji ANOVA di bawah)** = hari kerja ke-1 s.d. ke-5 setiap bulan

````python
from scipy.stats import f_oneway


def anova_index(data, group_col, label):
    # One Way ANOVA: apakah rata-rata indeks harian berbeda antar kelompok
    stat, pvalue = f_oneway(*[group['indeks'] for _, group in data.groupby(group_col, observed=True)])
    result = 'Tolak Ho: rata-rata indeks berbeda' if pvalue < 0.05 else 'Gagal menolak Ho: tidak terbukti berbeda'
    print(f'{label}: p-value = {pvalue:.3g}. {result}.')


weekdays_only = df_pattern[df_pattern['hari'] < 5].copy()
weekdays_only['masa_gajian'] = np.where(weekdays_only['hari_kerja_ke'] <= 5, 'Hari kerja ke-1 s.d. ke-5', 'Hari kerja lainnya')

anova_index(df_pattern, 'hari', 'Hari dalam seminggu')
anova_index(df_pattern, 'periode_tanggal', 'Periode tanggal')
anova_index(weekdays_only, 'masa_gajian', 'Masa gajian (Senin-Jumat saja)')

pd.concat([
    df_pattern.groupby('periode_tanggal', observed=True)['indeks'].agg(['count', 'mean']),
    weekdays_only.groupby('masa_gajian')['indeks'].agg(['count', 'mean']),
]).rename(columns={'count': 'jumlah_hari', 'mean': 'rata_rata_indeks'}).round(3)
````

````python
# cara bayar: kartu kredit dengan cicilan membuat belanja tidak harus menunggu gajian
payment_share = df_order['payment_type_main'].value_counts(normalize=True).mul(100).round(1)
credit_card = df_order['payment_type_main'] == 'credit_card'
installment_of_credit = (df_order.loc[credit_card, 'payment_installments_max'] > 1).mean() * 100
installment_of_all = (credit_card & (df_order['payment_installments_max'] > 1)).mean() * 100

print(f"Pesanan dibayar dengan kartu kredit         : {payment_share['credit_card']:.1f}%")
print(f'Pesanan kartu kredit yang dicicil (> 1 kali): {installment_of_credit:.1f}%')
print(f'Pesanan dicicil dari seluruh pesanan        : {installment_of_all:.1f}%')
````

**Insight (pola permintaan):**
- **Hari Black Friday sangat ekstrem:** **1.176 pesanan** dalam sehari, **±7×** hari biasa di November (median 175 pesanan). Pesanan tetap tinggi ±5 hari setelahnya (±320–500 pesanan per hari, 25–29 Nov), sehingga beban pengiriman terbawa ke minggu berikutnya (sejalan dengan 5.2.5a). Black Friday juga berdekatan dengan batas pembayaran setengah gaji ke-13 (paling lambat November), tetapi keterkaitannya tidak dapat dibuktikan dengan data ini.
- **Pola mingguan jelas dan signifikan** (ANOVA: **Tolak Ho**):
  - Senin dan Selasa paling ramai (**1,13×** rata-rata harian);
  - setelah itu turun bertahap hingga **Sabtu, yang paling sepi (0,77×)**;
  - selisih Senin dan Sabtu ±47%.
- **Pola gajian seperti di Indonesia tidak terlihat** (ANOVA: **gagal menolak Ho**):
  - tanggal 1–7 justru sedikit di bawah rata-rata (0,96×), dan tanggal 22–akhir sedikit di atasnya (1,03×), tetapi perbedaannya tidak signifikan (p = 0,23);
  - hari kerja ke-1 s.d. ke-5 (masa gajian) tidak lebih ramai dari hari kerja lainnya (1,03× vs 1,08×, p = 0,13).
- **Dugaan penyebab:** **75,4%** pesanan dibayar dengan kartu kredit, dan **68,0%** di antaranya dicicil (51,2% dari seluruh pesanan). Dengan cicilan, pelanggan tidak harus menunggu gajian untuk berbelanja. Ini hanya dugaan, karena data tidak mencatat tanggal gajian pelanggan.

Pola hari juga dicek dari sisi operasional: apakah pesanan yang dibuat pada hari tertentu diproses lebih lama atau lebih sering telat?

````python
df_weekday = df_delivered.assign(hari_beli=df_delivered['order_purchase_timestamp'].dt.dayofweek)
weekday_ops = df_weekday.groupby('hari_beli').agg(
    jumlah_pesanan=('order_id', 'count'),
    median_proses_seller_hari=('seller_processing_days', 'median'),
    median_lama_kirim_hari=('delivery_days', 'median'),
    late_rate_pct=('is_late', 'mean'),
    low_review_rate_pct=('is_low_review', 'mean'),
)
weekday_ops['late_rate_%'] = (weekday_ops.pop('late_rate_pct') * 100).round(1)
weekday_ops['low_review_rate_%'] = (weekday_ops.pop('low_review_rate_pct') * 100).round(1)
weekday_ops[['median_proses_seller_hari', 'median_lama_kirim_hari']] = weekday_ops[['median_proses_seller_hari', 'median_lama_kirim_hari']].round(1)
weekday_ops.index = DAY_NAMES

handover_day = df_delivered['order_delivered_carrier_date'].dropna().dt.dayofweek.value_counts(normalize=True).sort_index().mul(100).round(1)
weekday_ops['porsi_serah_ke_kurir_%'] = handover_day.reindex(range(7), fill_value=0).values

chi_square_late(df_weekday, 'hari_beli')
weekday_ops
````

**Insight (sisi operasional):**
- Serah ke kurir hampir hanya terjadi **Senin–Jumat** (Sabtu 1,8%, Minggu 0%). Selasa adalah hari serah tersibuk (**22,3%** paket seminggu).
- Pesanan yang dibuat hari **Jumat** diproses seller paling lama: median **3,2 hari**, dibanding ±1,3–1,5 hari untuk pesanan Senin–Kamis. Pesanan Sabtu menyusul (2,5 hari). Pesanan ini harus menunggu akhir pekan berlalu sebelum diserahkan ke kurir.
- Lama pengiriman pesanan Kamis–Sabtu juga ±2 hari lebih lama (11,0–11,4 hari vs 9,3–9,4 hari untuk Senin–Rabu).
- Late Delivery Rate hanya berbeda tipis antar hari (6,2–7,5%; Chi-Square: **Tolak Ho**). Yang tertinggi adalah pesanan **Senin (7,5%)**, hari dengan pesanan terbanyak. Low Review Rate hampir sama di semua hari (12,3–13,8%).

**Implikasi:**
1. **Kapasitas harian perlu mengikuti pola mingguan, bukan tanggal gajian.** Kapasitas kurir dan tenaga packing seller perlu paling besar pada **Senin–Selasa**, karena hari itu menampung pesanan terbanyak sekaligus sisa pesanan akhir pekan. Kekurangan ±244 paket per hari kerja pada 5.2.5a adalah angka rata-rata. Pada hari Selasa (22,3% paket seminggu), kebutuhannya menjadi **±272 paket** (1.220 × 22,3%).
2. **Pesanan Jumat–Sabtu perlu perhatian khusus pada periode puncak**, misalnya dengan opsi serah atau penjemputan kurir hari Sabtu, karena proses sellernya ±2× lebih lama.
3. **Promo tidak perlu dipusatkan di awal bulan**, karena tidak ada lonjakan masa gajian. Kampanye dapat dipertimbangkan pada hari yang sepi (Jumat–Minggu) agar beban lebih merata, kecuali untuk momen besar seperti Black Friday.

### 5.2.6 Apa yang Dapat Dilakukan pada 11 Rute Prioritas?

Bagian 5.2.4 menentukan **rute mana** yang diprioritaskan. Bagian ini menjawab **tindakan apa** yang cocok untuk setiap rute:

| Langkah | Pertanyaan |
|---|---|
| a | Apa jenis masalah setiap rute: karena jaraknya jauh, atau telat melebihi wajar untuk jaraknya? |
| b | Apakah menaruh stok di gudang dekat tujuan dapat membantu? |
| c | Di kota mana gudang sebaiknya diletakkan, dan kenapa? |
| d | Berapa pesanan telat yang dapat dicegah? |
| e | Apa target Olist untuk setiap rute? |

#### a. Jenis Masalah Setiap Rute

Rute jarak jauh wajar lebih sering telat, sehingga Late Delivery Rate setiap rute dibandingkan dengan **rate wajar menurut jaraknya**:
- **Rate wajar menurut jarak** = Late Delivery Rate rute-rute lain pada kelompok jarak yang sama, ditimbang sesuai sebaran jarak pesanan pada rute tersebut.
- Diuji dengan **one sample z-test proporsi satu sisi**: Ho = Late Delivery Rate rute ≤ rate wajar menurut jaraknya. Sama seperti uji seller dan kategori berisiko (5.1.8), dipakai **α = 0,01** karena 11 rute diuji sekaligus.

````python
route_rows = []
for route in priority_routes.index:
    data = df_route[df_route['route_state'] == route]
    other_rate = df_route[df_route['route_state'] != route].groupby('distance_group', observed=True)['is_late'].mean()
    distance_mix = data['distance_group'].value_counts(normalize=True)
    expected_rate = (distance_mix * other_rate.reindex(distance_mix.index)).sum()
    pvalue = proportions_ztest(count=data['is_late'].sum(), nobs=len(data), value=expected_rate, alternative='larger')[1]
    seller_orders = data['seller_id'].value_counts()
    route_rows.append({
        'rute': route,
        'region_tujuan': data['customer_region'].iloc[0],
        'jumlah_pesanan': len(data),
        'late_rate_%': round(data['is_late'].mean() * 100, 1),
        'telat_berlebih': priority_routes.loc[route, 'telat_berlebih'],
        'median_jarak_km': round(data['distance_km'].median()),
        'rate_wajar_jarak_%': round(expected_rate * 100, 1),
        'p_value': pvalue,
        'median_transit_hari': round(data['carrier_transit_days'].median(), 1),
        'low_review_rate_%': priority_routes.loc[route, 'low_review_rate_%'],
        'seller_50%_pesanan': (seller_orders.cumsum() / seller_orders.sum() < 0.5).sum() + 1,
    })

route_detail = pd.DataFrame(route_rows).set_index('rute')
route_detail['jenis_masalah'] = np.select(
    [route_detail['p_value'] < 0.01, route_detail['median_jarak_km'] > 1500],
    ['Melebihi wajar untuk jaraknya', 'Wajar, karena jarak sangat jauh'],
    default='Wajar untuk jaraknya',
)
route_detail['p_value'] = route_detail['p_value'].map(lambda p: f'{p:.2g}')
route_detail
````

**Insight (jenis masalah):**
- **5 rute telat melebihi wajar untuk jaraknya** (p < 0,01): SP → RJ, PR → RJ, SP → BA, SP → ES, dan SP → SC.
  - Paling mencolok adalah **SP → RJ**: jaraknya hanya ±383 km dan rate wajarnya ±5,3%, tetapi Late Delivery Rate-nya **14,4% (±2,7×)**.
  - Masalah rute-rute ini bukan jarak, melainkan kemungkinan ada di **jalur atau mitra logistik** menuju tujuan. Penyebab pastinya tidak tersedia di data.
- **3 rute telat karena jaraknya sangat jauh** (> 2.000 km, tidak signifikan pada α = 0,01): SP → CE, SP → PE, dan SP → PA. Late Delivery Rate-nya (9,9–13,9%) setara dengan rute lain yang sama jauhnya. Untuk rute ini, yang perlu dikurangi adalah **jaraknya**.
- **3 rute wajar untuk jaraknya**: MG → RJ, RJ → SP, dan SP → GO. Late Delivery Rate-nya hanya sedikit di atas baseline (7,6–8,3%), dengan telat berlebih kecil (12–17 pesanan per rute).
- Pesanan setiap rute terkonsentrasi pada sedikit seller: **8–60 seller** sudah mencakup separuh pesanan rute. Hal ini penting untuk opsi gudang (d).

#### b. Apakah Gudang Dekat Tujuan Dapat Membantu?

Gudang membuat barang sudah berada dekat pelanggan sebelum dipesan, sehingga pengiriman menjadi **jarak pendek di dalam wilayah tujuan**. Data tidak memiliki pesanan yang dikirim dari gudang, sehingga dipakai **gambaran kasar**: Late Delivery Rate pesanan ke wilayah yang sama yang **sudah dikirim dari dekat**. Dua wilayah tujuan dengan rute prioritas terbanyak dicek: **RJ** (3 rute) dan **Nordeste** (3 rute: BA, CE, PE).

````python
to_rj = df_route['customer_state'] == 'RJ'
to_nordeste = df_route['customer_region'] == 'Nordeste'

origin_distance = pd.concat({
    'Tujuan RJ': df_route[to_rj].groupby('distance_group', observed=True)['is_late'].agg(['count', 'mean']),
    'Tujuan Nordeste': df_route[to_nordeste].groupby('distance_group', observed=True)['is_late'].agg(['count', 'mean']),
}, axis=1)
origin_distance.columns = [f'{dest} - {"jumlah_pesanan" if stat == "count" else "late_rate_%"}' for dest, stat in origin_distance.columns]
for col in origin_distance.columns[1::2]:
    origin_distance[col] = (origin_distance[col] * 100).round(1)

# gambaran kasar Late Delivery Rate jika dikirim dari gudang di wilayah tujuan
local_rate = {
    'RJ': df_route.loc[df_route['route_state'] == 'RJ → RJ', 'is_late'].mean(),
    'Nordeste': df_route.loc[to_nordeste & (df_route['distance_km'] <= 700), 'is_late'].mean(),
}
print(f"RJ → RJ (dikirim dari dalam RJ)          : {local_rate['RJ'] * 100:.1f}% ({(df_route['route_state'] == 'RJ → RJ').sum()} pesanan)")
print(f"Tujuan Nordeste, dikirim dari <= 700 km  : {local_rate['Nordeste'] * 100:.1f}% ({(to_nordeste & (df_route['distance_km'] <= 700)).sum()} pesanan)")
print(f"Tujuan Nordeste, dikirim dari > 700 km   : {df_route.loc[to_nordeste & (df_route['distance_km'] > 700), 'is_late'].mean() * 100:.1f}%")
origin_distance
````

**Insight (gudang dekat tujuan):**
- **RJ:** pesanan ke RJ yang dikirim dari < 100 km hanya telat **5,4%**, dan pengiriman di dalam RJ (RJ → RJ) **4,8%**. Begitu dikirim dari 300–700 km (terutama dari SP), Late Delivery Rate melonjak ke **13,5%**. Jadi hambatannya ada pada **jalur masuk ke RJ**, bukan pada pengiriman di dalam RJ.
- **Nordeste:** pesanan ke Nordeste yang dikirim dari ≤ 700 km telat **7,0%**, hampir separuh dari yang dikirim dari > 700 km (**13,1%**).
- Artinya, jika barang **sudah disimpan di gudang dalam wilayah tujuan**, pengiriman ke pelanggan menjadi jarak pendek yang selama ini jarang telat.
- **Catatan:** angka lokal ini hanya gambaran kasar. Pesanan ke Nordeste dari ≤ 700 km hanya 300 pesanan, dan seller lokal belum tentu sama dengan seller dari SP.

#### c. Di Mana Gudang Diletakkan?

Lokasi gudang dipilih dengan cara sederhana: dari **8 kota dengan pesanan terbanyak** di wilayah tersebut, dipilih kota yang membuat **rata-rata jarak ke seluruh pelanggan di wilayah itu paling kecil**. Jarak dihitung dengan rumus Haversine (4.7), dari titik kota kandidat ke koordinat setiap pesanan. Juga dihitung porsi pelanggan yang berada dalam radius 300 km dan 700 km dari gudang, karena pengiriman ≤ 700 km ke Nordeste hanya telat ±7% (b).

````python
def hub_candidates(region_mask, top_n=8):
    # rata-rata jarak dari kota kandidat ke seluruh pesanan di wilayah tersebut
    customers = df_delivered[region_mask & df_delivered['customer_lat'].notna()]
    top_cities = customers['customer_city'].value_counts().head(top_n).index
    city_point = (customers[customers['customer_city'].isin(top_cities)]
                  .groupby(['customer_city', 'customer_state'])[['customer_lat', 'customer_lng']].median())
    rows = []
    for (city, state), point in city_point.iterrows():
        distance = haversine_km(customers['customer_lat'], customers['customer_lng'], point['customer_lat'], point['customer_lng'])
        rows.append({
            'kota_kandidat': f'{city.title()} ({state})',
            'rata_rata_jarak_km': round(distance.mean()),
            'pelanggan_<=300_km_%': round((distance <= 300).mean() * 100, 1),
            'pelanggan_<=700_km_%': round((distance <= 700).mean() * 100, 1),
        })
    return pd.DataFrame(rows).set_index('kota_kandidat').sort_values('rata_rata_jarak_km')


print('Wilayah RJ:')
display(hub_candidates(df_delivered['customer_state'] == 'RJ'))
print('Wilayah Nordeste:')
hub_candidates(df_delivered['customer_region'] == 'Nordeste')
````

**Insight (lokasi gudang):**
- **Gudang RJ: kota Rio de Janeiro.** Rata-rata jaraknya ke seluruh pelanggan RJ paling kecil (**±44 km**), dan **99,9%** pelanggan RJ berada dalam radius 300 km. Kota sekitarnya (Niterói, Duque de Caxias) hampir sama, karena berada di wilayah metropolitan yang sama.
- **Gudang Nordeste: Aracaju (SE)** adalah pilihan terbaik menurut jarak:
  - rata-rata **±520 km** ke seluruh pelanggan Nordeste;
  - **72,9%** pelanggan Nordeste berada dalam radius 700 km, jarak yang selama ini hanya telat ±7%.
- **Maceió (AL)** dan **Recife (PE)** hampir setara (selisih < 55 km, 68,6–71,6% pelanggan dalam radius 700 km), sehingga dapat menjadi alternatif.
- Kota di pinggir wilayah seperti **Fortaleza** dan **São Luís** kurang cocok, karena rata-rata jaraknya jauh lebih besar (±690–1.000 km).
- Pilihan akhir tetap perlu mempertimbangkan biaya sewa dan ketersediaan mitra logistik, yang tidak tersedia di data.

#### d. Berapa Pesanan Telat yang Dapat Dicegah oleh Gudang?

> Pesanan telat yang dicegah = jumlah pesanan rute × porsi pesanan yang dilayani gudang × (Late Delivery Rate rute − Late Delivery Rate lokal)

- **Late Delivery Rate lokal** = gambaran kasar dari langkah b (RJ → RJ untuk gudang RJ; pesanan ke Nordeste dari ≤ 700 km untuk gudang Nordeste).
- **Porsi pesanan yang dilayani gudang = 50% (asumsi).** Gudang tidak mungkin menyimpan semua produk, sehingga diasumsikan hanya menyimpan produk dari seller terbesar yang bersama-sama mencakup ±50% pesanan rute (jumlah sellernya dihitung di bawah).
- **Ulasan buruk yang berkurang** = pesanan telat yang dicegah × porsi pesanan berulasan × selisih Low Review Rate telat vs tepat waktu (5.1.2), sama seperti cara di 5.4.

````python
HUB_SHARE = 0.5  # asumsi porsi pesanan rute yang dilayani dari gudang
low_review_gap = (timeliness_summary.loc['Telat', 'low_review_rate_%'] - timeliness_summary.loc['Tepat waktu', 'low_review_rate_%']) / 100
reviewed_share = df_route.groupby('route_state')['is_valid_review'].mean()

hub_routes = {
    'Gudang Rio de Janeiro': (['SP → RJ', 'PR → RJ', 'MG → RJ'], local_rate['RJ']),
    'Gudang Nordeste': (['SP → BA', 'SP → CE', 'SP → PE'], local_rate['Nordeste']),
}
hub_rows = []
for hub, (routes, rate_local) in hub_routes.items():
    hub_orders = df_route[df_route['route_state'].isin(routes)]
    seller_orders = hub_orders['seller_id'].value_counts()
    print(f'{hub}: {len(hub_orders)} pesanan dari {len(seller_orders)} seller; '
          f'{(seller_orders.cumsum() / seller_orders.sum() < HUB_SHARE).sum() + 1} seller terbesar mencakup {HUB_SHARE:.0%} pesanan')
    for route in routes:
        n_orders = route_detail.loc[route, 'jumlah_pesanan']
        route_rate = route_detail.loc[route, 'late_rate_%'] / 100
        late_avoided = n_orders * HUB_SHARE * (route_rate - rate_local)
        hub_rows.append({
            'gudang': hub,
            'rute': route,
            'jumlah_pesanan': n_orders,
            'late_rate_%': round(route_rate * 100, 1),
            'late_rate_lokal_%': round(rate_local * 100, 1),
            'pesanan_telat_dicegah': round(late_avoided),
            'ulasan_buruk_berkurang': round(late_avoided * reviewed_share[route] * low_review_gap),
        })

hub_impact = pd.DataFrame(hub_rows).set_index(['gudang', 'rute'])
hub_total = hub_impact.groupby(level='gudang')[['jumlah_pesanan', 'pesanan_telat_dicegah', 'ulasan_buruk_berkurang']].sum()
display(hub_impact)
hub_total
````

**Insight (dampak gudang):**
- **Gudang Rio de Janeiro** berpotensi mencegah **±437 pesanan telat** dan **±231 ulasan buruk**, terutama dari SP → RJ (±382). Gudang cukup menyimpan produk dari **90 seller terbesar** (dari 1.313 seller) untuk melayani separuh pesanan ketiga rute.
- **Gudang Nordeste** berpotensi mencegah **±118 pesanan telat** dan **±61 ulasan buruk**, dengan produk dari **54 seller terbesar** (dari 760 seller).
- Kedua gudang (±555 pesanan telat) mencakup sekitar **separuh** target skenario A (±1.132). Sisanya dicapai lewat evaluasi mitra logistik dan estimasi yang realistis (e).
- **Gudang RJ didahulukan**, karena volumenya terbesar dan selisih antara rate rute dan rate lokal paling besar (14,4% vs 4,8%).
- **Asumsi yang paling berpengaruh adalah porsi 50%.** Jika gudang hanya melayani 30% pesanan, dampaknya turun menjadi ±60% dari angka di atas (gudang RJ ±262, gudang Nordeste ±71).
- Gudang Nordeste juga dapat melayani state Nordeste lain di luar rute prioritas (misalnya AL, SE, dan PB), sehingga dampak sebenarnya dapat lebih besar.

## 5.3 Pilar 3 — WHY & WHO: Seberapa Besar Andil Seller, dan Seller Mana yang Perlu Diprioritaskan?

**Pertanyaan:**
1. **WHY:** apakah keterlambatan lebih banyak terjadi di **sisi seller** (sebelum barang diserahkan ke kurir) atau di **tahap transit** (setelah diserahkan)?
2. **WHO:** seller mana yang paling sering terlambat menyerahkan barang dan perlu diprioritaskan untuk pembinaan?

**Data:**
- Perbandingan durasi tahap memakai pesanan delivered dengan durasi tahap yang valid (`is_valid_stage`, langkah 4.4).
- Analisis handover dan seller memakai **pesanan delivered 1 seller dengan data handover valid** (`is_valid_handover`, langkah 4.5).

### 5.3.1 Di Tahap Mana Pesanan Telat Kehilangan Waktu?

Durasi setiap tahap dibandingkan antara pesanan yang **tepat waktu** dan yang **telat**. Distribusi durasi terlihat sangat miring (langkah 3.8), sehingga normalitasnya diuji terlebih dahulu. Jika tidak normal, digunakan **median** dan uji **Mann-Whitney** (uji non-parametrik untuk membandingkan dua kelompok):
- **Ho:** durasi tahap pada pesanan telat dan tepat waktu tidak berbeda.
- **Ha:** durasi tahap pada pesanan telat lebih lama.

````python
from scipy.stats import mannwhitneyu

df_stage = df_order[df_order['is_valid_stage']].copy()
df_stage['ketepatan'] = np.where(df_stage['is_late'], 'Telat', 'Tepat waktu')

stage_cols = ['approval_days', 'seller_processing_days', 'carrier_transit_days']
stage_comparison = df_stage.groupby('ketepatan')[stage_cols + ['delivery_days']].median().round(2).T
stage_comparison['selisih_hari'] = (stage_comparison['Telat'] - stage_comparison['Tepat waktu']).round(2)
stage_comparison.index = ['Approval pembayaran', 'Proses seller', 'Transit kurir', 'Total lama pengiriman']
stage_comparison
````

**Uji normalitas.** Sebelum memilih uji perbandingan, distribusi durasi setiap tahap diuji normalitasnya dengan **D'Agostino-Pearson** (`normaltest`), yang menggabungkan uji skewness dan kurtosis. Uji ini dipilih karena jumlah datanya besar (puluhan ribu pesanan), sedangkan p-value Shapiro-Wilk tidak akurat untuk data di atas 5.000 baris.
- **Ho:** data berdistribusi normal.
- **Ha:** data tidak berdistribusi normal.

````python
from scipy.stats import normaltest

for col, label in [('seller_processing_days', 'Proses seller'), ('carrier_transit_days', 'Transit kurir')]:
    for group in ['Tepat waktu', 'Telat']:
        values = df_stage.loc[df_stage['ketepatan'] == group, col]
        stat, pvalue = normaltest(values)
        pvalue_text = '< 1e-300' if pvalue == 0 else f'= {pvalue:.3g}'
        result = 'Tolak Ho (tidak normal)' if pvalue < 0.05 else 'Gagal menolak Ho (normal)'
        print(f'{label:13s} | {group:11s}: skewness {values.skew():.2f}, p-value {pvalue_text}. {result}.')
````

Seluruh kelompok **tidak berdistribusi normal** dan miring ke kanan (skewness positif), sehingga perbandingan durasi menggunakan **median** dan uji non-parametrik **Mann-Whitney**:

````python
for col, label in [('seller_processing_days', 'Proses seller'), ('carrier_transit_days', 'Transit kurir')]:
    late_values = df_stage.loc[df_stage['ketepatan'] == 'Telat', col]
    on_time_values = df_stage.loc[df_stage['ketepatan'] == 'Tepat waktu', col]
    stat, pvalue = mannwhitneyu(late_values, on_time_values, alternative='greater')
    pvalue_text = '< 1e-300' if pvalue == 0 else f'= {pvalue:.3g}'
    result = 'Tolak Ho' if pvalue < 0.05 else 'Gagal menolak Ho'
    print(f'{label:15s}: p-value {pvalue_text}. {result}.')
````

````python
fig, ax = plt.subplots(figsize=(10, 3.5))
stage_plot = stage_comparison.loc[['Approval pembayaran', 'Proses seller', 'Transit kurir'], ['Tepat waktu', 'Telat']].T
left = np.zeros(len(stage_plot))
for stage, color in zip(stage_plot.columns, ['#cccccc', '#ff9896', '#d62728']):
    ax.barh(stage_plot.index, stage_plot[stage], left=left, color=color, label=stage)
    for i, value in enumerate(stage_plot[stage]):
        if value > 1:
            ax.text(left[i] + value / 2, i, f'{value:.1f} hari', ha='center', va='center', color='white' if stage == 'Transit kurir' else 'black')
    left += stage_plot[stage].values
ax.set_title(f"Pesanan telat kehilangan waktu terutama di tahap transit ({stage_comparison.loc['Transit kurir', 'Telat']:.0f} vs "
             f"{stage_comparison.loc['Transit kurir', 'Tepat waktu']:.0f} hari)\n(median durasi per tahap: tepat waktu vs telat)")
ax.set_xlabel('Hari')
ax.legend(loc='lower right')
plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **Approval pembayaran** = dari pembelian sampai pembayaran disetujui
- **Proses seller** = dari pembayaran disetujui sampai barang diserahkan ke kurir
- **Transit kurir** = dari barang diserahkan ke kurir sampai diterima pelanggan
- **Telat** = sampai setelah tanggal estimasi

**Insight:**
- Pesanan telat membutuhkan total ±31 hari (median), dibanding ±10 hari pada pesanan tepat waktu.
- **Perbedaan durasi terbesar terlihat pada tahap transit kurir**: median **±26 hari** pada pesanan telat vs **±7 hari** pada pesanan tepat waktu (**+19 hari**), sedangkan proses seller **±3,1 vs ±1,8 hari** (**+1,3 hari**).
- Kedua uji Mann-Whitney: **Tolak Ho**, kedua tahap memang lebih lama pada pesanan telat, tetapi **perbedaan terbesar terlihat setelah barang diserahkan ke kurir**.

**Interpretasi:** sesuai batasan project, dataset tidak memiliki data kurir, sehingga kesimpulan yang dapat diambil adalah *terdapat hambatan pada tahap transit setelah barang diserahkan ke kurir*, bukan bahwa kurir tertentu berkinerja buruk.

### 5.3.2 Apakah Telat Serah Berkaitan dengan Pesanan Telat Sampai?

````python
df_handover = df_order[df_order['is_valid_handover'] & df_order['is_delivered']].copy()
df_handover['status_handover'] = np.where(df_handover['is_late_handover'], 'Telat serah ke kurir', 'Tepat serah ke kurir')

baseline_handover = df_handover['is_late_handover'].mean() * 100
baseline_late_handover_data = df_handover['is_late'].mean() * 100
print(f'Pesanan delivered 1 seller dengan handover valid: {df_handover.shape[0]}')
print(f'Seller Late Handover Rate (data ini)            : {baseline_handover:.2f}%')
print(f'Late Delivery Rate (data ini)                   : {baseline_late_handover_data:.2f}%')
````

````python
handover_summary = df_handover.groupby('status_handover').agg(
    jumlah_pesanan=('order_id', 'count'),
    late_rate_pct=('is_late', 'mean'),
    low_review_rate_pct=('is_low_review', 'mean'),
    median_lama_kirim_hari=('delivery_days', 'median'),
    median_transit_hari=('carrier_transit_days', 'median'),
)
handover_summary['late_rate_pct'] = (handover_summary['late_rate_pct'] * 100).round(2)
handover_summary['low_review_rate_pct'] = (handover_summary['low_review_rate_pct'] * 100).round(2)
handover_summary = handover_summary.rename(columns={'late_rate_pct': 'late_rate_%', 'low_review_rate_pct': 'low_review_rate_%'})
handover_summary.round(2)
````

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 3.5))
for ax, col, title in [(axes[0], 'late_rate_%', 'Late Delivery Rate'), (axes[1], 'low_review_rate_%', 'Low Review Rate')]:
    ax.barh(handover_summary.index, handover_summary[col], color=['#d62728', '#9e9e9e'])
    for i, value in enumerate(handover_summary[col]):
        ax.text(value + 0.3, i, f'{value:.1f}%', va='center')
    ax.set_title(f'{title} berdasarkan Ketepatan Seller Serah ke Kurir')
    ax.set_xlabel(f'{title} (%)')
    ax.set_xlim(0, handover_summary[col].max() * 1.25)
fig.suptitle(f"Pesanan yang telat diserahkan seller "
             f"{handover_summary.loc['Telat serah ke kurir', 'late_rate_%'] / handover_summary.loc['Tepat serah ke kurir', 'late_rate_%']:.1f}× "
             'lebih sering telat sampai', fontsize=12)
plt.tight_layout()
plt.show()

table = pd.crosstab(df_handover['status_handover'], df_handover['is_late'].astype(bool))
stat, pvalue, dof, expected = chi2_contingency(table)
print(f"Chi-Square handover vs keterlambatan: p-value {'< 1e-300' if pvalue == 0 else f'= {pvalue:.3g}'}. "
      f"{'Tolak Ho' if pvalue < 0.05 else 'Gagal menolak Ho'}.")
````

**Keterangan istilah:**

- **Telat serah ke kurir** = seller menyerahkan barang ke kurir setelah batas kirim (`shipping_limit_date`)
- **Cakupan** = pesanan delivered dari 1 seller yang tanggal serahnya valid
- **Late Delivery Rate** = pesanan yang sampai setelah tanggal estimasi ÷ pesanan delivered
- **Low Review Rate** = jumlah ulasan buruk (bintang 1–2) ÷ jumlah pesanan berulasan valid

````python
df_late_orders = df_handover[df_handover['is_late']]
share_from_seller = df_late_orders['is_late_handover'].mean() * 100
print(f'Pesanan telat sampai                                : {df_late_orders.shape[0]}')
print(f'... yang juga telat serah ke kurir                  : {share_from_seller:.1f}%')
print(f'... yang diserahkan tepat waktu oleh seller          : {100 - share_from_seller:.1f}%')
````

**Insight:**
- Pesanan yang **telat diserahkan** seller ke kurir memiliki Late Delivery Rate **21,3%**, dibanding **5,5%** pada pesanan yang diserahkan tepat waktu (**±3,9×**), dan Low Review Rate **23,7% vs 11,3%**. Uji Chi-Square: **Tolak Ho**.
- Median waktu transit pada kedua kelompok hampir sama (±7 hari). Artinya, **tidak terlihat percepatan** di tahap transit pada pesanan yang telat diserahkan: selisih hari di sisi seller tetap terlihat sampai pesanan diterima pelanggan.
- Di antara pesanan yang telat sampai, **27,8% juga mengalami telat serah** oleh seller, sedangkan **72,2% telah diserahkan seller tepat waktu**. Angka ini adalah **pembagian pesanan telat** menurut status serah ke kurir, bukan pembagian penyebab keterlambatan.

**Implikasi:** keterlambatan terlihat pada dua tahap yang ditangani pihak berbeda:
1. **Telat serah oleh seller** (27,8% pesanan telat juga telat serah): dapat dikendalikan langsung oleh Olist melalui aturan dan pembinaan seller → dilanjutkan ke daftar seller prioritas (5.3.3).
2. **Tahap transit** (72,2% pesanan telat sudah diserahkan tepat waktu): sejalan dengan temuan Pilar 2 (jarak, rute, dan lonjakan bulan tertentu) → ditangani melalui perbaikan rute dan estimasi pengiriman.

### 5.3.3 Seller Mana yang Perlu Diprioritaskan?

**Minimum jumlah pesanan.** Seller dengan sedikit pesanan dapat terlihat ekstrem (misalnya 1 dari 2 pesanan telat = 50%). Sesuai asumsi Business Understanding, seller hanya diberi peringkat jika memiliki jumlah pesanan minimum. Distribusi jumlah pesanan per seller dicek terlebih dahulu:

````python
seller_volume = df_handover.groupby('seller_id').agg(
    jumlah_pesanan=('order_id', 'count'),
    jumlah_telat_serah=('is_late_handover', 'sum'),
)
print(f'Jumlah seller: {seller_volume.shape[0]}')
print(seller_volume['jumlah_pesanan'].describe(percentiles=[0.5, 0.75, 0.9, 0.95]).round(1).to_string())
````

````python
threshold_check = pd.DataFrame([
    {
        'minimal_pesanan': t,
        'jumlah_seller': (seller_volume['jumlah_pesanan'] >= t).sum(),
        'cakupan_pesanan_%': round(seller_volume.loc[seller_volume['jumlah_pesanan'] >= t, 'jumlah_pesanan'].sum() / seller_volume['jumlah_pesanan'].sum() * 100, 1),
        'cakupan_telat_serah_%': round(seller_volume.loc[seller_volume['jumlah_pesanan'] >= t, 'jumlah_telat_serah'].sum() / seller_volume['jumlah_telat_serah'].sum() * 100, 1),
    }
    for t in [10, 20, 30, 50, 100]
])
threshold_check
````

Separuh seller hanya memiliki ±7 pesanan, sehingga persentase mereka tidak stabil. Dipilih batas **minimal 30 pesanan** sebagai **jalan tengah antara rate yang stabil dan cakupan data yang luas**:
- batas yang lebih rendah (10–20) membuat rate seller kecil tidak stabil, sedangkan batas yang lebih tinggi (50–100) membuang terlalu banyak pesanan dan kejadian telat serah (tabel di atas),
- batas ini masih mencakup **±83% pesanan** dan **±81% kejadian telat serah**, sehingga sebagian besar masalah tetap tercakup.

**Segmentasi seller.** Seller dengan minimal 30 pesanan dikelompokkan menjadi 4 segmen berdasarkan dua metrik, dibandingkan dengan baseline:

| Segmen | Telat serah ke kurir | Telat sampai ke pelanggan | Arti |
|---|---|---|---|
| **Prioritas** | di atas baseline | di atas baseline | Telat serah dan pesanannya memang sering telat sampai |
| **Pantau SLA** | di atas baseline | di bawah baseline | Sering melanggar batas serah, tetapi belum berdampak besar ke pelanggan |
| **Masalah rute/transit** | di bawah baseline | di atas baseline | Seller patuh, tetapi pesanannya sering telat karena tahap transit |
| **Baik** | di bawah baseline | di bawah baseline | Tidak perlu tindakan khusus |

````python
MIN_ORDERS_SELLER = 30

seller_summary = df_handover.groupby('seller_id').agg(
    seller_state=('seller_state', 'first'),
    jumlah_pesanan=('order_id', 'count'),
    jumlah_telat_serah=('is_late_handover', 'sum'),
    jumlah_telat_sampai=('is_late', 'sum'),
    low_review_rate_pct=('is_low_review', 'mean'),
    median_proses_seller_hari=('seller_processing_days', 'median'),
)
seller_summary = seller_summary[seller_summary['jumlah_pesanan'] >= MIN_ORDERS_SELLER].copy()
seller_summary['late_handover_rate_%'] = (seller_summary['jumlah_telat_serah'] / seller_summary['jumlah_pesanan'] * 100).round(1)
seller_summary['late_delivery_rate_%'] = (seller_summary['jumlah_telat_sampai'] / seller_summary['jumlah_pesanan'] * 100).round(1)
seller_summary['low_review_rate_%'] = (seller_summary.pop('low_review_rate_pct') * 100).round(1)

handover_high = seller_summary['late_handover_rate_%'] > baseline_handover
delivery_high = seller_summary['late_delivery_rate_%'] > baseline_late_handover_data
seller_summary['segmen'] = np.select(
    [handover_high & delivery_high, handover_high, delivery_high],
    ['Prioritas', 'Pantau SLA', 'Masalah rute/transit'],
    default='Baik',
)

segment_summary = seller_summary.groupby('segmen').agg(
    jumlah_seller=('jumlah_pesanan', 'count'),
    jumlah_pesanan=('jumlah_pesanan', 'sum'),
    jumlah_telat_serah=('jumlah_telat_serah', 'sum'),
    jumlah_telat_sampai=('jumlah_telat_sampai', 'sum'),
)
segment_summary['porsi_pesanan_%'] = (segment_summary['jumlah_pesanan'] / df_handover.shape[0] * 100).round(1)
segment_summary['porsi_telat_serah_%'] = (segment_summary['jumlah_telat_serah'] / df_handover['is_late_handover'].sum() * 100).round(1)
segment_summary['porsi_telat_sampai_%'] = (segment_summary['jumlah_telat_sampai'] / df_handover['is_late'].sum() * 100).round(1)
segment_summary.loc[['Prioritas', 'Pantau SLA', 'Masalah rute/transit', 'Baik']]
````

````python
fig, ax = plt.subplots(figsize=(11, 6.5))
segment_colors = {'Prioritas': '#d62728', 'Pantau SLA': '#ff7f0e', 'Masalah rute/transit': '#1f77b4', 'Baik': '#bbbbbb'}
for segment, color in segment_colors.items():
    data = seller_summary[seller_summary['segmen'] == segment]
    ax.scatter(data['late_handover_rate_%'], data['late_delivery_rate_%'], s=data['jumlah_pesanan'] / 4 + 8,
               color=color, alpha=0.6, edgecolor='white', label=f'{segment} ({len(data)} seller)')
ax.axvline(baseline_handover, color='black', linestyle='--', linewidth=1)
ax.axhline(baseline_late_handover_data, color='black', linestyle=':', linewidth=1)
ax.set_title(f"{(seller_summary['segmen'] == 'Prioritas').sum()} seller Prioritas telat serah sekaligus pesanannya sering telat sampai\n"
             f'(segmentasi seller, min. {MIN_ORDERS_SELLER} pesanan; ukuran titik = jumlah pesanan)')
ax.set_xlabel('Seller Late Handover Rate (%)')
ax.set_ylabel('Late Delivery Rate (%)')
ax.legend()
plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **Seller Late Handover Rate** = pesanan yang diserahkan ke kurir setelah batas kirim (`shipping_limit_date`) ÷ pesanan seller tersebut
- **Prioritas** = Seller Late Handover Rate di atas baseline (9,1%) DAN Late Delivery Rate di atas baseline (6,9%)
- **Pantau SLA** = hanya Seller Late Handover Rate yang di atas baseline (SLA = batas waktu serah ke kurir)
- **Masalah rute/transit** = hanya Late Delivery Rate yang di atas baseline: seller tepat serah, tetapi pesanannya sering telat sampai
- **Baik** = keduanya pada atau di bawah baseline
- **Garis putus-putus / titik-titik** = baseline Seller Late Handover Rate / baseline Late Delivery Rate

**Daftar seller prioritas**, diurutkan berdasarkan **telat serah berlebih** = jumlah telat serah − jumlah pesanan × baseline (9,07%), sama seperti cara menilai rute dan state. Dengan cara ini, seller bervolume besar yang rate-nya hanya sedikit di atas baseline tidak otomatis berada di urutan atas:

````python
# telat serah berlebih = jumlah telat serah − jumlah pesanan × baseline, sama seperti analisis rute dan state
seller_summary['telat_serah_berlebih'] = (seller_summary['jumlah_telat_serah'] - seller_summary['jumlah_pesanan'] * baseline_handover / 100).round(1)
priority_sellers = (
    seller_summary[seller_summary['segmen'] == 'Prioritas']
    .sort_values('telat_serah_berlebih', ascending=False)
    [['seller_state', 'jumlah_pesanan', 'jumlah_telat_serah', 'telat_serah_berlebih', 'late_handover_rate_%',
      'late_delivery_rate_%', 'low_review_rate_%', 'median_proses_seller_hari']]
)
print(f'Jumlah seller prioritas: {len(priority_sellers)}')
print(f"Seller prioritas dengan telat serah berlebih < 5 pesanan: {(priority_sellers['telat_serah_berlebih'] < 5).sum()}")
priority_sellers.head(15)
````

````python
# seberapa terkonsentrasi masalah telat serah pada seller prioritas teratas
cumulative_share = priority_sellers['jumlah_telat_serah'].cumsum() / df_handover['is_late_handover'].sum() * 100
for top_n in [10, 20, 50, len(priority_sellers)]:
    print(f'Top {top_n:3d} seller prioritas menyumbang {cumulative_share.iloc[top_n - 1]:.1f}% dari seluruh pesanan telat serah')

print('\nState asal seller prioritas:')
print(priority_sellers['seller_state'].value_counts().head(5).to_string())
````

**Insight:**
- Dari 604 seller dengan minimal 30 pesanan, **121 seller masuk segmen Prioritas**. Mereka hanya menangani **±17% pesanan**, tetapi menyumbang **±40% dari seluruh pesanan telat serah** dan **±28% dari seluruh pesanan telat sampai**.
- Masalah sangat terkonsentrasi: **10 seller prioritas teratas** saja sudah menyumbang **±13%** dari seluruh pesanan telat serah, dan 50 seller teratas ±29%.
- Sebanyak **42 dari 121 seller prioritas** hanya memiliki telat serah berlebih kurang dari 5 pesanan. Seller ini masuk segmen Prioritas karena rate-nya sedikit di atas baseline, sehingga berada di urutan bawah dan cukup dipantau. Beberapa seller bervolume besar memiliki Late Handover Rate jauh di atas baseline 9,1% (hingga ±30–52%).
- Sebagian besar seller prioritas (**89 dari 121**) berasal dari **SP**, sejalan dengan konsentrasi seller Olist di SP (langkah merge 5).
- Segmen **Masalah rute/transit** menunjukkan seller yang patuh menyerahkan barang tepat waktu, tetapi pesanannya tetap sering telat sampai. Untuk seller ini, pembinaan seller **tidak akan efektif**; perbaikannya ada di sisi rute dan estimasi pengiriman (Pilar 2).

**Catatan:** daftar ini adalah **daftar awal untuk evaluasi dan pembinaan**, bukan daftar hitam otomatis. Data tidak menjelaskan penyebab internal keterlambatan seller (stok, kapasitas, proses packing), sesuai batasan project.

**Hubungan dengan seller berisiko (5.1.8).** Pada breakdown pesanan tepat waktu ditemukan **seller berisiko**, yaitu seller yang ulasan buruknya tinggi walaupun pengirimannya tepat waktu, dan keluhannya didominasi kualitas serta ketidaksesuaian produk (5.1.8). Apakah seller ini sama dengan seller prioritas di atas?

````python
risky_segment = seller_summary['segmen'].reindex(risky_sellers).fillna('Tidak dinilai (< 30 pesanan)')

print('Seller berisiko (5.1.8)       :', len(risky_sellers))
print('Juga termasuk seller prioritas:', (risky_segment == 'Prioritas').sum())
risky_segment.value_counts().rename_axis('segmen di 5.3.3').to_frame('jumlah_seller_berisiko')
````

**Temuan:** hanya **14 dari 30** seller berisiko yang juga termasuk seller prioritas. 16 sisanya berada di segmen lain, termasuk 7 seller yang ketepatan serah dan pengirimannya tergolong **baik**. Artinya, kedua daftar **saling melengkapi**, bukan duplikat:

| Daftar | Masalah utama | Tindakan |
|---|---|---|
| **121 seller prioritas** (5.3.3) | Telat menyerahkan barang ke kurir | Pembinaan ketepatan serah |
| **30 seller berisiko** (5.1.8) | Ulasan buruk tinggi walaupun tepat waktu; keluhan kualitas dan ketidaksesuaian produk (5.1.8) | Evaluasi deskripsi, foto, dan kualitas produk (Rekomendasi 3) |

**14 seller** yang masuk di kedua daftar memiliki kedua masalah sekaligus, sehingga paling layak didahulukan.

**Rute rawan yang sebenarnya masalah seller.** Sebuah rute bisa terlihat rawan padahal pesanannya hanya berasal dari satu seller. Dalam kasus seperti ini, tindakan yang tepat adalah pembinaan seller, bukan perbaikan jalur logistik. Dicek rute dengan minimal 100 pesanan dan Late Delivery Rate di atas baseline yang **≥ 80% pesanannya berasal dari 1 seller**:

````python
route_seller = df_route.groupby('route_state').agg(
    jumlah_pesanan=('order_id', 'count'),
    late_rate=('is_late', 'mean'),
    seller_terbesar=('seller_id', lambda s: s.value_counts().index[0]),
    porsi_seller_terbesar=('seller_id', lambda s: s.value_counts(normalize=True).iloc[0]),
)
route_seller['late_rate_%'] = (route_seller.pop('late_rate') * 100).round(1)
route_seller['porsi_seller_terbesar_%'] = (route_seller.pop('porsi_seller_terbesar') * 100).round(1)
route_seller['segmen_seller'] = route_seller['seller_terbesar'].map(seller_summary['segmen'])

route_seller[
    (route_seller['jumlah_pesanan'] >= 100)
    & (route_seller['late_rate_%'] > baseline_late)
    & (route_seller['porsi_seller_terbesar_%'] >= 80)
]
````

**Temuan:** rute **MA → SP** (123 pesanan, Late Delivery Rate 20,3%, ±3× baseline) **seluruhnya** berasal dari 1 seller (`06a2c3af`), yang juga termasuk **segmen Prioritas**.

**Implikasi:** rute MA → SP sebenarnya **masalah satu seller**, bukan masalah jalur logistik, sehingga tindakannya adalah pembinaan seller tersebut. Saat memakai daftar rute rawan, perlu dicek dulu apakah rute itu hanya dilayani satu seller, agar tindakan yang dipilih tepat.

Dari sisi pelanggan, ketika keluhan menyebut pihak yang disalahkan, **seller disebut ±68%** dan kurir hanya ±9% (5.1.8), baik untuk keterlambatan maupun barang kurang/salah. Karena itu, pembinaan seller perlu mencakup dua hal: **ketepatan serah barang** (121 seller prioritas) dan **kualitas serta kesesuaian produk** (seller berisiko dari 5.1.8, yang sebagian besar berbeda dari seller prioritas).

|---|---|---|---|---|
| **1 — WHAT** | Faktor apa yang paling berkaitan dengan ulasan buruk? | Keterlambatan (dampak terbesar), pesanan tidak sampai, dan pesanan kompleks (multi-barang/multi-seller) yang sering tidak lengkap. Ongkir dan nilai pesanan hanya berkaitan tidak langsung; metode bayar tidak terbukti berkaitan (p ≈ 0,057) | Telat 62,4% vs tepat waktu 9,2%; ±3.400 ulasan buruk berlebih | Menekan keterlambatan adalah tuas utama menurunkan Low Review Rate |
| **1 — WHAT** (isi ulasan) | Apa yang dikeluhkan pelanggan? | Telat/belum sampai, barang kurang/salah, dan kualitas produk. Pada pesanan tepat waktu didominasi barang kurang/salah dan kualitas | 35% / 30% / 16%; seller disalahkan ±68% vs kurir ±9% | Selain logistik, kelengkapan pengiriman dan kualitas produk seller perlu dibenahi |
| **2 — WHERE** | Di rute dan wilayah mana keterlambatan terkonsentrasi? | Beda-state dan jarak > 1.500 km; tujuan Nordeste dan RJ; 11 rute prioritas, terutama SP → RJ | 11 rute = ±23% pesanan, ±40% pesanan telat | Perbaikan bisa difokuskan pada sedikit rute |
| **2 — WHEN** | Kapan keterlambatan meningkat? | Nov 2017 (lonjakan volume) dan Feb–Mar 2018 (transit melambat, estimasi tidak disesuaikan) | 12,4% / 14,1% / 19,0% | Siapkan kapasitas dan sesuaikan estimasi pada periode puncak |
| **3 — WHY** | Hambatan terbesar di seller atau transit? | Perbedaan durasi terbesar ada di transit; 27,8% pesanan telat juga telat serah oleh seller | Transit ±26 vs ±7 hari; 27,8% pesanan telat juga telat serah | Perbaikan logistik dan pembinaan seller berjalan bersamaan |
| **3 — WHO** | Seller mana yang perlu diprioritaskan? | 121 seller segmen Prioritas | ±17% pesanan → ±40% telat serah | Pembinaan terarah, dimulai dari 10 seller teratas |

**Yang tidak bisa diklaim dari analisis ini:**
- Semua hubungan adalah **keterkaitan**, bukan sebab-akibat (data historis, bukan eksperimen; bagian 1.9).
- Hambatan transit tidak bisa dikaitkan ke **perusahaan kurir tertentu**, karena identitas kurir tidak tersedia.
- Label isi ulasan dibuat oleh Claude dan sudah dicek manual pada 150 sampel: Pain Points ±94%, Review Stage ±95%, tetapi Blame Target hanya ±79%, sebagian karena perbedaan tafsir "pihak yang disebut" vs "pihak yang salah" (5.1.8a). Label hanya mencakup ±75% ulasan buruk yang berkomentar.
- Data akhir Agustus 2018 tidak lengkap (3.9).

## 5.4 Prioritas dan Estimasi Dampak (Skenario)

Sesuai Goal 5 pada Business Understanding, bagian ini memperkirakan **berapa banyak keterlambatan dan ulasan buruk yang berpotensi berkurang** jika area prioritas dari Pilar 2 dan Pilar 3 diperbaiki.

**Penting:** angka di bawah adalah **skenario berdasarkan data historis**, bukan prediksi pasti. Dampak sebenarnya baru dapat diukur setelah perbaikan diterapkan.

### Cara Menghitung

Estimasi dilakukan dalam dua langkah sederhana:

**Langkah 1: berapa pesanan telat yang berkurang?**
Setiap skenario mengasumsikan kelompok prioritas memiliki Late Delivery Rate yang sama dengan kelompok pembandingnya:

| Skenario | Kelompok prioritas | Diasumsikan menjadi |
|---|---|---|
| **A. Perbaikan rute** | 11 rute prioritas (5.2.4) | Late Delivery Rate = **baseline nasional** (6,79%) |
| **B. Pembinaan seller** | Pesanan telat serah dari 121 seller prioritas (5.3.3) | Seller Late Handover Rate seller prioritas turun ke **baseline (9,1%)**. Pesanan telat serah yang menjadi tepat serah memiliki peluang telat seperti pesanan tepat serah (5,5% vs 21,3%) |
| **C. Kesiapan bulan sibuk** | Pesanan pada bulan lonjakan (Nov 2017, Feb & Mar 2018) | Late Delivery Rate = **rata-rata bulan normal** |

**Langkah 2: berapa ulasan buruk yang berkurang?**
Dari Pilar 1, Low Review Rate pesanan telat **62,4%** dan pesanan tepat waktu **9,2%**. Dalam skenario, setiap pesanan yang berubah dari telat menjadi tepat waktu **diasumsikan** memiliki peluang ulasan buruk ±53 poin persen lebih rendah:

> Ulasan buruk yang berkurang = pesanan telat yang berkurang (yang memiliki ulasan) × (62,4% − 9,2%)

````python
# Langkah 2: selisih Low Review Rate antara pesanan telat dan tepat waktu (dari Pilar 1)
low_review_if_late = df_review.loc[df_review['is_late'].fillna(False).astype(bool), 'is_low_review'].mean()
low_review_if_on_time = df_review.loc[is_on_time, 'is_low_review'].mean()
low_review_gap = low_review_if_late - low_review_if_on_time
print(f'Low Review Rate pesanan telat      : {low_review_if_late * 100:.1f}%')
print(f'Low Review Rate pesanan tepat waktu: {low_review_if_on_time * 100:.1f}%')
print(f'Selisih                            : {low_review_gap * 100:.1f} poin persen')
````

Setiap pesanan delivered diberi nilai **peluang telat** dalam skenario. Awalnya nilainya sama dengan kondisi sebenarnya (1 = telat, 0 = tepat waktu). Untuk pesanan dalam kelompok prioritas, peluang telat dikecilkan sesuai skenario. Contoh: rute dengan Late Delivery Rate 14,4% yang diasumsikan turun ke 6,79% berarti peluang telat setiap pesanan telat di rute itu dikali 6,79 ÷ 14,4.

Dengan cara ini, skenario A dan B dapat **digabung tanpa menghitung dua kali** pesanan yang termasuk rute prioritas sekaligus seller prioritas.

````python
df_scenario = df_order[df_order['is_delivered']].copy()
df_scenario['telat_aktual'] = df_scenario['is_late'].astype(float)

# Skenario A: rute prioritas turun ke baseline nasional
route_late_rate = df_route.groupby('route_state')['is_late'].mean()
in_priority_route = df_scenario['route_state'].isin(priority_routes.index)
route_factor = (baseline_late / 100) / df_scenario.loc[in_priority_route, 'route_state'].map(route_late_rate)

# Skenario B: Seller Late Handover Rate seller prioritas turun ke baseline
late_rate_late_handover = handover_summary.loc['Telat serah ke kurir', 'late_rate_%'] / 100
late_rate_on_time_handover = handover_summary.loc['Tepat serah ke kurir', 'late_rate_%'] / 100
handover_target = df_order.loc[df_order['is_valid_handover'], 'is_late_handover'].astype(bool).mean()
priority_handover_rate = df_handover.loc[df_handover['seller_id'].isin(priority_sellers.index), 'is_late_handover'].astype(bool).mean()
share_fixed = 1 - handover_target / priority_handover_rate  # porsi pesanan telat serah yang menjadi tepat serah
in_priority_seller = df_scenario['seller_id'].isin(priority_sellers.index) & (df_scenario['is_late_handover'] == True)
# sebagian pesanan telat serah menjadi tepat serah (peluang telatnya mengikuti pesanan tepat serah), sisanya tetap
seller_factor = (1 - share_fixed) + share_fixed * late_rate_on_time_handover / late_rate_late_handover

# Skenario C: bulan lonjakan turun ke rata-rata bulan normal
in_spike_month = df_scenario['purchase_month'].isin(spike_months)
normal_month_rate = df_scenario.loc[~in_spike_month, 'is_late'].mean()
spike_factor = normal_month_rate / df_scenario.loc[in_spike_month, 'is_late'].mean()

df_scenario['A_rute'] = df_scenario['telat_aktual']
df_scenario.loc[in_priority_route, 'A_rute'] *= route_factor

df_scenario['B_seller'] = df_scenario['telat_aktual']
df_scenario.loc[in_priority_seller, 'B_seller'] *= seller_factor

df_scenario['A_dan_B'] = df_scenario['A_rute']
df_scenario.loc[in_priority_seller, 'A_dan_B'] *= seller_factor

df_scenario['C_bulan_sibuk'] = df_scenario['telat_aktual']
df_scenario.loc[in_spike_month, 'C_bulan_sibuk'] *= spike_factor

print('Pesanan pada rute prioritas            :', in_priority_route.sum())
print('Pesanan telat serah dari seller prioritas:', in_priority_seller.sum())
print('Pesanan yang termasuk keduanya          :', (in_priority_route & in_priority_seller).sum())
print('Pesanan pada bulan lonjakan             :', in_spike_month.sum())
print(f'Seller Late Handover Rate seller prioritas: {priority_handover_rate * 100:.1f}% -> {handover_target * 100:.1f}% '
      f'({share_fixed * 100:.0f}% pesanan telat serah menjadi tepat serah)')
print(f"Late Delivery Rate kelompok B: {df_scenario.loc[in_priority_seller, 'telat_aktual'].mean() * 100:.1f}% "
      f"-> {df_scenario.loc[in_priority_seller, 'B_seller'].mean() * 100:.1f}% setelah skenario")
````

````python
total_reviewed = df_review.shape[0]
total_low_review = df_review['is_low_review'].sum()
months_in_data = df_scenario['purchase_month'].nunique()
reviewed_mask = df_scenario['is_valid_review']

scenario_labels = {
    'A_rute': 'A. Perbaikan 11 rute prioritas',
    'B_seller': 'B. Pembinaan 121 seller prioritas',
    'A_dan_B': 'A + B (tanpa hitung ganda)',
    'C_bulan_sibuk': 'C. Kesiapan bulan sibuk',
}

scenario_rows = []
for col, label in scenario_labels.items():
    late_avoided = df_scenario['telat_aktual'].sum() - df_scenario[col].sum()
    late_avoided_reviewed = df_scenario.loc[reviewed_mask, 'telat_aktual'].sum() - df_scenario.loc[reviewed_mask, col].sum()
    low_review_avoided = late_avoided_reviewed * low_review_gap
    scenario_rows.append({
        'skenario': label,
        'pesanan_telat_berkurang': round(late_avoided),
        'late_delivery_rate_baru_%': round(df_scenario[col].mean() * 100, 2),
        'ulasan_buruk_berkurang': round(low_review_avoided),
        'low_review_rate_baru_%': round((total_low_review - low_review_avoided) / total_reviewed * 100, 2),
        'ulasan_buruk_berkurang_per_tahun': round(low_review_avoided / months_in_data * 12),
    })

scenario_result = pd.DataFrame(scenario_rows).set_index('skenario')
print(f'Kondisi saat ini: Late Delivery Rate {baseline_late:.2f}%, Low Review Rate {baseline_low_review:.2f}%')
scenario_result
````

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
plot_data = scenario_result.iloc[::-1]

axes[0].barh(plot_data.index, plot_data['pesanan_telat_berkurang'], color=['#9e9e9e', '#d62728', '#ff9896', '#ff9896'])
axes[0].set_title('Estimasi Pesanan Telat yang Berkurang')
axes[0].set_xlabel('Jumlah pesanan (Jan 2017 – Agu 2018)')
for i, v in enumerate(plot_data['pesanan_telat_berkurang']):
    axes[0].text(v + 20, i, f'{v:,}'.replace(',', '.'), va='center')

axes[1].barh(plot_data.index, plot_data['ulasan_buruk_berkurang'], color=['#9e9e9e', '#d62728', '#ff9896', '#ff9896'])
axes[1].set_title('Estimasi Ulasan Buruk (bintang 1-2) yang Berkurang')
ab = scenario_result.loc['A + B (tanpa hitung ganda)']
fig.suptitle(f"Perbaikan rute dan seller prioritas (A + B) berpotensi mencegah ±{int(ab['pesanan_telat_berkurang']):,} pesanan telat "
             f"dan ±{int(ab['ulasan_buruk_berkurang']):,} ulasan buruk".replace(',', '.'), fontsize=12)
axes[1].set_xlabel('Jumlah pesanan (Jan 2017 – Agu 2018)')
axes[1].set_yticklabels([])
for i, v in enumerate(plot_data['ulasan_buruk_berkurang']):
    axes[1].text(v + 10, i, f'{v:,}'.replace(',', '.'), va='center')

plt.tight_layout()
plt.show()
````

**Keterangan istilah:**

- **A. Perbaikan rute** = Late Delivery Rate 11 rute prioritas diasumsikan turun ke baseline (6,79%)
- **B. Pembinaan seller** = Seller Late Handover Rate 121 seller prioritas diasumsikan turun ke baseline (9,1%)
- **A + B** = gabungan A dan B; pesanan yang termasuk keduanya hanya dihitung sekali
- **C. Kesiapan bulan sibuk** = Late Delivery Rate bulan lonjakan (Nov 2017, Feb 2018, Mar 2018) diasumsikan turun ke rata-rata bulan normal
- **Ulasan buruk berkurang** = pesanan telat yang berkurang (yang berulasan) × selisih Low Review Rate telat vs tepat waktu (53,2 pp)

**Insight:**
- **Skenario A (perbaikan 11 rute prioritas)** berpotensi mencegah **±1.130 pesanan telat** dan **±590 ulasan buruk**, menurunkan Late Delivery Rate dari 6,79% menjadi **±5,6%**.
- **Skenario B (pembinaan 121 seller prioritas)**: jika Seller Late Handover Rate seller prioritas turun dari ±21% ke baseline 9,1%, ±57% pesanan telat serahnya menjadi tepat serah. Hasilnya berpotensi mencegah **±370 pesanan telat** dan **±190 ulasan buruk**.
- **Gabungan A + B** (837 pesanan yang termasuk keduanya tidak dihitung dua kali) berpotensi menurunkan Late Delivery Rate menjadi **±5,3%**, atau ±22% lebih sedikit pesanan telat. Ulasan buruk yang berpotensi dicegah **±750**, atau sekitar **±450 per tahun**. Low Review Rate turun dari 14,61% menjadi **±13,84%**.
- **Skenario C (kesiapan bulan sibuk)** memiliki potensi terbesar (**±2.200 pesanan telat** dan **±1.150 ulasan buruk**), karena lonjakan keterlambatan pada 3 bulan tersebut sangat besar. Namun skenario ini **tidak dapat dijumlahkan** dengan A dan B (kelompok pesanannya tumpang tindih), dan penyebab lonjakannya tidak tersedia di data.

**Cara membaca dampak:** penurunan Low Review Rate secara keseluruhan terlihat kecil (±0,8 poin persen pada A + B, atau ±5% ulasan buruk), karena keterlambatan hanya ±6,8% dari pesanan. Namun dampaknya **terkonsentrasi**: perbaikan hanya perlu menyasar ±23% pesanan (rute prioritas) dan 121 seller, bukan seluruh operasional. Selain itu, Pilar 1 menunjukkan bahwa sebagian ulasan buruk berasal dari faktor di luar keterlambatan (pesanan tidak sampai, kompleksitas pesanan, serta kualitas dan ketidaksesuaian produk, 5.1.4 dan 5.1.8), yang memerlukan tindakan terpisah.

**Asumsi dan batasan skenario:**
- Diasumsikan pesanan yang tidak lagi telat memiliki peluang ulasan buruk yang sama dengan pesanan tepat waktu pada umumnya (9,2%).
- Hubungan yang dipakai adalah **keterkaitan**, bukan sebab-akibat. Angka menunjukkan **potensi**, bukan jaminan.
- Estimasi berbasis data historis Jan 2017 – Agu 2018; angka per tahun hanya proyeksi sederhana dari periode tersebut.

### Batas Atas dan Hubungan Antar-Metrik

Dua angka tambahan membantu membaca skenario di atas:
1. **Batas atas:** jika **seluruh** pesanan telat menjadi tepat waktu, berapa ulasan buruk yang paling banyak bisa dikurangi lewat perbaikan pengiriman?
2. **Hubungan antar-metrik:** Seller Late Handover Rate → Late Delivery Rate → Low Review Rate. Seberapa besar selisih Late Delivery Rate dan Low Review Rate antara pesanan yang telat dan yang tepat diserahkan ke kurir?

````python
# 1) batas atas: jika SELURUH pesanan telat menjadi tepat waktu
late_reviewed = df_scenario.loc[reviewed_mask, 'telat_aktual'].sum()
max_low_review_avoided = late_reviewed * low_review_gap
print(f'Pesanan telat yang memiliki ulasan     : {late_reviewed:,.0f}')
print(f'Ulasan buruk yang berkurang (maksimal) : {max_low_review_avoided:,.0f} '
      f'({max_low_review_avoided / total_low_review * 100:.1f}% dari seluruh ulasan buruk)')
print(f'Low Review Rate jika tidak ada telat   : {(total_low_review - max_low_review_avoided) / total_reviewed * 100:.2f}%')

# 2) hubungan antar-metrik (observasional): telat serah vs tepat serah
low_review_late_handover = handover_summary.loc['Telat serah ke kurir', 'low_review_rate_%']
low_review_on_time_handover = handover_summary.loc['Tepat serah ke kurir', 'low_review_rate_%']
print()
print(f'Late Delivery Rate telat serah vs tepat serah: {late_rate_late_handover * 100:.2f}% vs {late_rate_on_time_handover * 100:.2f}% '
      f'(selisih {(late_rate_late_handover - late_rate_on_time_handover) * 100:.1f} pp)')
print(f'Low Review Rate telat serah vs tepat serah   : {low_review_late_handover:.2f}% vs {low_review_on_time_handover:.2f}% '
      f'(selisih {low_review_late_handover - low_review_on_time_handover:.1f} pp)')
print(f'Low Review Rate telat vs tepat waktu         : selisih {low_review_gap * 100:.1f} pp')
````

**Insight:**
- **Batas atas:** walaupun **seluruh** keterlambatan dihilangkan, ulasan buruk hanya berkurang **±3.391 (±23,6%)**, dan Low Review Rate turun dari 14,61% ke **±11,2%**. Artinya, sekitar 3 dari 4 ulasan buruk berasal dari hal lain di luar keterlambatan, misalnya pesanan tidak sampai, pesanan kompleks yang tidak lengkap (5.1.4), atau kualitas dan ketidaksesuaian produk (5.1.8). Target perbaikan perlu realistis.
- **Hubungan antar-metrik:** pesanan dengan telat serah memiliki Late Delivery Rate **±15,7 pp lebih tinggi** (21,25% vs 5,51%) dan Low Review Rate **±12,4 pp lebih tinggi** (23,68% vs 11,27%) dibanding pesanan yang diserahkan tepat waktu. Sebagai pembanding, selisih Low Review Rate antara pesanan telat dan tepat waktu adalah ±53,2 pp. Hubungan ini hanya **keterkaitan pada data historis, bukan sebab-akibat**.
- Selisih Low Review Rate pada telat serah lebih kecil, karena tidak semua pesanan yang telat diserahkan berujung telat sampai. Karena itu, dalam skenario, per pesanan yang disasar, perbaikan di tahap pengiriman (rute) memberi estimasi penurunan Low Review Rate yang lebih besar daripada perbaikan di sisi seller. Pembinaan seller tetap penting karena bagian ini dapat dikendalikan langsung oleh Olist.

````python
orders_per_customer = df_order.groupby('customer_unique_id')['order_id'].nunique()
print(f'Pelanggan yang berbelanja lebih dari sekali: {(orders_per_customer > 1).mean() * 100:.1f}%')
````

# 6. Kesimpulan dan Rekomendasi

## 6.1 Kesimpulan

**Pertanyaan utama:** *Seller dan rute pengiriman mana yang perlu diprioritaskan perbaikannya agar proporsi ulasan buruk di Olist dapat ditekan?*

Sekitar **1 dari 7 pesanan Olist (14,61%)** berakhir dengan ulasan buruk. Analisis menelusuri faktor operasional yang berkaitan dengannya secara bertahap:

**Pilar 1 — WHAT.** **Keterlambatan pengiriman** adalah faktor operasional yang paling berkaitan dengan ulasan buruk. Pesanan telat memiliki Low Review Rate **62,4%**, ±6,8× lipat pesanan tepat waktu (9,2%), dan polanya bertingkat: semakin lama telat, semakin tinggi ulasan buruknya (32% saat telat 1–3 hari, ±80% saat telat lebih dari 1 minggu). Meskipun hanya ±6,5% pesanan, pesanan telat menyumbang **±28% dari seluruh ulasan buruk**. Faktor lain yang berkaitan adalah pesanan yang tidak sampai (78,3%) dan kompleksitas pesanan (multi-seller 47,2%, pesanan 4+ barang 35,3%), yang terbukti **tidak dijelaskan oleh keterlambatan**. Ongkos kirim dan nilai pesanan hanya berkaitan secara tidak langsung (melalui jumlah barang). Metode pembayaran **tidak terbukti berkaitan** dengan ulasan buruk (p = 0,057 ≥ 0,05, gagal menolak Ho). Isi ulasan buruk menunjukkan keluhan utamanya adalah keterlambatan (±35%), barang kurang/salah (±30%), dan kualitas produk (±16%); pada pesanan tepat waktu, keluhannya didominasi barang kurang/salah dan kualitas produk.

**Pilar 2 — WHERE & WHEN.** Keterlambatan terkonsentrasi pada:
- pengiriman **beda-state** (8,2% vs 4,6%) dan **jarak > 1.500 km** (11,7%), terutama ke region **Nordeste** (12,8%);
- **11 rute prioritas** yang hanya mencakup ±23% pesanan tetapi menyumbang **±40% pesanan telat**, dengan **SP → RJ** sebagai rute berdampak terbesar (14,4% telat, ±1.150 pesanan telat), padahal jaraknya hanya ±383 km;
- **lonjakan menyeluruh** pada November 2017 dan Februari–Maret 2018 (hingga 19,0%) di semua region. Lonjakan November bertepatan dengan kenaikan volume pesanan (Black Friday), sedangkan pada Februari–Maret perubahan terbesar yang teramati terjadi pada transit yang melambat tanpa penyesuaian estimasi.

**Pilar 3 — WHY & WHO.** Perbedaan durasi terbesar antara pesanan telat dan tepat waktu terlihat di **tahap transit** (±26 vs ±7 hari). Di antara pesanan telat, **27,8% juga telat diserahkan** oleh seller. Pesanan yang telat diserahkan ±3,9× lebih sering telat sampai (21,3% vs 5,5%), dan pesanannya tidak menjadi lebih cepat di tahap transit. Masalah ini terkonsentrasi pada **121 seller prioritas** yang menangani ±17% pesanan tetapi menyumbang **±40% pesanan telat serah**.

**Jawaban pertanyaan utama:**
- **Rute yang perlu diprioritaskan:** 11 rute prioritas, terutama **SP → RJ**, **SP → BA**, **SP → ES**, **SP → SC**, dan **SP → CE**, serta pengiriman jarak jauh ke **Nordeste**.
- **Seller yang perlu diprioritaskan**, dengan dua jenis masalah yang berbeda:
  - **121 seller prioritas** yang sering telat menyerahkan barang, dimulai dari 10 seller dengan telat serah berlebih terbesar yang menyumbang ±13% pesanan telat serah;
  - **30 seller berisiko** yang ulasan buruknya tinggi walaupun pengirimannya tepat waktu, dengan keluhan kualitas dan ketidaksesuaian produk (5.1.8).
  - **14 seller** masuk di kedua daftar, sehingga paling layak didahulukan.
- Berdasarkan skenario, perbaikan pada rute prioritas dan ketepatan serah seller berpotensi mengurangi **±22% pesanan telat** dan mencegah **±750 ulasan buruk** (±450 per tahun) pada periode data. Dalam metrik utama, **Low Review Rate turun dari 14,61% menjadi ±13,84%** (−0,8 poin persen, atau ±5% ulasan buruk lebih sedikit). Sebagai batas atas, jika **seluruh** keterlambatan hilang pun, Low Review Rate hanya turun ke **±11,16%** (±24% ulasan buruk berkurang), sehingga sumber ulasan buruk di luar keterlambatan tetap perlu ditangani.

## 6.2 Rekomendasi

Rekomendasi disusun berdasarkan prioritas dan pemiliknya, sesuai stakeholder pada Business Understanding.

### Rekomendasi 1 — Tindak lanjut 11 rute prioritas: gudang dekat tujuan, evaluasi mitra logistik, dan estimasi realistis *(Tim Operasional & Logistik)*

**Dasar:**
- 11 rute prioritas menyumbang ±40% pesanan telat (5.2.4).
- **5 rute telat melebihi wajar untuk jaraknya**, terutama SP → RJ: 14,4% vs rate wajar ±5,3% untuk jarak ±383 km (5.2.6).
- **3 rute telat karena jaraknya > 2.000 km** (SP → CE, SP → PE, SP → PA).
- Pengiriman jarak pendek di wilayah tujuan jarang telat: di dalam RJ hanya **4,8%**, dan ke Nordeste dari ≤ 700 km hanya **7,0%** (5.2.6).

**11 rute prioritas dan tindakannya:**

| Rute | Late Delivery Rate | Pesanan telat berlebih | Jenis masalah | Tindakan utama |
|---|---|---|---|---|
| SP → RJ | 14,4% | ±606 | Melebihi wajar untuk jaraknya | Gudang Rio de Janeiro + evaluasi mitra logistik |
| SP → BA | 13,1% | ±143 | Melebihi wajar untuk jaraknya | Gudang Nordeste + evaluasi mitra logistik |
| SP → ES | 12,5% | ±82 | Melebihi wajar untuk jaraknya | Evaluasi mitra logistik |
| SP → SC | 9,8% | ±68 | Melebihi wajar untuk jaraknya | Evaluasi mitra logistik |
| SP → CE | 13,9% | ±68 | Wajar, jarak sangat jauh | Gudang Nordeste + estimasi realistis |
| PR → RJ | 12,1% | ±52 | Melebihi wajar untuk jaraknya | Gudang Rio de Janeiro + evaluasi mitra logistik |
| SP → PE | 9,9% | ±34 | Wajar, jarak sangat jauh | Gudang Nordeste + estimasi realistis |
| SP → PA | 11,8% | ±34 | Wajar, jarak sangat jauh | Estimasi realistis |

| RJ → SP | 8,0% | ±16 | Wajar untuk jaraknya | Pantau mingguan |
| SP → GO | 7,6% | ±12 | Wajar untuk jaraknya | Pantau mingguan |

**Tindakan:**
1. **Buka gudang di kota Rio de Janeiro (prioritas pertama).**
   - Isinya produk dari ±90 seller terbesar pada rute SP/PR/MG → RJ, yang mencakup ±50% pesanan ketiga rute.
   - Lokasi dipilih karena rata-rata jaraknya ke pelanggan RJ paling kecil (±44 km).
   - Potensi: ±437 pesanan telat dan ±231 ulasan buruk lebih sedikit.
2. **Buka gudang Nordeste di Aracaju** (alternatif Maceió atau Recife) setelah gudang RJ dievaluasi.
   - Isinya produk dari ±54 seller terbesar pada rute SP → BA/CE/PE.
   - 72,9% pelanggan Nordeste berada dalam radius 700 km dari Aracaju.
   - Potensi: ±118 pesanan telat dan ±61 ulasan buruk lebih sedikit.
3. **Evaluasi mitra logistik** pada 5 rute yang telat melebihi wajar untuk jaraknya (SP → RJ, PR → RJ, SP → BA, SP → ES, SP → SC). Bandingkan kecepatan transit antar mitra, dan alihkan pengiriman ke mitra alternatif jika tidak membaik.
4. **Sesuaikan estimasi tanggal sampai untuk rute > 1.500 km** (SP → CE, SP → PE, SP → PA), setidaknya sampai gudang Nordeste beroperasi, agar janji ke pelanggan lebih realistis.
5. **Pantau Late Delivery Rate per rute setiap minggu** melalui dashboard, termasuk rute yang cukup dipantau (MG → RJ, RJ → SP, SP → GO).

**Target:** Late Delivery Rate setiap rute prioritas turun ke baseline nasional (±6,8%).
**Potensi dampak (skenario A):** ±1.130 pesanan telat dan ±590 ulasan buruk lebih sedikit. Kedua gudang sendiri menyumbang ±555 pesanan telat dari target tersebut.

### Rekomendasi 2 — Persiapan kapasitas pada bulan sibuk *(Tim Operasional & Logistik)*

**Dasar:**
- Late Delivery Rate melonjak 2–3× lipat pada November 2017 dan Februari–Maret 2018, di semua region, terutama pada tahap transit (5.2.5).
- **November 2017** bertepatan dengan volume Black Friday. Pada minggu puncak, volume mencapai **2,79×** normal, jauh di atas batas yang selama ini masih tertangani (**1,59×**).
- **Februari–Maret 2018** terjadi saat volume stabil, tetapi transit melambat dan estimasi tidak menyesuaikan.

**Tindakan:**
1. **Pesan kapasitas kurir tambahan sebelum Black Friday:** **+75%** dari kapasitas normal pada minggu setelah Black Friday dan **+22%** pada minggu berikutnya.
   - Rumus: armada tambahan = (volume puncak − 1,59 × volume normal) ÷ 5 hari kerja ÷ kapasitas 1 armada per hari.
   - Pada volume dataset, kekurangannya ±244 paket per hari kerja, setara **±3 armada** jika 1 armada mengantar 100 paket/hari.
2. **Siapkan seller bervolume besar 2–4 minggu sebelumnya.** Sasarannya:
   - **±60 seller** yang menyumbang 50% pesanan Black Friday;
   - seller yang diperkirakan lonjakannya **> 5×**, yang telat serahnya 21,0% (normal 7,9%).

   Mereka diminta menyiapkan stok dan tenaga packing, dan penyerahan ke kurirnya dipantau setiap hari.
3. **Tambahkan buffer estimasi +3 hari** pada periode lonjakan. Buffer ini sudah menurunkan Late Delivery Rate sekitar 3–5 poin persen.
   - Buffer penuh (+7 sampai +10 hari) hanya dipakai jika kapasitas tidak dapat ditambah, karena berisiko menurunkan minat beli.
   - Estimasi **tidak boleh diperpendek** saat transit sedang melambat.
4. **Pantau tanda bahaya mingguan: median transit > 8 hari.** Jika muncul:
   - aktifkan buffer untuk pesanan baru,
   - kirim notifikasi proaktif ke pelanggan,
   - alihkan sebagian pengiriman ke kurir cadangan.

   Pada 2018, tanda ini muncul ±4 minggu sebelum puncak keterlambatan. Kejadian terjadwal seperti Karnaval disiapkan sebelum tanggalnya.
5. **Atur kapasitas harian mengikuti pola mingguan, bukan tanggal gajian** (5.2.5):
   - Senin–Selasa ±13% di atas rata-rata harian, dan Selasa menampung 22,3% serah ke kurir mingguan. Pada minggu puncak Black Friday, kekurangan di hari Selasa ±272 paket.
   - Pada periode puncak, sediakan opsi serah ke kurir hari Sabtu, karena pesanan Jumat diproses seller ±3,2 hari (vs ±1,3–1,4 hari untuk Senin–Rabu).
   - Tidak ada lonjakan pada masa gajian, sehingga kapasitas tidak perlu ditambah khusus di awal bulan.

**Target:** Late Delivery Rate pada bulan sibuk turun ke rata-rata bulan normal (±4,5%).
**Potensi dampak (skenario C):** hingga ±2.200 pesanan telat dan ±1.150 ulasan buruk lebih sedikit (sebagian tumpang tindih dengan skenario A dan B).

- Label isi ulasan menunjukkan bahwa setiap jenis keluhan memiliki asal dan pemilik yang berbeda (5.1.8g).

## 6.3 Trade-off

| Rekomendasi | Manfaat | Trade-off / risiko |
|---|---|---|
| Perbaikan estimasi tanggal sampai (R1, R2) | Janji lebih realistis, keterlambatan berkurang | Estimasi yang lebih lama dapat **menurunkan minat beli** saat checkout; perlu diterapkan selektif pada rute/periode berisiko, bukan menyeluruh |
| Ganti/tambah mitra logistik (R1) | Transit lebih cepat pada rute bermasalah | **Biaya pengiriman dapat naik**; perlu dibandingkan dengan nilai pesanan yang terdampak |
| Gudang Rio de Janeiro dan Nordeste (R1) | Pengiriman ke pelanggan menjadi jarak pendek di wilayah tujuan | Membutuhkan **biaya sewa gudang dan stok**, serta risiko stok menumpuk; karena itu dimulai dari gudang RJ dengan produk ±90 seller terbesar, lalu dievaluasi sebelum membuka gudang Nordeste |
| Kapasitas bulan sibuk (R2) | Mencegah lonjakan keterlambatan | Membutuhkan **biaya kapasitas tambahan** yang mungkin tidak terpakai jika permintaan lebih rendah dari perkiraan; karena itu kapasitas dipesan hanya untuk **±2 minggu setelah Black Friday** (+75% lalu +22%), bukan sebulan penuh, dan buffer estimasi dibatasi +3 hari |
| Penundaan survei pesanan multi-paket (R3) | Skor lebih mencerminkan pengalaman utuh | Jumlah ulasan yang masuk dapat **berkurang** atau tertunda |
| Evaluasi produk & listing seller berisiko (R3) | Mengurangi ulasan buruk akibat barang salah, tidak sesuai, atau berkualitas buruk | Membutuhkan **tenaga untuk meninjau listing**, dan seller dapat merasa dihukum; karena itu dilakukan sebagai pembinaan, dimulai dari 14 seller dengan dua masalah sekaligus |

**Urutan pelaksanaan yang disarankan:**
- **Segera:** Rekomendasi 1, karena daftar rute prioritas sudah tersedia di notebook ini dan dapat dipantau lewat dashboard.
- **Uji coba terbatas:** penundaan survei pada Rekomendasi 3, untuk mengukur dampaknya sebelum diterapkan luas.
- **Menjelang periode puncak:** Rekomendasi 2, disiapkan sebelum Black Friday.
- **Jangka menengah:** gudang Rio de Janeiro lalu gudang Nordeste pada Rekomendasi 1, serta pelacakan per paket pada Rekomendasi 3, karena membutuhkan investasi dan perubahan sistem.

## 6.4 Keterbatasan dan Future Work

**Keterbatasan analisis:**
- Seluruh temuan adalah **keterkaitan** pada data historis Jan 2017 – Agu 2018, bukan bukti sebab-akibat.
- Tidak ada data kurir, sehingga hambatan transit tidak dapat dikaitkan ke mitra logistik tertentu.
- Tidak ada data penyebab internal seller (stok, kapasitas, proses packing), sehingga Seller Late Handover Rate menunjukkan **seberapa sering** seller telat, bukan **mengapa**.
- Analisis seller dan rute hanya memakai pesanan 1 seller (±98,7% pesanan dengan barang).
- Jarak adalah jarak garis lurus antar titik tengah kode pos, bukan jarak tempuh sebenarnya.
- Estimasi dampak bersifat skenario dan mengasumsikan pesanan yang tidak lagi telat berperilaku seperti pesanan tepat waktu pada umumnya.
- Label isi ulasan (5.1.8) dibuat oleh Claude dari teks komentar saja. QA manual pada 150 sampel menunjukkan akurasi Pain Points ±94%, Review Stage ±95%, dan Blame Target ±79% (5.1.8a). Label hanya mencakup ±75% ulasan buruk yang berkomentar. Keluhan yang dilabel adalah **persepsi pelanggan**, bukan penyebab yang terbukti.
- Data akhir Agustus 2018 tidak lengkap (3.9), sehingga angka bulan tersebut perlu dibaca hati-hati.

**Future work:**
1. **Uji coba terbatas (pilot)** pada beberapa rute dan seller prioritas, lalu bandingkan Late Delivery Rate dan Low Review Rate sebelum–sesudah dengan kelompok pembanding, untuk mengukur dampak sebenarnya.
2. **Menambahkan data pelacakan pengiriman** (identitas kurir, waktu tiba di hub, status per paket) agar hambatan transit dapat dipecah lebih rinci.
3. **Menelusuri penyebab lonjakan Februari–Maret 2018** dengan data operasional atau kejadian eksternal pada periode tersebut.
4. **Mengukur pengaruh pengiriman multi-paket** dengan data status per paket, untuk menguji dugaan bahwa barang yang tiba terpisah berkaitan dengan ulasan buruk.
5. **Memperjelas definisi Blame Target** (pihak yang disebut vs pihak yang salah) dan memperbesar sampel QA, lalu memperluas pelabelan ke ulasan bintang 3.
6. **Mendalami rute prioritas per kota asal dan tujuan.** Nama kota sudah bersih (3.7d), sehingga rute prioritas dapat dipecah per kota untuk mencari titik masalah yang lebih spesifik, dengan tetap membatasi pada pasangan kota yang pesanannya cukup (5.2.4).
